# Why NumPy &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.

```python
import math
from datetime import datetime, timedelta

import numpy as np

STATIONS = {"Bergen": 8.0, "Oslo": 6.5, "Svalbard": -4.5, "Tromso": 3.5}   # each station's mean for the year
START = datetime(2025, 1, 1)
HOURS = 365 * 24


def year_of_readings(station):
    """A reading for every hour of 2025 at one station, as the sqlite3 guide makes them.
    Svalbard was silent on 2 March, so those 24 hours are None."""
    i = list(STATIONS).index(station)
    row = []
    for n in range(HOURS):
        hour = START + timedelta(hours=n)
        if station == "Svalbard" and hour.strftime("%Y-%m-%d") == "2025-03-02":
            row.append(None)
        else:
            season = -math.cos(2 * math.pi * (n - 400) / HOURS)
            day = -math.cos(2 * math.pi * (hour.hour - 3) / 24)
            wobble = ((n * 37 + i * 101) % 17 - 8) / 10
            row.append(round(STATIONS[station] + 9 * season + 3 * day + wobble, 1) + 0.0)
    return row


readings_list = [year_of_readings(station) for station in STATIONS]
readings = np.array(readings_list, dtype=float)       # None becomes nan

print("numpy", np.__version__, "|", len(readings_list), "stations,", len(readings_list[0]), "hours each |", readings.shape)
```

**1.** Make an array from the list `[12.5, 13.0, 11.75]`, and print it, its shape and its dtype.

```python
small = np.array([12.5, 13.0, 11.75])

print(small)
print(small.shape)
print(small.dtype)
```

Three decimals make a one-dimensional array of shape (3,), and decimals give `float64`.

**2.** Convert Oslo's readings, row 1 of `readings`, to kelvin by adding 273.15, and print the first five.

```python
kelvin = readings[1] + 273.15

print(kelvin[:5])
```

Adding a number to an array adds it to every value. `readings[1]` is Oslo's whole row.

**3.** Find Tromso's warmest hour of the year, row 3, and print the reading with its date and time.

```python
where = np.nanargmax(readings[3])

print(readings[3, where], START + timedelta(hours=int(where)))
```

`np.nanargmax` is `np.nanargmin`'s partner. Tromso has no missing hours, so `readings[3].argmax()` gives the same position.

**4.** Print every station's lowest reading of the year, leaving Svalbard's missing hours out.

```python
for name, row in zip(STATIONS, readings):
    print(name, np.nanmin(row))
```

`np.nanmin` skips `nan`. `row.min()` would print `nan` for Svalbard.

**5.** Make an array of the difference between Bergen and Oslo at every hour, and print its mean rounded to two places.

```python
difference = readings[0] - readings[1]

print(difference.shape)
print(round(difference.mean(), 2))
```

Subtracting one row from another subtracts hour from hour, and gives 8,760 differences. Their mean is the gap between the two stations' yearly means.
