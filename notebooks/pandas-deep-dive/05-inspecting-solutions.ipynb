```python
import math
from datetime import datetime, timedelta

import pandas as pd

STATIONS = {"Bergen": 8.0, "Oslo": 6.5, "Svalbard": -4.5, "Tromso": 3.5}   # each station's mean for the year
LATITUDES = {"Bergen": 60.39, "Oslo": 59.91, "Svalbard": 78.22, "Tromso": 69.65, "Kirkenes": 69.73}


def year_of_readings():
    """Every hour of 2025 at the four stations, as the sqlite3 guide makes them, with Svalbard silent on 2 March."""
    for n in range(365 * 24):
        hour = datetime(2025, 1, 1) + timedelta(hours=n)
        season = -math.cos(2 * math.pi * (n - 400) / (365 * 24))
        day = -math.cos(2 * math.pi * (hour.hour - 3) / 24)
        for i, (station, mean) in enumerate(STATIONS.items()):
            if station == "Svalbard" and hour.strftime("%Y-%m-%d") == "2025-03-02":
                celsius = None
            else:
                wobble = ((n * 37 + i * 101) % 17 - 8) / 10
                celsius = round(mean + 9 * season + 3 * day + wobble, 1) + 0.0
            yield station, hour.strftime("%Y-%m-%dT%H:%M"), celsius


readings = pd.DataFrame(list(year_of_readings()), columns=["station", "time", "celsius"])
readings["time"] = pd.to_datetime(readings["time"], format="%Y-%m-%dT%H:%M")
stations = pd.DataFrame({"station": list(LATITUDES), "latitude": list(LATITUDES.values())})

print("pandas", pd.__version__, "|", len(readings), "readings,", len(stations), "stations")
```

**1.** Print the shape of `stations` and its column types.

```python
print(stations.shape)
stations.dtypes
```

Five stations, Kirkenes included, and two columns: the names as `object` and the latitudes as `float64`.

**2.** Show the last five rows of `readings` with `tail`, and say which day and hours they are.

```python
readings.tail()
```

The last five rows are Tromso at 22:00 and all four stations at 23:00 on 31 December, because each hour has a row for every station.

**3.** Use `describe` on Oslo's readings alone, rounded to two places.

```python
oslo = readings[readings["station"] == "Oslo"]

oslo["celsius"].describe().round(2)
```

Oslo has all 8,760 readings, a mean of 6.5 and a range from -6.3 to 19.3, the same numbers as Oslo's row in the summary for each station in the **Inspecting** notebook.

**4.** Find the reading that occurs most often at Svalbard, with `value_counts`.

```python
svalbard = readings[readings["station"] == "Svalbard"]

svalbard["celsius"].value_counts().head(3)
```

-10.6 degrees occurs 61 times, more than any other Svalbard reading. `value_counts` sorts from the most common value down, so the first row is the answer.

**5.** Use `describe` on `stations`, and find the station whose latitude is the `max` it reports.

```python
print(stations.describe())

stations[stations["latitude"] == stations["latitude"].max()]
```

`describe` reports a largest latitude of 78.22, and the station with it is Svalbard, the northernmost.
