# Functions


## What you will be able to do

Write a **function**: a named piece of code you can run again with different values. You will be
able to give it inputs, get an answer back, provide sensible defaults, and avoid the one kind of default
that silently produces wrong results.


## The idea

### The problem

Everything written so far runs once, where it sits. When the same work is needed again with
different numbers, the only tool available has been to copy the cell and edit it.

That fails in three ways, and they get worse together.

**The copies drift.** Fix a mistake in one and the others keep it. Nobody can tell which version
is current.

**The intent disappears.** Six lines computing a tax-inclusive total look like six lines of
arithmetic. There is nothing that says *this is how we price things*.

**It cannot be handed anywhere.** Code that lives in the middle of a cell cannot be passed to
`sorted` as a sorting rule, or called from twenty places, or tested on its own.

### What a function is

> A **function** is a named block of code that takes **arguments**, does something with them, and
> hands back a value with `return`.
>
> ```
> def name(parameter, other_parameter):
>     body
>     return result
> ```
>
> Defining a function does not run it. It runs when you **call** it, by writing its name followed
> by parentheses.

The words are worth keeping straight, because error messages use them. A **parameter** is the
name in the definition. An **argument** is the value you pass in when calling.

### What `return` does

`return` sends a value back and **ends the function immediately**. Nothing after it runs.

A function with no `return` still returns something: `None`. That is not an error, and it is the
cause of a common surprise, because printing a value and returning one are different acts. A
function that prints has shown you something. A function that returns has given you something
you can keep.

### Where you will meet this

Everywhere from here on. Every library you use is functions. The `sorted(key=...)` argument the
**Lists** notebook used takes a function. The **Testing and Packaging** guide is about testing
functions and shipping them. The **Object-Oriented Python** guide builds on them.

They also settle a debt. The **Conditionals** notebook flattened nested `if` statements into
guard clauses and said the shape gets better still once you can `return` early. This is where
that happens.

### What this notebook covers

- `def`, calling, and the difference between a parameter and an argument
- `return`, and what a function gives back when you forget it
- Returning several values at once
- Default values, so common cases need no arguments
- Calling by keyword, and why it makes a call readable
- Guard clauses with early `return`
- Docstrings, which is how `help()` learns what your function does
- `*args` and `**kwargs`, for functions that take any number of arguments

- Four errors, including a default value that silently corrupts every later call

## Worked examples

### return against print

These two functions look similar and are not.


````python
def shout(name):
    print(f"HI {name}")

def greet(name):
    return f"Hello, {name}"

a = shout("Ada")
b = greet("Ada")

print("shout gave back:", a)
print("greet gave back:", b)
````

`shout` displayed something and handed back `None`. `greet` handed back a string, which is now
in `b` and can be joined, stored, or passed on.

Print when a person needs to see it. Return when the program needs to use it. A function that
only prints gives the caller nothing to work with.


### return ends the function

Nothing after a `return` runs.


````python
def check(n):
    if n < 0:
        return "negative"
    return "zero or more"

print(check(-5))
print(check(5))
````

The first `return` fires for `-5` and the second line is never reached. That is not a special
case; it is what `return` always does, and it is what makes the next section work.

### Guard clauses with an early return

The **Conditionals** notebook flattened this shape as far as `elif` allowed. With `return`, the
awkward cases can leave immediately.


````python
def status(user):
    if user is None:
        return "no user"
    if not user.get("id"):
        return "no id"
    if not user.get("active"):
        return "inactive"
    return "ready"

for u in [None, {}, {"id": 1}, {"id": 1, "active": True}]:
    print(f"{str(u):<26} {status(u)}")
````

No `else` anywhere, and no nesting. Each line says: if this is wrong, stop here and say why. The
normal answer is the last line, at the same indentation as everything else.

Compare that with the four-level nested version in the **Conditionals** notebook, which gives
identical answers. This is the shape to aim for.

### Several inputs, and several outputs


````python
def rectangle_area(width, height):
    return width * height

print(rectangle_area(3, 4))
````

To hand back more than one value, return them separated by commas. That builds a tuple, which
the **Tuples and Unpacking** notebook covered, and it unpacks on the way out.


````python
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

low, high, mean = stats([88, 72, 95, 64])

print(low, high, mean)
print(stats([88, 72, 95, 64]))     # the raw return is a tuple
````

### Default values

Give a parameter a default and callers can leave it out.


````python
def price(amount, tax=0.08):
    return round(amount * (1 + tax), 2)

print(price(100))          # uses the default
print(price(100, 0.05))    # overrides it
````

Parameters with defaults must come after those without, because Python matches unnamed arguments
by position and would not know where the required ones ended otherwise.

### Calling by name

Any argument can be passed by naming its parameter, which lets you skip over defaults and makes
a call readable on its own.


````python
def price(amount, tax=0.08, discount=0.0):
    return round(amount * (1 + tax) * (1 - discount), 2)

print(price(100, discount=0.1))        # skip past tax
print(price(amount=100, tax=0.05))     # order stops mattering
````

Compare `price(100, 0.08, 0.1)` with `price(100, discount=0.1)`. The second says what the number
means. For anything that is not obvious from the value, name it.

### Docstrings

A string on the first line of a function is its documentation, and `help()` finds it.


````python
def price(amount, tax=0.08):
    """Return amount with tax added, rounded to two decimal places."""
    return round(amount * (1 + tax), 2)

help(price)
````

This is the same `help()` from the **Running Python** notebook, now working on something you
wrote. One sentence saying what the function returns is enough.


### Any number of arguments

`*args` collects extra positional arguments into a tuple.


````python
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))
print(total(10, 20))
print(total())
````

`**kwargs` does the same for named arguments, collecting them into a dictionary.


````python
def describe(**fields):
    for key, value in fields.items():
        print(f"{key}: {value}")

describe(name="Ada", age=36, city="London")
````

The names `args` and `kwargs` are convention, not rules. The `*` and `**` are what matter.

You will read these more often than you write them, most commonly in library functions that
pass arguments through to something else.


### A function is a value

Functions can be stored in names, put in lists, and passed to other functions. This is what the
`key` argument in the **Lists** notebook was doing.


````python
def apply_twice(fn, value):
    return fn(fn(value))

print(apply_twice(str.upper, "ab"))
````

Note `str.upper` with no parentheses. With them you would be **calling** it; without them you are
handing over the function itself.

When the function is small and needed only once, `lambda` writes it inline.


````python
words = ["banana", "fig", "apple"]

print(sorted(words, key=len))
print(sorted(words, key=lambda w: w[-1]))     # sort by last letter
````

`lambda w: w[-1]` is a function taking `w` and returning `w[-1]`. There is no `def` and no name,
and the whole thing has to be a single expression.

Use `lambda` for a one-line rule passed directly into
something else. Anything longer, or needed
twice, should be a real function with a name that says what it does.

## Common errors

Each cell below is run on purpose so you can see the real message.

### TypeError: an argument is missing

Python counts what you passed against what the function needs.


````python
def add(a, b):
    return a + b

print(add(1))
````

`add() missing 1 required positional argument: 'b'` names the function and the parameter, which
is usually enough to spot it immediately. Giving `b` a default would make it optional.


### TypeError: too many arguments, or a name it does not know


````python
def add(a, b):
    return a + b

print(add(1, 2, 3))
````

`takes 2 positional arguments but 3 were given` is the mirror image. Passing a keyword the
function has no parameter for gives a different message naming the keyword:
`got an unexpected keyword argument 'c'`, which is usually a typo in the parameter name.


### SyntaxError: a required parameter after an optional one


````python
def order(quantity=1, item):
    return f"{quantity} x {item}"
````

`parameter without a default follows parameter with a default`. Python matches unnamed arguments
left to right, so it could never tell which value was meant for which parameter. Put the
required ones first.


### The quiet one: a mutable default value

This does not raise. It gives a wrong answer that gets worse every time you call it.


````python
def add_item(item, basket=[]):
    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("pear"))
print(add_item("fig"))
````

Each call was given one item and should have returned a basket holding only that item. Instead
the basket accumulates.

The reason goes back to the **Values and Variables** notebook. **A default value is created once,
when the function is defined**, not each time it is called. That one list is attached to the
function, and because a list is mutable, every call appends to the same one.

The fix is `None` as the default, with the real value built inside:


````python
def add_item(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(add_item("apple"))
print(add_item("pear"))
````

Now a fresh list is made on each call that needs one.

The rule is short and worth following without exception: **never use a list, dictionary or set
as a default value.** Numbers, strings, `True`, `False` and `None` are all immutable and
therefore safe.


## Recap

- `def name(parameters):` defines a function; it runs only when called.
- A **parameter** is the name in the definition, an **argument** is the value passed in.
- `return` hands a value back and ends the function immediately.
- A function with no `return` gives back `None`, so printing and returning are not the same.
- Returning several values makes a tuple, which unpacks into separate names.
- Defaults make arguments optional and must come after the required parameters.
- Naming arguments at the call site makes the call readable and the order irrelevant.
- Early `return` turns nested conditions into flat guard clauses.
- A function is a value: it can be stored, passed, and written inline as a `lambda`.
- **Never use a mutable default.** It is created once and shared by every call.
