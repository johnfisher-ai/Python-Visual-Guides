**1.** Loop over `["red", "green", "blue"]` and print each one in capitals.


```python
for color in ["red", "green", "blue"]:
    print(color.upper())
```

`.upper()` returns a new string, as **Strings** showed. Printing it directly is fine because
nothing needs to be kept.


**2.** Use `range` to print the numbers 1 through 10, each on its own line.


```python
for n in range(1, 11):
    print(n)
```

Two adjustments from `range(10)`. Starting at `1` needs the first argument, and reaching `10`
needs a stop of `11`, because the stop value is never included. That exclusive end is the same
rule as slicing.


**3.** Given `prices = [4.50, 12.00, 7.25, 30.00]`, use a loop to add them up and print the
total, then print `sum(prices)` to check yourself.


```python
prices = [4.50, 12.00, 7.25, 30.00]

total = 0
for price in prices:
    total = total + price

print(total)
print(sum(prices))
```

Starting at `0` matters: the running total has to exist before the first addition.

`total += price` is shorthand for the same line and you will see it constantly.


**4.** Given the same `prices`, build a new list holding only those under 10, and print it.


```python
cheap = []
for price in prices:
    if price < 10:
        cheap.append(price)

print(cheap)
```

Empty list first, then append what qualifies. This is the collect pattern, and building a new
list rather than removing from `prices` is what keeps it correct, for the reason **Lists**
demonstrated.


**5.** Given `names = ["Ada", "Grace", "Alan"]`, print a numbered list starting at 1,
like `1. Ada`.


```python
names = ["Ada", "Grace", "Alan"]

for n, name in enumerate(names, start=1):
    print(f"{n}. {name}")
```

`start=1` shifts the numbering for a human reader without touching the list. Doing it as
`enumerate(names)` and printing `n + 1` works too, and this says it once instead of on every
pass.


**6.** Given `words = ["apple", "fig", "banana", "kiwi"]`, stop at the first word longer
than five letters and print it. If there is none, print `"none found"`.


```python
words = ["apple", "fig", "banana", "kiwi"]

for word in words:
    if len(word) > 5:
        print(word)
        break
else:
    print("none found")
```

`banana` is the first word over five letters, so the loop breaks there and `kiwi` is never
checked. Because a `break` happened, the `else` is skipped.

Take `banana` out of the list and the loop runs to the end with no `break`, so the `else` runs
and prints `none found`. That is the whole purpose of a loop `else`: it saves keeping a separate
`found = False` flag and checking it afterward.


```python
for word in ["apple", "fig", "kiwi"]:
    if len(word) > 5:
        print(word)
        break
else:
    print("none found")
```
