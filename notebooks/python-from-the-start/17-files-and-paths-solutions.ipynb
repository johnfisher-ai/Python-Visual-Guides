# Files and Paths &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
from pathlib import Path
import shutil

work = Path("practice")
work.mkdir(exist_ok=True)
```

**1.** Write `"hello\nworld\n"` into `practice/greeting.txt` and print it back.


```python
greeting = work / "greeting.txt"

greeting.write_text("hello\nworld\n")

print(greeting.read_text())
```

`write_text` replaces whatever was there. Running the cell twice gives the same file rather
than two copies of the text.


**2.** Print the `name`, `stem`, `suffix` and `parent` of the file.


```python
print("name:  ", greeting.name)
print("stem:  ", greeting.stem)
print("suffix:", greeting.suffix)
print("parent:", greeting.parent)
```

`suffix` keeps the dot, which is worth knowing when comparing it: the test is
`suffix == ".txt"`, not `"txt"`.


**3.** Open the file with `with` and print each line numbered, without the trailing
newline.


```python
with open(greeting) as f:
    for number, line in enumerate(f, start=1):
        print(number, line.rstrip())
```

`enumerate` with `start=1` numbers for a reader, as the **Loops** notebook showed, and
`.rstrip()` removes the newline that every line carries.


**4.** Append a third line, then print the whole file.


```python
with open(greeting, "a") as f:
    f.write("again\n")

print(greeting.read_text())
```

`"a"` keeps what is there. Using `"w"` would have replaced both existing lines, with no
warning.

The `\n` at the end is needed. Without it the next append would join onto the same line.


**5.** Write two more files, one `.txt` and one `.csv`, then print only the `.txt` filenames
using `glob`.


```python
(work / "notes.txt").write_text("some notes\n")
(work / "data.csv").write_text("a,b\n1,2\n")

print(sorted(p.name for p in work.glob("*.txt")))
print(sorted(p.name for p in work.glob("*")))
```

`glob("*.txt")` matched two files and skipped the CSV. Sorting makes the output predictable,
since the order files come back in is not guaranteed.


**6.** Try to read `practice/nothing.txt` and print `"not found"` instead of raising, then
delete the folder.


```python
missing = work / "nothing.txt"

try:
    print(missing.read_text())
except FileNotFoundError:
    print("not found")

shutil.rmtree(work)
print("practice still there:", work.exists())
```

`missing.exists()` would answer the same question without the `try`. Either is right here; the
`try` version is the one to reach for when reading the file is the normal case and its absence
is the exception.

`rmtree` removes the folder and everything in it, which is why the cleanup comes last.
