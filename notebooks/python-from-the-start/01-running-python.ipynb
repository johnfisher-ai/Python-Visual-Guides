## The idea

A notebook is a stack of **cells**. A cell holds either text, like this one, or code. You run a
code cell and its result appears directly underneath it.

That is the whole idea, and it is why notebooks suit learning: you can change one line, run it
again, and see immediately what changed.


```python
print("Hello from Python")
```

Two things happened. Python read the line, and it printed the text between the quotes. The
output appeared below the cell, not in some other window.

`print` is a **function**. You give it something in the brackets and it shows it to you.


## Worked examples

### Printing more than one thing

`print` takes several values, separated by commas, and puts a space between them.


```python
print("Python", "is", "running")
print("2 + 2 is", 2 + 2)
```

Notice `2 + 2` was worked out before it was printed. Python evaluates what you give it, then
prints the answer, not the question.

### The last line shows itself

In a notebook you do not always need `print`. If the last line of a cell produces a value, the
notebook displays it for you.


```python
7 * 6
```

That is a notebook convenience, not a Python rule. Only the **last** line behaves this way, so
if you want to see two things you still need `print`.


```python
3 + 3
7 * 6
```

Only `42` appeared. The `6` was calculated and thrown away, because nothing asked to see it.

### Cells remember

This is the part that surprises people. Every cell in a notebook shares the same memory. A name
you create in one cell is still there in the next.


```python
message = "stored in memory"
```

That cell printed nothing, because assigning a value is not the same as displaying one. But the
name now exists, and the next cell can use it.


```python
print(message)
```

### Notes to yourself

Anything after a `#` on a line is a **comment**. Python ignores it completely. Comments are for
the human reading the code, which is usually you, later.


```python
# This line does nothing at all.
print("This line does something")  # and this bit is ignored too
```

### Asking Python what something does

You do not have to remember everything, and nobody does. Python will tell you.

`help()` prints the documentation for whatever you give it.


```python
help(print)
```

In Colab and Jupyter there is a shorter way: put a `?` after a name and run the cell. The
documentation opens in a panel at the bottom of the screen rather than in the output.

Try it: make a new cell, type `print?`, and run it.

Getting comfortable with `help()` early is worth more than memorising anything on this page.


### Other ways to run Python

You are using a notebook, which is the right tool for learning and for exploring data. It is
not the only one, and it is worth knowing what else exists so the words are not a surprise
later.

- **A script**: Python code saved in a `.py` file and run all at once, start to finish. This is
  how finished programs are shipped. Notebook 18 covers writing your own.
- **The interpreter**: type `python` in a terminal and you get a prompt that runs one line at a
  time. Useful for a quick check.
- **A notebook**: what you are in. Cells, in any order, with the output kept beside the code.

The Python is identical in all three. Only the way you feed it code changes.

## Common errors

Every notebook in this guide ends with the errors that part of Python actually produces. They
are not hypothetical: each one below is run on purpose so you can see the real message.

Learning to read these is most of learning to program.

### NameError: you used something that does not exist yet

This is the error you will hit most often in a notebook, and it is almost never a typo. It
usually means you **ran the cells out of order**, or skipped one.


```python
print(total_apples)
```

Read the last line first. `NameError: name 'total_apples' is not defined` means Python has no
idea what `total_apples` is, because no cell that created it has been run.

The fix is almost always to run the earlier cell you skipped. In Colab, **Runtime > Run all**
runs everything from the top, in order, and clears up most of these at once.

### SyntaxError: Python could not read the line

A syntax error means the line is not valid Python at all, so nothing ran. Here a quote is left
open.


```python
print("this string never ends)
```

Syntax errors point at where Python gave up, which is often slightly **after** the real mistake.
The unclosed quote is at the start of the string; the complaint arrives at the end of the line.
When the arrow looks like it is pointing at innocent code, check just before it.

### TypeError: those two things do not go together

Python will not quietly guess what you meant when the types do not fit.


```python
print("I have " + 5 + " apples")
```

`can only concatenate str (not "int") to str` means you tried to join text and a number with
`+`. Python does not assume you wanted the number turned into text.

Two ways to fix it, both fine:


```python
print("I have", 5, "apples")
print("I have " + str(5) + " apples")
```

## Recap

- A notebook is cells, and running a code cell shows its result underneath.
- **Shift + Enter** runs a cell.
- `print` displays things; a bare last line displays itself, but only the last one.
- Cells share memory, so **the order you run them in matters**.
- `#` starts a comment, which Python ignores.
- `NameError` usually means a cell was skipped, not that you mistyped.
- Syntax errors point at where Python gave up, which can be after the actual mistake.
- `help(print)` or `print?` tells you what something does, and you will use it constantly.
