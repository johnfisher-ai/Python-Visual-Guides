# Running Python


You are looking at a notebook. It is a document you can run.


## What you will be able to do

why the order you run things in matters more here than in any other kind of program.


## The idea

### The problem

Most programs are written in one place and run in another. You type a whole file of
instructions, save it, hand it to the computer, and it runs from the first line to the last.
You find out at the end whether any of it did what you wanted.

For finished software that is exactly right. For **learning**, and for **looking at data**, it
is painful. You want to try one small thing, see what happened, and adjust. You want to keep
the result of a slow step and poke at it, not recompute it every time. You want the code and
its output side by side so you can see whether the answer is sensible.

### What a notebook is

> A **notebook** is a document made of **cells**. Each cell holds either text or code. When you
> run a code cell, Python executes it and its result is stored in the document, directly
> underneath the code that produced it.

Three things follow from that definition, and they are the whole reason notebooks exist.

**The output lives in the document.** It is not printed to a terminal and lost. It is saved
alongside the code, so anyone opening the file later sees what happened without running
anything.

**Cells share one memory.** A name created in the first cell is still available in the
twentieth. The notebook is one continuous session, not twenty separate programs.

**You choose the order.** Cells are laid out top to bottom, but nothing forces you to run them
that way. You can re-run one cell fifty times while the rest sit untouched.

That third point is the one to hold onto. It is what makes a notebook good for exploring, and
it is the source of the single most common confusion beginners hit, which this notebook covers
directly rather than leaving you to discover it.

### Where you will meet this

Notebooks are standard equipment anywhere the work is exploratory rather than finished. Data
analysts use them to look at a dataset one column at a time. Researchers publish them so a
reader can rerun the analysis and check it. Teaching material like this guide is written in
them because you can read a paragraph and immediately run the thing it describes.

You are reading one now. Every guide in this collection is a notebook.

## Worked examples

### Printing more than one thing

`print` accepts several arguments, separated by commas, and puts a space between them.


````python
print("Python", "is", "running")
print("2 + 2 is", 2 + 2)
````

Notice `2 + 2` was worked out before it was printed. Python evaluates what you give it, then
prints the answer, not the question.

### The last line shows itself

In a notebook you do not always need `print`. If the last line of a cell produces a value, the
notebook displays it for you.


````python
7 * 6
````

That is a notebook convenience, not a Python rule. Only the **last** line behaves this way, so
if you want to see two things you still need `print`.


````python
3 + 3
7 * 6
````

Only `42` appeared. The `6` was calculated and thrown away, because nothing asked to see it.

### Cells remember

This is the part that surprises people. Every cell in a notebook shares the same memory. A name
you create in one cell is still there in the next.


````python
message = "stored in memory"
````

That cell printed nothing, because assigning a value is not the same as displaying one. But the
name now exists, and the next cell can use it.


````python
print(message)
````

### Notes to yourself

Anything after a `#` on a line is a **comment**. Python ignores it completely. Comments are for
the human reading the code, which is usually you, later.


````python
# This line does nothing at all.
print("This line does something")  # and this bit is ignored too
````

### Asking Python what something does

You do not have to remember everything, and nobody does. Python will tell you.

`help()` prints the documentation for whatever you give it.


````python
help(print)
````

In Colab and Jupyter there is a shorter way: put a `?` after a name and run the cell. The
documentation opens in a panel at the bottom of the screen rather than in the output.

Try it: make a new cell, type `print?`, and run it.

Getting comfortable with `help()` early is worth more than memorizing anything on this page.


### Other ways to run Python

You are using a notebook, which is the right tool for learning and for exploring data. It is
not the only one, and it is worth knowing what else exists so the words are not a surprise
later.

- **A script**: Python code saved in a `.py` file and run all at once, start to finish. This is
  how finished programs are shipped. The **Modules and Imports** notebook covers writing your own.
- **The interpreter**: type `python3` in a terminal and you get a prompt that runs one line at
  a time. Useful for a quick check.
- **A notebook**: what you are in. Cells, in any order, with the output kept beside the code.

The Python is identical in all three. Only the way you feed it code changes.


#### Why `python3` and not `python`

The extra 3 is a leftover from a long changeover.

Python 3 arrived in 2008 and was deliberately not backward compatible with Python 2, so for
years both were installed side by side and `python` was ambiguous. `python3` was unambiguous.
Python 2 reached end of life on 1 January 2020 and is gone, but the name stuck.

On macOS and Linux, type `python3`. Plain `python` often does not exist at all: recent macOS
ships no Python 2, so the command is simply missing. On Windows the installer usually provides
both `python` and a `py` launcher.

None of this affects you inside Colab, which is already running Python 3. It matters the moment
you search the web for help, because plenty of answers online are still written in Python 2 and
will not run. Common errors, below, shows the giveaway.

## Common errors

Every notebook in this guide ends with the errors that part of Python actually produces. They
are not hypothetical: each one below is run on purpose so you can see the real message.

Learning to read these is most of learning to program.

### NameError: you used something that does not exist yet

This is the error you will hit most often in a notebook, and it is almost never a typo. It
usually means you **ran the cells out of order**, or skipped one.


````python
print(total_apples)
````

Read the last line first. `NameError: name 'total_apples' is not defined` means Python has no
idea what `total_apples` is, because no cell that created it has been run.

The fix is almost always to run the earlier cell you skipped. In Colab, **Runtime > Run all**
runs everything from the top, in order, and clears up most of these at once.

### SyntaxError: Python could not read the line

A syntax error means the line is not valid Python at all, so nothing ran. Here a quote is left
open.


````python
print("this string never ends)
````

Syntax errors point at where Python gave up, which is often slightly **after** the real mistake.
The unclosed quote is at the start of the string; the complaint arrives at the end of the line.
When the arrow looks like it is pointing at innocent code, check just before it.

### TypeError: those two things do not go together

Python will not quietly guess what you meant when the types do not fit.


````python
print("I have " + 5 + " apples")
````

`can only concatenate str (not "int") to str` means you tried to join text and a number with
`+`. Python does not assume you wanted the number turned into text.

Two ways to fix it, both fine:


````python
print("I have", 5, "apples")
print("I have " + str(5) + " apples")
````

### SyntaxError: you copied Python 2 code

This is the one that catches people searching for help. In Python 2, `print` was not a function
and needed no parentheses. Any answer written that way fails immediately.


````python
print "hello"
````

Missing parentheses in call to 'print' is a good error message: Python worked out what you
meant and told you. Add the parentheses and it runs.

When code from a web page fails like this, you are reading a Python 2 answer. Look for a newer
one, because the `print` difference is rarely the only thing that has changed.


### IndentationError: the lines do not line up

Python cares about space at the **start** of a line, which is unusual and catches everyone at
least once. It usually happens when pasting part of a larger program from a web page.


````python
print("this line starts at the left")
  print("this one does not")
````

`unexpected indent` means Python found a line pushed to the right for no reason. Space at the
start of a line is how Python marks a block belonging to something above it, which you meet in
the **Conditionals** notebook. Here there is nothing to belong to.

One thing worth knowing, because it looks inconsistent: a cell where **every** line is indented
by the same amount runs perfectly well, since the notebook quietly removes the shared
indentation. It is lines that disagree with each other that fail.

Space *inside* a line is free. `x=1+2` and `x = 1 + 2` are identical to Python, and the second
is easier to read.
