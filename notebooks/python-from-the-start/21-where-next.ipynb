## Worked examples

### A program you can now read

This is the kind of thing a first real task looks like: lines of text from somewhere, some of
them broken, and a summary wanted at the end.

Read it before running it, and name what each part is doing.


````python
rows = [
    "2026-03-01,north,18.5",
    "2026-03-01,south,22.1",
    "2026-03-02,north,",
    "2026-03-02,south,23.4",
    "bad line with no commas",
    "2026-03-03,north,19.2",
]


def parse(line):
    """Turn one line into a record, or None if it cannot be read."""
    parts = line.split(",")

    if len(parts) != 3:
        return None

    date, region, value = parts

    if not value.strip():
        return None

    return {"date": date, "region": region, "value": float(value)}


records = [r for r in (parse(line) for line in rows) if r is not None]

print(f"kept {len(records)} of {len(rows)} lines")
````

Every piece of that has a notebook behind it:

| In the code | From |
|---|---|
| `rows = [...]` and indexing into it | **Lists** |
| `def parse(line):` and `return` | **Functions** |
| The docstring on `parse` | **Modules and Imports** |
| `line.split(",")` and `value.strip()` | **Strings** |
| `if len(parts) != 3:` with an early `return` | **Conditionals** |
| `date, region, value = parts` | **Tuples and Unpacking** |
| `if not value.strip():` as a truth test | **Booleans and Comparison** |
| `float(value)` | **Numbers** |
| `{"date": ..., "region": ...}` | **Dictionaries** |
| `[r for r in (...) if r is not None]` | **Comprehensions** |
| `f"kept {len(records)} of {len(rows)} lines"` | **Strings** |

Two guard clauses do the work of validation, and returning `None` for a bad line means the
caller decides what to do about it rather than the parser crashing.

### Summarizing it


````python
by_region = {}

for r in records:
    by_region.setdefault(r["region"], []).append(r["value"])

for region, values in sorted(by_region.items()):
    mean = sum(values) / len(values)
    print(f"{region:<6} n={len(values)}  mean={mean:.2f}")
````

`setdefault` grouping from the **Dictionaries** notebook, `.items()` unpacked in the loop
header, `sorted` for a predictable order, and an f-string with alignment and two decimal places.

Note the denominator again: `len(values)`, the readings actually used, not `len(rows)`. That
distinction was the subject of the last section of the **Debugging** notebook and it does not
stop mattering.

If you followed all of that, you can read most Python you will meet. The rest is libraries.

## Common errors

These are not new. They are the mistakes from this guide that keep catching people long after
they know better, gathered in one place so you can recognize the shape rather than the specific
example.

### The five that will still catch you


````python
# 1. A mutable default is created once, not per call
def add(item, basket=[]):
    basket.append(item)
    return basket

print("mutable default: ", add("a"), add("b"))
````

````python
# 2. Assignment gives a second name, not a copy
a = [1, 2]
b = a
b.append(3)

print("aliasing:        ", a)
````

````python
# 3. Floats are approximations
print("float equality:  ", 0.1 + 0.2 == 0.3)
````

````python
# 4. == compares values, is compares identity
x = [1, 2, 3]
y = [1, 2, 3]

print("== vs is:        ", x == y, x is y)
````

````python
# 5. Never change a collection while looping over it
nums = [2, 4, 6]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)

print("mutate in a loop:", nums, "  (every item was even; one survived)")
````

Not one of those raises. All five produce a plausible wrong answer, which is what makes them
worth memorizing rather than looking up.

They also have one thing in common, and it is the single most useful idea in this guide: **a
name is a label on a value, not a box containing one.** Numbers 1, 2, 4 and 5 are all that fact
in different clothes. If you carry one sentence out of these twenty notebooks, carry that one.

### The sixth, which is about you rather than Python

The most expensive mistake is not on that list. It is deciding what the code does by reading it
instead of by running it.

Every quiet bug in this guide looked correct. The version that returned `17.0` instead of
`21.25` looked correct. The comprehension that dropped a row looked correct. The only reliable
way to know what code does is to run it and look, which is exactly what a notebook is for.

That last one is worth being concrete about. Whenever a notebook states a rule, write a cell
that does the opposite and read what Python says:

- A tuple cannot be changed. Try assigning to one and read the `TypeError`.
- A default argument is created once, not per call. Call such a function three times and print
  the result each time.
- Two names can point at one list. Change it through one name and print the other.

Each of those is one cell and takes a few seconds, and the error message or the surprising
output teaches the rule better than the sentence stating it does. The exercises in these guides
touch nothing outside the notebook, so the worst outcome is an error, which is the thing you
were trying to see.
