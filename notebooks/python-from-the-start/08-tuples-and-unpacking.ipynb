# Tuples and Unpacking


## What you will be able to do

Use tuples where a list would be wrong, and pull several values out of one in a single line.
You will also be able to explain the swap from the **Values and Variables** notebook, which has been waiting for this
notebook.


## The idea

### The problem

A list is the right shape for many things of the same kind: all the temperatures, all the rows,
all the responses. It is the wrong shape for **one thing that has parts**.

A point on a map is not a collection of two numbers you might add a third to later. It is a
single location that will always have exactly two parts, in a fixed order, and swapping them
gives a different place. A date is three numbers that mean year, month and day, always in that
arrangement. A color written as red, green and blue is one color, not a growing pile of
numbers.

Store any of those in a list and nothing prevents code from appending a fourth number to a
coordinate, sorting a date into ascending order, or removing the green. Nothing prevents it,
and eventually something does it.

### What a tuple is

> A **tuple** is an ordered collection that **cannot be changed** after it is created. Reading
> works exactly as it does for a list: positions, slices, `in`, `len`. What is missing is every
> operation that would alter it.
>
> A tuple is written with commas. Parentheses are usually present and are usually optional; it
> is the **comma** that creates the tuple.

### Why a restriction is a feature

Removing an ability sounds like a downside. Three things come back in exchange.

**It can be a dictionary key.** the **Dictionaries** notebook stores values under keys, and a key has to be
something that cannot change, or the entry would become unfindable. A tuple qualifies. A list
does not. Coordinate keys, date keys and composite keys all depend on this.

**It is safe to hand over.** the **Lists** notebook showed that passing a list to other code means that
code can alter yours. A tuple cannot be altered, so there is nothing to defend against.

**It signals intent.** A reader who sees a tuple knows the shape is fixed. That is information
the code carries by itself, with no comment required.

### Unpacking, which is why they are everywhere

The second half of this notebook is a technique rather than a type. **Unpacking** assigns
several names from one collection in a single line, and it is used constantly.

It also settles something the **Values and Variables** notebook left unfinished. There,
`a, b = b, a` swapped two values with no temporary variable, and the explanation was deferred. The reason is
that the right side builds a tuple before anything is assigned. This notebook shows that
happening.

### Where you will meet this

Every function that returns more than one value returns a tuple. `divmod` does. `enumerate` and
`zip`, which the **Loops** notebook uses in almost every loop, hand you tuples. `re.search` groups came
back as one in the **Regular Expressions** notebook. Dictionary `.items()` in the **Dictionaries** notebook gives you a stream of them.

## Worked examples

### The comma makes the tuple, not the parentheses

This surprises people, and it causes a real bug.


````python
print(type((5)))      # just a number in brackets
print(type((5,)))     # a tuple with one item
print(type(1, ))      # still just a number
````

`(5)` is `5` with brackets around it, exactly as in arithmetic. A one-item tuple needs the
trailing comma: `(5,)`.

The parentheses are often optional altogether.


````python
pair = 1, 2
print(pair, type(pair))

empty = ()
print(empty, len(empty))
````

Write the parentheses anyway in most places. They cost nothing and they make the intent
obvious.

### What a tuple cannot do

A list has `append`, `insert`, `remove`, `sort`. A tuple has none of them, because all of them
change the thing.


````python
point = (3, 7, 11)

print(point.count(7))
print(point.index(11))
print(7 in point)
````

`count`, `index` and `in` are the whole set. They ask questions; they do not change anything.
Trying to change a tuple raises, which Common errors shows.

### Unpacking: the reason tuples are everywhere

Put several names on the left and Python hands out the parts.


````python
point = (3, 7, 11)

x, y, z = point

print(x, y, z)
````

The count has to match, or you get an error naming the mismatch. That strictness is useful:
it fails immediately rather than silently giving you the wrong piece.

This is what makes the swap from the **Values and Variables** notebook work.


````python
a, b = 1, 2
print("before:", a, b)

print("the right side builds:", (b, a))

a, b = b, a
print("after: ", a, b)
````

`b, a` builds a tuple from the current values, **then** the two names are assigned from it.
Because the tuple is finished before any name moves, nothing is overwritten halfway and no
temporary variable is needed.

### Unpacking in a loop

This is the form you will write most often.


````python
scores = [("Ada", 91), ("Grace", 88), ("Alan", 95)]

for name, score in scores:
    print(f"{name:<6} {score}")
````

Each item is a two-item tuple, and `for name, score in ...` unpacks it on the way in. Compare
that with `pair[0]` and `pair[1]` everywhere, which is what you would otherwise write.

Two built-ins you will meet in the **Loops** notebook hand you tuples for exactly this reason.


````python
print(list(enumerate(["a", "b", "c"])))
print(list(zip([1, 2, 3], ["a", "b", "c"])))
````

`enumerate` pairs each item with its position, `zip` pairs up two collections. Both are lists
of tuples, and both are normally used with unpacking.


````python
for i, letter in enumerate(["a", "b", "c"]):
    print(i, letter)
````

### Taking some and keeping the rest

A `*` in front of one name collects everything left over, as a list.


````python
first, *rest = [10, 20, 30, 40]
print(first, rest)

*start, last = [10, 20, 30, 40]
print(start, last)
````

Only one name can carry the `*`, because otherwise there would be no way to decide where the
split falls.

When you want a piece and do not care about the others, `_` is the conventional name for
"ignore this".


````python
date = (2026, 9, 4)

_, month, _ = date

print("month:", month)
````

`_` is an ordinary name with no special power. The convention is a message to whoever reads the
code next.

### Returning more than one thing

A function can only return one value, and a tuple is how you return several anyway. `divmod`
is a built-in that does it.


````python
result = divmod(17, 5)

print(result, type(result))

whole, remainder = divmod(17, 5)
print(whole, "remainder", remainder)
````

`17 = 5 * 3 + 2`, so `divmod` gives back both halves of the division from the **Numbers** notebook in one
call. Your own functions do the same thing in the **Functions** notebook.


### Why the restriction is worth having

The **Lists** notebook showed the aliasing problem: two names on one list, and a change through either is
visible through both. A tuple cannot be changed, so that problem does not exist.


````python
original = (1, 2)
alias = original

print(original is alias)      # one tuple, two names, exactly as with a list
print(original)               # and nothing can change it, so it does not matter
````

That makes a tuple safe to hand to code you did not write. There is no way for it to come back
altered.

The second reason is more concrete. A dictionary key has to be unchangeable, so a tuple can be
one and a list cannot. Dictionaries come next; this is the part that needs a tuple.


````python
locations = {
    (0, 0): "origin",
    (3, 7): "the point we care about",
}

print(locations[(3, 7)])
````

A coordinate pair is a natural key, and it works only because tuples are immutable. Common
errors below shows what happens if you try a list instead.

### Choosing between a tuple and a list

| Use a tuple | Use a list |
|---|---|
| The parts mean different things, like `(x, y)` | The items are all the same kind of thing |
| The size is fixed and known | Items get added or removed |
| You want it usable as a dictionary key | You want to sort or append |
| Returning several values at once | Collecting results as you go |

When neither fits obviously, use a list. It is the more flexible of the two.

Converting is easy in both directions.


````python
as_list = list((1, 2, 3))
as_tuple = tuple([1, 2, 3])

print(as_list, type(as_list))
print(as_tuple, type(as_tuple))
````

## Common errors

Each cell below is run on purpose so you can see the real message.

### TypeError: a tuple cannot be changed

The same error a string gave in the **Strings** notebook, for the same reason.


````python
point = (3, 7, 11)
point[0] = 99
````

`'tuple' object does not support item assignment` is immutability doing its job. If you need to
change it, you wanted a list.

To produce a changed version, build a new tuple:


````python
point = (3, 7, 11)
point = (99,) + point[1:]

print(point)
````

### AttributeError: tuples have no append

Not a restriction on when you can call it. The method does not exist.


````python
point = (3, 7)
point.append(11)
````

`'tuple' object has no attribute 'append'` is Python saying the operation is not part of what a
tuple is. Convert to a list if you need to build something up, and convert back at the end if
you want it fixed again.


### ValueError: the counts do not match

Unpacking is strict about how many names you provide.


````python
a, b = (1, 2, 3)
````

`too many values to unpack (expected 2, got 3)` names both numbers, which usually tells you immediately
whether the data or the line is wrong.

Too few gives the matching complaint, `not enough values to unpack`. When you genuinely want
only some of them, use `*rest` or `_`.


````python
a, b, *_ = (1, 2, 3)

print(a, b)
````

### TypeError: a list cannot be a dictionary key

This is the practical consequence of everything above.


````python
locations = {[0, 0]: "origin"}
````

`unhashable type: 'list'` means the key has to be something that cannot change. A dictionary
finds things by key, and a key that could change underneath it would make the entry
unfindable.

A tuple is the fix, and it is why coordinate keys are always written as tuples.


````python
locations = {(0, 0): "origin"}

print(locations[(0, 0)])
````

## Recap

- A tuple is an ordered collection that cannot be changed after it is made.
- The **comma** makes a tuple, not the brackets, so a single item needs `(5,)`.
- Positions, slicing, `in`, `index` and `count` work as they do for lists.
- Unpacking assigns several names at once, and the counts must match.
- `a, b = b, a` works because the right side becomes a tuple before any name moves.
- `for name, score in pairs` is the everyday form, and what `enumerate` and `zip` are built for.
- `*rest` collects what is left over; `_` is the convention for a value you are ignoring.
- Only immutable things can be dictionary keys, which is why a tuple can be one and a list
  cannot.
