# Strings &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** Put your own city in a name called `home`. Print its length, its first character,
and its last character.


```python
home = "Philadelphia"

print(len(home))
print(home[0])
print(home[-1])
```

`home[-1]` avoids writing `home[len(home) - 1]`, which is the same thing said the long way.


**2.** Using `home`, print the first three characters and the last three characters, each
on its own line.


```python
print(home[:3])
print(home[-3:])
```

Leaving a side of the slice blank means the beginning or the end, so neither line needs
`len`.


**3.** The string `"  Ada Lovelace  "` has spaces on both ends. Clean it up and print it
with `repr` so you can see the quotes and prove the spaces are gone.


```python
padded = "  Ada Lovelace  "

print(repr(padded))
print(repr(padded.strip()))
```

Without `repr` both lines would look almost identical in the output. The quotes are what
make the difference visible.


**4.** With `first = "Ada"` and `last = "Lovelace"`, use an f-string to print
`Ada Lovelace has 11 letters`. Count the letters in code, not by hand, and do not count the
space.


```python
first = "Ada"
last = "Lovelace"

letters = len(first) + len(last)

print(f"{first} {last} has {letters} letters")
```

Adding the two lengths counts only the letters. Measuring `len(first + " " + last)` would give
12, because the space is a character too.


**5.** Split `"red,green,blue"` on the comma, then join the pieces back together separated
by `" and "`.


```python
colors = "red,green,blue"

pieces = colors.split(",")
print(pieces)
print(" and ".join(pieces))
```

The string you call `join` on is the glue. `" and "` goes between the pieces, not around
them.


**6.** Print `"racecar"` reversed, then print `True` or `False` for whether it reads the
same backward as forward. Do it with a slice, in one comparison.


```python
word = "racecar"

print(word[::-1])
print(word == word[::-1])
```

A step of `-1` walks the string backward. Comparing the result with the original is the whole
palindrome test, with no loop.

Try it with `"Racecar"` and it becomes `False`, because `R` and `r` are different characters.
Comparing `word.lower()` instead handles that.
