**1.** Write `double(n)` that returns `n` multiplied by two, and print the result for `21`.


```python
def double(n):
    return n * 2

print(double(21))
```

`return`, not `print`. The value comes back so it can be used; here it is used by the
`print` on the last line.


**2.** Write `initials(first, last)` that returns something like `"A.L."` for `"Ada"` and
`"Lovelace"`.


```python
def initials(first, last):
    return f"{first[0]}.{last[0]}."

print(initials("Ada", "Lovelace"))
```

`first[0]` is indexing from the **Strings** notebook, and the f-string assembles the result.
`.upper()` on each would make it robust against a lowercase name.


**3.** Write `shipping(weight, rate=2.50)` returning weight times rate. Print it once with
the default and once with a rate of `4.00`.


```python
def shipping(weight, rate=2.50):
    return weight * rate

print(shipping(10))
print(shipping(10, 4.00))
print(shipping(10, rate=4.00))    # the same call, said more clearly
```

The third line names the argument. With one extra number the meaning is still guessable; in a
call with four numbers it would not be, which is why naming is worth the habit.


**4.** Write `describe(numbers)` that returns the count, the smallest and the largest, then
unpack the result into three names.


```python
def describe(numbers):
    return len(numbers), min(numbers), max(numbers)

count, lowest, highest = describe([88, 72, 95, 64])

print(count, lowest, highest)
```

Three values separated by commas make a tuple, and three names on the left unpack it. The counts
have to match, as the **Tuples and Unpacking** notebook showed.


**5.** Write `grade(score)` using early returns, with no `else` anywhere.


```python
def grade(score):
    if score >= 90:
        return "A"
    if score >= 80:
        return "B"
    if score >= 70:
        return "C"
    return "F"

for s in [95, 85, 72, 40]:
    print(s, grade(s))
```

No `else` is needed because `return` ends the function. Reaching the second `if` already proves
the score was under 90, so each test only needs its own lower bound.

The order matters for the same reason it did in the **Conditionals** notebook: reversed, every
score of 70 or more would come back `"C"`.


**6.** Sort `["banana", "fig", "apple", "kiwi"]` by the last letter of each word, using a
`lambda`.


```python
words = ["banana", "fig", "apple", "kiwi"]

print(sorted(words, key=lambda w: w[-1]))
```

`lambda w: w[-1]` takes one word and returns its last character, and `sorted` uses that as the
sorting rule rather than the word itself.

The same thing with a named function is longer and does exactly the same job, which is the
choice `lambda` exists to offer:


```python
def last_letter(word):
    return word[-1]

print(sorted(words, key=last_letter))
```
