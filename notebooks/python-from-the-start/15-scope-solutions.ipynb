**1.** Write a function that creates a name inside itself, then try to print that name
after calling it.


```python
def compute():
    working = 42
    return "done"

print(compute())
print(working)
```

`name 'working' is not defined`. The name lived only for the length of the call. Returning it is
the way to get a value out.


```python
def compute():
    working = 42
    return working

print(compute())
```

**2.** Create `rate = 0.1` outside a function, then write a function that reads it and
returns `100 * rate`. No parameters.


```python
rate = 0.1

def commission():
    return 100 * rate

print(commission())
```

Reading an outer name needs no declaration. Python looked locally, found nothing, and
searched outward.


**3.** Rewrite it so that `rate` is a parameter with a default of `0.1`.


```python
def commission(rate=0.1):
    return 100 * rate

print(commission())
print(commission(0.2))

# The parameter version is easier to maintain. It says what it needs, it can be
# tested with any rate, and it does not break when something renames the outer name.
```

The first version silently depends on a name defined somewhere else in the notebook. The second
carries its own default and can be called with anything.


**4.** Write `add_zero(items)` that appends `0` to the list it is given, then show the
original list changed.


```python
def add_zero(items):
    items.append(0)

numbers = [1, 2, 3]
add_zero(numbers)

print(numbers)
```

`items` and `numbers` are two names on one list, so appending through either is visible through
both. The function returns nothing and still changed something, which is worth being deliberate
about.


**5.** Write `clear(items)` that sets `items = []` inside the function, then explain why the
list still has its contents.


```python
def clear(items):
    items = []

numbers = [1, 2, 3]
clear(numbers)

print(numbers)

# items = [] points the local name at a brand new list. The caller's name still
# points at the original, which was never touched.
```

This is the mutate-against-rebind distinction. To actually empty the caller's list, change the
list itself rather than the local name:


```python
def clear(items):
    items.clear()

numbers = [1, 2, 3]
clear(numbers)

print(numbers)
```

**6.** Write `adder(n)` that returns a function adding `n` to its argument.


```python
def adder(n):
    def add(value):
        return value + n
    return add

add_five = adder(5)

print(add_five(10))
print(adder(100)(1))
```

`adder(5)` has finished by the time `add_five` is called, and `n` is still there because the
returned function carries it. That is a closure.

`adder(100)(1)` builds one and calls it immediately, which is the same thing without keeping a
name for it.
