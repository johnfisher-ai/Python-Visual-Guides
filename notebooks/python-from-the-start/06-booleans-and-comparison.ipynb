## The idea

There are exactly two boolean values, written `True` and `False` with capitals.

You rarely type them. Almost always they arrive as the answer to a comparison.


```python
print(3 < 5)
print(3 > 5)
print(type(3 < 5))
```

`3 < 5` is not a statement of fact you are making. It is a question Python answers, and the
answer is a value you can store, print, or act on.


## Worked examples

### The comparisons

Six of them, and only the first two are unfamiliar in shape.

| Written | Asks |
|---|---|
| `==` | are these equal |
| `!=` | are these different |
| `<` `>` | less than, greater than |
| `<=` `>=` | less than or equal, greater or equal |

`==` has two equals signs because one equals sign already means assignment, which notebook 2
covered.


```python
print(7 == 7)
print(7 != 7)
print(7 <= 7)
print(7 == 7.0)     # same value, different types
```

`7 == 7.0` is `True` because `==` compares **values**. An int and a float can hold the same
quantity, and Python says so.

### Comparing text

Text compares alphabetically, but it is the character codes doing the work, and every capital
letter sorts before every lowercase one.


```python
print("apple" < "banana")
print("Apple" < "apple")
print("Zebra" < "apple")     # a capital Z before a lowercase a
```

`"Zebra" < "apple"` is `True`, which is almost never what a person means by alphabetical.
Compare `text.lower()` on both sides when you want the human answer.


```python
print("Zebra".lower() < "apple".lower())
```

### Chaining, which reads the way you would say it

Python lets you write a range check the way it appears in mathematics. Most languages do not.


```python
age = 7

print(0 < age < 18)
print(0 < 70 < 18)
```

`0 < age < 18` means both comparisons, and it only evaluates `age` once. Written out it would
be `0 < age and age < 18`.

### and, or, not

These combine answers.


```python
print(True and False)     # both must hold
print(True or False)      # either will do
print(not True)           # flip it
```

```python
temperature = 71
raining = False

print(temperature > 65 and not raining)
```

### Truthiness: values that count as true or false

An `if` does not need an actual boolean. Every value in Python counts as one, and the rule is
short enough to memorize.

**These are false.** Everything else is true.

| Falsy | |
|---|---|
| `False` | the boolean |
| `None` | the absence of a value |
| `0`, `0.0` | any zero number |
| `""` | the empty string |
| `[]`, `{}` | an empty list or dictionary |

The pattern behind it: **empty is false, and zero is false**.


```python
print(bool(0), bool(0.0), bool(""), bool([]), bool(None))
print(bool(1), bool(-5), bool("a"), bool([0]))
```

Two of those catch people out, and they are worth looking at twice.


```python
print(bool(" "))     # a space is a character, so the string is not empty
print(bool("0"))     # the text zero is not the number zero
```

`bool([0])` is also `True`: a list holding one item is not empty, and it does not matter that
the item happens to be zero.

This is what `if m:` in notebook 5 was relying on. A match object is truthy and `None` is
falsy, so the test reads naturally without mentioning `None` at all.


```python
name = ""

if name:
    print("got a name")
else:
    print("name is empty")
```

### and and or hand back a value, not a boolean

This surprises everyone the first time, and then becomes useful.


```python
print("" or "default")
print("given" or "default")
print(0 and 99)
print(5 and 99)
```

`or` returns the first thing that is truthy, and `and` returns the first thing that is falsy.
Neither converts to `True` or `False`.

That gives the common idiom for a fallback:


```python
user_input = ""
city = user_input or "Philadelphia"

print(city)
```

### `is` is not `==`

This is where notebook 2 pays off. A name is a label on a value, and two labels can be on the
same value or on two equal values that are not the same object.

- `==` asks **do these have the same value**
- `is` asks **are these the same object**


```python
a = [1, 2, 3]
b = [1, 2, 3]     # an equal list, built separately
c = a             # a second label on the same list

print(a == b, "  a and b hold equal values")
print(a is b, "  but they are two different lists")
print(a is c, "   a and c are one list with two names")
```

`a == b` is `True` and `a is b` is `False`, and both are correct. Change `b` and `a` is
unaffected; change `c` and `a` changes with it, because there is only one list.


```python
c.append(4)

print("a:", a)
print("b:", b)
```

Use `==` for almost everything. `is` has one everyday use, and it is the right way to test for
`None`:


```python
result = None

print(result is None)
print(result is not None)
```

Do **not** use `is` on numbers or text. Whether two equal numbers are the same object depends
on details of the Python build that are not worth learning and can change between versions.
Recent versions of Python will warn you if you try.

### in and not in

`in` asks whether something appears inside something else. For text it means a substring.


```python
city = "Philadelphia"

print("delp" in city)
print("xyz" in city)
print("xyz" not in city)
```

This is usually clearer than `find(...) != -1` from notebook 4, and it is the same test.

### None is a value

`None` means "no value here". It is not zero, and it is not an empty string.


```python
answer = None

print(answer)
print(type(answer))
print(answer is None, answer == 0, answer == "")
```

It is what a function gives back when it has nothing to return, which is notebook 14, and what
`re.search` returned in notebook 5 when nothing matched.


### One oddity worth knowing

`True` and `False` are secretly numbers: `True` is `1` and `False` is `0`.


```python
print(True + True)
print(sum([True, False, True, True]))
```

That last line is genuinely useful. Counting how many things in a collection satisfy a test is
just adding up the answers, which is a trick that comes back in notebook 13 and again in NumPy.

## Common errors

Each cell below is run on purpose so you can see the real message.

### SyntaxError: one equals sign in a test

Assignment and comparison look similar and mean completely different things.


```python
x = 5

if x = 5:
    print("five")
```

`Maybe you meant '==' or ':=' instead of '='?` is Python telling you exactly what happened. An
`if` needs a question, and `=` is not a question.

Ignore the `:=` suggestion for now. It is a real operator with a narrow use, and it is not what
you want here.


### TypeError: those two cannot be ordered

`==` works between any two values and simply answers `False` when they differ in type. `<` does
not.


```python
print("3" < 5)
```

`'<' not supported between instances of 'str' and 'int'` means Python refuses to guess whether
you wanted `3 < 5` or `"3" < "5"`. Convert one side yourself.


```python
print(int("3") < 5)
print("3" < str(5))
```

Note that `==` across types does not raise; it just says they are different.


```python
print("3" == 3)
```

### The quiet one: comparing floats

This does not raise, which is why it belongs in a list of errors.


```python
print(0.1 + 0.2 == 0.3)
```

`False`, for the reason notebook 3 gave: floats are approximations, and `0.1 + 0.2` lands a
hair above `0.3`. Nothing here is broken and nothing will warn you.

Compare with a tolerance whenever floats are involved.


```python
import math

print(math.isclose(0.1 + 0.2, 0.3))
```

## Recap

- A comparison produces a boolean, and `==` compares values while `=` assigns.
- `0 < x < 10` chains, reads like mathematics, and evaluates `x` once.
- Every value is truthy or falsy: **empty is false and zero is false**, everything else is true.
- `and` and `or` return one of the operands, which gives the `value or "default"` fallback.
- `==` asks about value, `is` asks about identity, and `is` is for `None`.
- `in` tests membership and is clearer than checking `find` against `-1`.
- `True` is `1`, so `sum()` over a list of tests counts how many passed.
