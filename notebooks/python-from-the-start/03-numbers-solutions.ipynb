```python
import math
```

**1.** A film runs `500` minutes over three parts. Print how many whole hours and leftover
minutes that is, using `//` and `%`.


```python
total_minutes = 500

hours = total_minutes // 60
minutes = total_minutes % 60

print(hours, "hours and", minutes, "minutes")
```

`//` gives the whole hours and `%` gives what did not fill another hour. Together they
account for every minute.


**2.** A meal costs `47.50` and you want to add a 15 percent tip. Print the tip rounded to
two decimal places, then the total rounded to two decimal places.


```python
bill = 47.50
tip = bill * 0.15

print("tip:", round(tip, 2))
print("total:", round(bill + tip, 2))
```

Round at the point of display, not before. Rounding the tip and then adding would be a
different calculation, and on larger bills the two answers drift apart.


**3.** Print the result of `0.1 * 3 == 0.3`, then print `math.isclose(0.1 * 3, 0.3)`. One
line each, so you can see the two answers disagree.


```python
print(0.1 * 3 == 0.3)
print(math.isclose(0.1 * 3, 0.3))
print(0.1 * 3)          # the reason
```

`0.1 * 3` lands a hair above `0.3`, so exact equality is `False` while `isclose` is `True`.
The third line shows the actual stored value, which is the whole explanation.


**4.** Print `2 ** 64`, then print how many digits that number has. `len(str(...))` will
tell you.


```python
big = 2 ** 64

print(big)
print(len(str(big)), "digits")
```

`str()` turns the number into text, and `len()` counts the characters. No float is involved,
so nothing is approximated.


**5.** Print `int(-7 / 2)` and `-7 // 2` on separate lines. They give different answers. Add
a comment saying which one rounds down and which one cuts toward zero.


```python
print(int(-7 / 2))    # cuts toward zero, giving -3
print(-7 // 2)        # rounds down, giving -4
```

`-7 / 2` is `-3.5`. Cutting toward zero drops the `.5` and leaves `-3`. Rounding down goes to
the next whole number below `-3.5`, which is `-4`.

For positive numbers the two agree, which is why this only shows up with negatives.
