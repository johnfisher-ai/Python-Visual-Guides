## Setup

```python
import math

print("Ready.")
```

## The idea

Python has two kinds of number that you will use constantly.

An **int** is a whole number. It is exact, and it can be as large as you like.

A **float** is a number with a decimal point. It is stored as an approximation, which is the
single most surprising thing in this notebook.


```python
print(7, type(7))
print(7.0, type(7.0))
```

`7` and `7.0` are the same quantity and different types. That distinction decides what happens
when you divide, round, or compare, so it is worth noticing early.


## Worked examples

### The four operations

Addition, subtraction and multiplication behave the way you expect.


```python
print(3 + 4)
print(10 - 4)
print(6 * 7)
```

Division is the one to look at. `/` **always** gives a float, even when the answer is whole.


```python
print(7 / 2)
print(6 / 3)          # exactly 2, but still a float
print(type(6 / 3))
```

This is a deliberate change from Python 2, where `6 / 3` gave an int and `7 / 2` quietly gave
`3`. If you find old code doing division and getting whole numbers, that is why.

### Whole-number division and remainder

When you want the whole part and the leftover, use `//` and `%`.


```python
print(7 // 2)         # how many whole 2s fit in 7
print(7 % 2)          # what is left over
```

These two go together. A common use is splitting one number into units.


```python
total_minutes = 500

hours = total_minutes // 60
minutes = total_minutes % 60

print(total_minutes, "minutes is", hours, "hours and", minutes, "minutes")
```

`%` is also the usual test for divisibility, because a remainder of zero means it divided
evenly.


```python
print(10 % 2)         # 0, so 10 is even
print(11 % 2)         # 1, so 11 is odd
```

### A trap with negative numbers

`//` does not cut the decimal off. It rounds **down**, toward the more negative number, which
for negatives is away from zero.


```python
print(7 / 2, 7 // 2)
print(-7 / 2, -7 // 2)      # not -3
```

`-7 / 2` is `-3.5`, and the next whole number down from `-3.5` is `-4`, not `-3`.

If you want the decimal simply cut off, that is `int()`, which always moves toward zero.


```python
print(int(-3.5), -7 // 2)   # these disagree, on purpose
```

Both are correct; they answer different questions. Pick the one you mean.

### Powers

`**` raises to a power. A fractional power is a root.


```python
print(2 ** 10)
print(2 ** 0.5)       # the square root of 2
print(math.sqrt(2))   # the same thing, said more clearly
```

### Order of operations

Python follows the usual rules: powers first, then multiply and divide, then add and subtract.


```python
print(2 + 3 * 4)      # 14, not 20
print((2 + 3) * 4)    # 20
```

When an expression gets long, break it across lines **inside parentheses**. Python treats
everything between an open bracket and its close as one statement, so you can put the line
breaks wherever they read best.


```python
total = (
    1200            # rent
    + 340           # food
    + 95            # transport
    + 60            # phone
)

print("monthly total:", total)
```

### Floats are approximations

This is the part that catches everyone.


```python
print(0.1 + 0.2)
```

That is not a bug in Python, and every language that stores numbers this way does the same
thing. The cause is that `0.1` cannot be written exactly in binary, any more than one third can
be written exactly as a decimal. You get the closest available value.

Ask for more digits and the stored value shows itself.


```python
print(format(0.1, ".20f"))
print(format(0.2, ".20f"))
print(format(0.1 + 0.2, ".20f"))
```

The tiny excess was there in `0.1` all along. Adding made it visible.

### So do not compare floats with `==`

The comparison you would naturally write is the one that fails.


```python
print(0.1 + 0.2 == 0.3)
```

Compare with a tolerance instead. `math.isclose` asks whether two numbers are near enough,
which is almost always the real question.


```python
print(math.isclose(0.1 + 0.2, 0.3))
```

For **display**, round to the number of decimals you actually want.


```python
print(round(0.1 + 0.2, 2))
```

Rounding for display and comparing with a tolerance are different jobs. Rounding changes the
number you show. `math.isclose` leaves the number alone and changes the question.

### `round` does not do what you were taught

Python rounds a halfway value to the **even** neighbor, not always upward.


```python
print(round(0.5), round(1.5), round(2.5), round(3.5))
```

`2.5` rounds to `2` and `3.5` rounds to `4`. This is the standard used across science and
finance, because always rounding halves upward introduces a small bias that accumulates over
many values.

It surprises people, and it is not a mistake.

### Ints are unbounded

Most languages have a largest integer. Python does not: an int grows to whatever it needs.


```python
print(2 ** 200)
print("2 ** 1000 has", len(str(2 ** 1000)), "digits")
```

Floats do have a limit, because they are stored in a fixed amount of space. You meet it in
Common errors below.

### Converting between them

`int()` and `float()` convert, and `int()` cuts toward zero rather than rounding.


```python
print(int(3.9), int(-3.9))    # cut toward zero, not rounded
print(float(3))
print(int("42"))              # text that looks like a whole number
print(float("3.5"))
```

## Common errors

Each cell below is run on purpose so you can see the real message.

### ZeroDivisionError: you divided by zero

Arithmetic that has no answer stops the program rather than inventing one.


```python
print(10 / 0)
```

`division by zero` is exactly what it says. `//` and `%` fail the same way for the same reason.

When the divisor comes from data rather than from you, check it first: `if count != 0:` before
dividing is the usual fix.


### ValueError: that text is not a whole number

`int()` accepts text, but only text that already looks like an int.


```python
print(int("3.5"))
```

`invalid literal for int() with base 10: '3.5'` means Python read the text, found a decimal
point, and refused to guess whether you wanted `3` or `4`.

Convert in two steps when the text has a decimal point, and decide the rounding yourself:


```python
print(int(float("3.5")))     # cut toward zero, giving 3
print(round(float("3.5")))   # round to even, giving 4
```

### OverflowError: too large for a float

Ints grow as needed. Floats do not.


```python
print(10.0 ** 400)
```

`Result too large` means the number no longer fits in the fixed space a float uses. The largest
is a little under `1.8e308`.

The same power as whole numbers is fine, because ints have no ceiling:


```python
print("10 ** 400 has", len(str(10 ** 400)), "digits")
```

### TypeError: arithmetic on something that is not a number

Text that looks like a number is still text.


```python
print(2 ** "10")
```

`unsupported operand type(s) for ** or pow(): 'int' and 'str'` means one side was text. Convert
it first with `int()` or `float()`.

This is the same family of error as joining text and a number with `+` in notebook 1. Python
does not convert types behind your back.


## Recap

- `int` is a whole number and is exact; `float` has a decimal point and is an approximation.
- `/` always gives a float. `//` gives the whole part, `%` gives the remainder.
- `//` rounds down and `int()` cuts toward zero, which differ for negative numbers.
- `0.1 + 0.2` is not `0.3`, so compare floats with `math.isclose` and round only for display.
- `round` sends a halfway value to the even neighbor, so `round(2.5)` is `2`.
- Ints have no upper limit; floats stop a little under `1.8e308`.
