# Dictionaries


## What you will be able to do

Use a **dictionary** to store and look up values by a key you choose instead of by position,
count things, group things, and read a value that might not be there without crashing.


## The idea

### The problem

A list finds things by position, and position is rarely what you actually know.

You do not want the third price, you want the price **of the blue shirt**. You do not want
reading number 47, you want the reading **from the sensor on the roof**. You do not want row
1,204, you want the record **for the patient with that ID**.

You can fake it. Keep two lists side by side, one of names and one of values, find the position
of the name in the first list, and use that position in the second. It works, and it keeps
working right up until the two lists fall out of step, which they will the first time anything
is inserted or sorted or filtered.

What you want is to store the value **under the name itself**, and look it up that way.

### What a dictionary is

> A **dictionary** stores **pairs**. Each pair has a **key** and a **value**. You look up a
> value by giving its key. Curly braces create one, a colon separates each key from its value,
> and commas separate the pairs.
>
> Keys must be **unchangeable**, which is why **Tuples and Unpacking** spent time on tuples. Values can be
> anything at all, including other dictionaries.

A list answers "what is at position 2?". A dictionary answers "what is stored under `"Ada"`?",
and the second is far more often the question you have.

### Why it is the container you will use most

Two reasons, and the second is the bigger one.

**It matches how data is actually shaped.** A record is a set of named fields, not a numbered
row. A configuration is a set of named settings. A count is a set of things paired with how
often each occurred. In every case the name is the natural handle, and a dictionary stores it
that way with no bookkeeping.

**It is the shape most real data arrives in.** Every JSON response from a web service is a
dictionary, which is the whole of the **APIs and JSON** guide. Configuration files load as dictionaries. A row from
a database or a spreadsheet is naturally a dictionary of column name to value, and that is
exactly what pandas hands you in the **Pandas** guide.

### The two habits worth taking away

Beyond the syntax, this notebook teaches two patterns you will write many times.

**Counting.** How often does each word appear, each category occur, each code repeat. One line
inside a loop does it, and the trick is a default value for keys not yet seen.

**Grouping.** Collect items into buckets by some property: files by extension, people by
department, records by date. Again one line, once you know the idiom.

Both are short, both look cryptic the first time, and both are worth recognizing on sight
because you will meet them in other people's code constantly.

```python
ages = {"Ada": 36, "Grace": 45, "Alan": 41}

print(ages)
print(ages["Ada"])
print(len(ages), "entries")
```

Curly braces, `key: value` inside, commas between pairs. Looking up uses square brackets like a

## Worked examples

### Adding and changing

Assigning to a key that does not exist creates it. Assigning to one that does replaces the
value. There is no separate "add" method.


```python
ages = {"Ada": 36}

ages["Grace"] = 45      # new key, so this adds
ages["Ada"] = 37        # existing key, so this replaces

print(ages)
```

That symmetry is convenient and occasionally dangerous: a typo in a key name silently creates a
new entry rather than raising.

### Reading a key that might not be there

This is the difference worth learning first. Square brackets raise when the key is missing.
`get` returns `None` instead.


```python
ages = {"Ada": 36, "Grace": 45}

print(ages.get("Ada"))
print(ages.get("Bob"))          # missing, and no error
```

`get` takes a second argument for what to return instead, which is usually better than `None`.


```python
print(ages.get("Bob", 0))
print(ages.get("Bob", "not on file"))
```

Use `ages["Ada"]` when a missing key means something is wrong and you want it to stop. Use
`ages.get("Ada", default)` when a miss is expected and normal. This is the same choice
**Strings** described between `index` and `find`.

`in` tests **keys**, not values.


```python
print("Ada" in ages)
print(36 in ages)          # 36 is a value, not a key
print(36 in ages.values())
```

### Removing


```python
ages = {"Ada": 36, "Grace": 45, "Alan": 41}

removed = ages.pop("Grace")     # hands back the value
print(removed, "|", ages)

del ages["Alan"]                # just removes it
print(ages)

print(ages.pop("Bob", "was not there"))    # a default, so no error
```

### Looping over a dictionary

Looping over the dictionary itself gives you the **keys**.


```python
ages = {"Ada": 36, "Grace": 45, "Alan": 41}

for name in ages:
    print(name, ages[name])
```

That works, but `.items()` is better, because it hands you both at once as a tuple, which
**Tuples and Unpacking** taught you to unpack.


```python
for name, age in ages.items():
    print(f"{name:<6} {age}")
```

`.keys()` and `.values()` give you one side each.


```python
print(list(ages.keys()))
print(list(ages.values()))
print(list(ages.items()))
```

The `list()` around each is only there to print them. In a loop you would not need it.

### The order is the order you inserted

Dictionaries keep their pairs in insertion order, and have done since Python 3.7. Older
material will tell you they are unordered, and that advice is out of date.


```python
d = {}
d["zebra"] = 1
d["apple"] = 2
d["mango"] = 3

print(list(d))
```

Insertion order, not alphabetical. When you want a different order, ask for it: `sorted` on a
dictionary sorts the keys and gives back a list.


```python
print(sorted(d))
print(sorted(d.values()))
```

### Counting things

This is the pattern you will use most, and `get` with a default of `0` is what makes it short.


```python
words = "the cat sat on the mat the end".split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

Read the key line as: whatever this word is on so far, or zero if it is new, plus one.

Without `get` you would need an `if` to check whether the key exists yet, every time.


### Grouping things

`setdefault` is the same trick for building lists instead of counting. It returns the value at
a key, inserting a default first if the key is missing.


```python
fruits = ["apple", "avocado", "banana", "blueberry", "cherry"]

groups = {}
for fruit in fruits:
    groups.setdefault(fruit[0], []).append(fruit)

print(groups)
```

The line does three things: get the list under this letter, create an empty one if there is
none, then append to it.

`setdefault` is worth knowing because you will read it in other people's code. If it reads as
too dense, the plain version is just as good:


```python
groups = {}
for fruit in fruits:
    letter = fruit[0]
    if letter not in groups:
        groups[letter] = []
    groups[letter].append(fruit)

print(groups)
```

### Sorting by value

Sorting by key is `sorted(d)`. Sorting by **value** is the question people actually ask, and it
needs `.items()` plus a rule for what to sort on.


```python
counts = {"the": 3, "cat": 1, "hat": 5, "end": 2}

print(sorted(counts.items(), key=lambda pair: pair[1]))
print(sorted(counts.items(), key=lambda pair: pair[1], reverse=True))
```

`lambda pair: pair[1]` is a small unnamed function saying "sort on the second part of each
tuple", which is the value. Functions come in **Functions**, and `lambda` is covered
there; for now this line is worth copying as an idiom.

For just the top one, `max` takes the same kind of rule.


```python
print(max(counts, key=counts.get))
```

### Building one from pairs

`dict()` accepts anything that produces key and value pairs, which is where `zip` from
**Tuples and Unpacking** earns its place.


```python
print(dict([("a", 1), ("b", 2)]))

names = ["Ada", "Grace", "Alan"]
scores = [91, 88, 95]
print(dict(zip(names, scores)))
```

### Merging

`update` changes a dictionary in place. `|` builds a new one.


```python
defaults = {"color": "blue", "size": "medium"}
chosen = {"size": "large"}

print(defaults | chosen)     # a new dictionary, chosen wins
print(defaults)              # unchanged
```

```python
defaults.update(chosen)      # changes defaults itself
print(defaults)
```

When the same key appears on both sides, the right side wins. That is what makes this the
normal way to apply settings over defaults.

### Dictionaries inside dictionaries

Real data is usually nested.


```python
people = {
    "ada":  {"age": 36, "city": "London"},
    "alan": {"age": 41, "city": "Cambridge"},
}

print(people["ada"]["city"])

for name, details in people.items():
    print(f"{name:<5} {details['age']}  {details['city']}")
```

`people["ada"]["city"]` reads left to right, the same as `grid[1][2]` in **Lists**.

Chaining `get` keeps it safe when a level might be missing.


```python
print(people.get("bob", {}).get("city", "unknown"))
```

If `"bob"` is absent, the first `get` returns an empty dictionary, and asking that for `"city"`
returns `"unknown"` rather than raising.

### What can be a key

Keys must be immutable, which **Tuples and Unpacking** explained. Text and numbers are the usual choice, and
tuples work when the key naturally has parts.


```python
grid = {
    (0, 0): "origin",
    (3, 7): "target",
}

print(grid[(3, 7)])
```

## Common errors

Each cell below is run on purpose so you can see the real message.

### KeyError: that key is not there

The most common dictionary error, and the reason `get` exists.


```python
ages = {"Ada": 36, "Grace": 45}
print(ages["Bob"])
```

`KeyError: 'Bob'` names the key it could not find, which is usually enough to spot a typo or a
casing difference. Keys are case sensitive: `"ada"` and `"Ada"` are different keys.

Three ways out, depending on what you mean:


```python
print(ages.get("Bob"))              # None when missing
print(ages.get("Bob", 0))           # a default of your choosing
print("Bob" in ages)                # ask first
```

### RuntimeError: the dictionary changed while you were looping

**Lists** showed that changing a list while looping over it gives a wrong answer silently.
Dictionaries are stricter, and complain.


```python
stock = {"apples": 3, "pears": 0, "plums": 7}

for item in stock:
    if stock[item] == 0:
        del stock[item]
```

`dictionary changed size during iteration` is a better outcome than the list version, because
it stops rather than quietly producing something wrong.

The fix is the same idea: loop over a snapshot, not the live thing. `list(stock)` makes a list
of the keys first.


```python
stock = {"apples": 3, "pears": 0, "plums": 7}

for item in list(stock):
    if stock[item] == 0:
        del stock[item]

print(stock)
```

### TypeError: keys and values are not lists

`.keys()` gives a view, not a list, so it cannot be indexed.


```python
ages = {"Ada": 36, "Grace": 45}
print(ages.keys()[0])
```

`'dict_keys' object is not subscriptable` means you need a real list first. A view is a live
window onto the dictionary, which is why it is cheap and why it does not support positions.


```python
print(list(ages.keys())[0])
```

Needing a position into a dictionary is usually a sign that a list of tuples would suit the
job better.


## Recap

- A dictionary stores values under keys you choose, and `d[key]` looks one up.
- Assigning to a new key adds it; assigning to an existing key replaces the value.
- `d[key]` raises on a missing key; `d.get(key, default)` does not.
- `in` tests **keys**. Use `d.values()` to test values.
- `.items()` gives key and value together as tuples, ready to unpack in a loop.
- Pairs stay in insertion order, and have since Python 3.7.
- `counts[k] = counts.get(k, 0) + 1` counts; `setdefault(k, []).append(v)` groups.
- Sort by value with `sorted(d.items(), key=lambda pair: pair[1])`.
- Never add or delete keys while looping; loop over `list(d)` instead.
