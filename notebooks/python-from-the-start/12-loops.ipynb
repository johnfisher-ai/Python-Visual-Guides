# Loops


## What you will be able to do

Run the same code once for every item in a collection, stop early when you have what you need,
skip the items you do not want, and combine looping with the conditions from **Conditionals** to
filter, count and summarize.


## The idea

### The problem

**Lists** gave you a list holding four hundred survey responses. **Dictionaries** gave you a
dictionary of every word and how often it appeared. In both cases you can reach any single item.
What you cannot yet do is anything to **all** of them.

Suppose you want the average of those four hundred numbers. Written out with what you know so
far, that is four hundred lines of addition, and it only works for a list of exactly that
length. Change the data and the code is wrong. Add a response and you have to edit it.

The work does not change from item to item. Only the item changes. That is the shape a **loop**
exists for.

### What a loop is

> A **`for` loop** runs a block of code once for each item in a collection, assigning the
> current item to a name you choose before each run.
>
> A **`while` loop** runs a block repeatedly for as long as a condition stays true, without
> reference to any collection.

The `for` form is the one you will write most:

```
for response in responses:
    print(response)
```

Read it as a sentence: for each response in responses, print it. The loop variable, `response`,
is created by the loop and holds a different item each time round.

### Why for and not while

Most languages make you manage a counter: start at zero, test against the length, add one, use
the counter to index in. Python can do that, and you should not. `for item in collection`
removes the counter entirely, and with it every off-by-one error the counter could cause.

Reach for `while` only when there is no collection to walk: repeat until the user types
something valid, until a value converges, until a queue is empty. If you know in advance what
you are looping over, `for` is the right tool.

### Where you will meet this

Every piece of data work is a loop over rows, records or files, whether you write the loop
yourself or a library writes it for you. **Comprehensions** introduces a shorter way to write the most
common loops. The **NumPy** and **Pandas** guides replace many of them with operations over whole arrays and columns
at once, which is faster and reads better, and both are built on exactly the idea here.

### What this notebook covers

- `for` over lists, strings, dictionaries and sets

Start with the smallest possible loop.


````python
for name in ["Ada", "Grace", "Alan"]:
    print(name)
````

Three items, so the block ran three times, with `name` holding a different one each time. The
indented block is the body, exactly as it was for `if` in **Conditionals**.

## Worked examples

### Anything you can walk through

A loop works over all four containers, and over a string.


````python
for letter in "hi":
    print("letter:", letter)

for number in (10, 20):
    print("from a tuple:", number)

for item in {"apples", "pears"}:
    print("from a set:", item)
````

Looping over a dictionary gives its **keys**, which **Dictionaries** covered. Use `.items()` when you
want both halves, and unpack them the way **Tuples and Unpacking** showed.


````python
ages = {"Ada": 36, "Grace": 45}

for name in ages:
    print(name)

for name, age in ages.items():
    print(f"{name} is {age}")
````

### range: looping a number of times

Sometimes you want to repeat something a fixed number of times, or you genuinely need the
numbers themselves.


````python
for i in range(5):
    print(i)
````

`range(5)` counts from `0` up to **but not including** `5`, which is the same half-open rule as
slicing in **Strings**. That is what makes `range(len(x))` line up exactly with the valid
positions of `x`.

It takes up to three numbers: where to start, where to stop, and what to step by.


````python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
````

The `list()` around each is only there so you can see the values. `range` does not build a list;
it produces the numbers as they are asked for, which is why `range(1000000)` costs nothing
until you loop over it.


### enumerate: the item and its position

When you need to know where you are as well as what you have, `enumerate` gives both.


````python
cities = ["Philadelphia", "Boston", "Denver"]

for i, city in enumerate(cities):
    print(i, city)
````

This is the tuple unpacking from **Tuples and Unpacking**, applied on each pass. It replaces the counter you
would otherwise keep by hand, and the mistakes that go with it.

`start=1` is useful when the numbers are for a person to read.


````python
for n, city in enumerate(cities, start=1):
    print(f"{n}. {city}")
````

### zip: two collections at once

`zip` walks several collections in step, handing you one item from each.


````python
names = ["Ada", "Grace", "Alan"]
scores = [91, 88, 95]

for name, score in zip(names, scores):
    print(f"{name:<6} {score}")
````

This is the right way to walk two lists together. Indexing into the second one with a counter
works and is easier to get wrong.

`zip` has one behavior worth knowing before it bites you, and Common errors returns to it: when
the collections are different lengths, it stops at the shortest and says nothing.


### The four things loops actually do

Almost every loop you write is one of these, and they are worth recognizing by shape.

**Total something up.** Start at zero, add each item.


````python
scores = [88, 72, 95, 64]

total = 0
for score in scores:
    total = total + score

print(total)
````

**Count the ones that qualify.** Start at zero, add one when the condition holds.


````python
passing = 0
for score in scores:
    if score >= 80:
        passing = passing + 1

print(passing, "of", len(scores))
````

**Collect the ones you want.** Start with an empty list, append what qualifies. This is the
filtering pattern from **Lists**, and the reason you never remove from the list you are
looping over.


````python
high = []
for score in scores:
    if score >= 80:
        high.append(score)

print(high)
````

**Find the best one.** Hold the winner so far, replace it when something beats it.


````python
best = scores[0]
for score in scores:
    if score > best:
        best = score

print(best)
````

Python has built-ins for three of those: `sum`, `len` and `max`. Use them when they fit; they
are faster and shorter. Write the loop when the rule is more complicated than the built-in
allows, and recognize the shape either way, because you will read it constantly.


````python
print(sum(scores), max(scores), min(scores))
````

### break: stop as soon as you have it

`break` leaves the loop immediately, skipping every remaining item.


````python
for n in [4, 9, 16, 25, 36]:
    print("checking", n)
    if n > 10:
        print("found", n)
        break
````

Without `break` this would check all five. With it, the loop ends the moment the answer is
found, which matters when the collection is long or each pass is slow.

### continue: skip this one and carry on

`continue` jumps to the next item without running the rest of the body.


````python
for n in range(10):
    if n % 3 != 0:
        continue
    print(n)
````

That prints only multiples of three. The same result comes from `if n % 3 == 0:` around the
body, and which reads better depends on how much comes after the test. `continue` keeps the
main work from drifting rightward, which is the guard clause idea from **Conditionals** applied
inside a loop.


### while: looping without a collection

A `while` loop repeats as long as its condition holds. Use it when you do not know in advance
how many passes are needed.


````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown = countdown - 1

print("Liftoff")
````

Something inside the loop **must** change the condition, or it will never end. Here it is
`countdown = countdown - 1`. Remove that line and the loop runs forever, printing `3` until you
stop it.

That is worth knowing how to survive rather than avoid. If a cell in Colab never finishes, press
the **stop** button on the cell, or use **Runtime > Interrupt execution**. Nothing is lost;
the cell simply stops.

The safest habit with `while` is to be able to point at the line that will eventually make the
condition false, before you run it.


### The loop else, which surprises everyone

A loop can have an `else`. It runs when the loop finished **without** hitting a `break`.


````python
for n in [1, 3, 5, 7]:
    if n % 2 == 0:
        print("found an even number")
        break
else:
    print("no even number in the list")
````

Nothing broke, so the `else` ran. Put an even number in the list and it will not.

`else` is a poor name for this. Read it as "if we got all the way through", which is what it
means. It saves keeping a `found = False` flag alongside the loop.

You will not need this often, and you will read it in other people's code.


### Loops inside loops

The inner loop runs completely on every single pass of the outer one.


````python
for row in range(1, 4):
    for col in range(1, 5):
        print(f"{row * col:4}", end="")
    print()
````

`end=""` tells `print` not to move to a new line, so the inner loop builds a row across, and the
bare `print()` afterward ends it.

Three passes of the outer loop times four of the inner is twelve runs of the body. That
multiplication is why nested loops get slow quickly: two loops over a thousand items each is a
million passes.

## Common errors

Each cell below is run on purpose so you can see the real message.

### TypeError: that is not something you can loop over

A loop needs a collection. A single number is not one.


````python
for n in 5:
    print(n)
````

`'int' object is not iterable` means Python was handed one value where it expected something
with items in it. This usually happens when `range` was meant:


````python
for n in range(5):
    print(n, end=" ")
````

### IndentationError: the body is missing

The same rule as `if` in **Conditionals**. A colon promises an indented block.


````python
for i in range(3):
print(i)
````

`expected an indented block after 'for' statement` names exactly what is missing.


### NameError: the loop variable outside the loop

The loop variable keeps whatever it held last, which is sometimes useful and often a trap.


````python
for temperature in [10, 20, 30]:
    pass

print(temperature)
````

`30` is still there after the loop, because the name was created in the enclosing code rather
than confined to the loop.

Relying on that breaks the moment the collection is empty, because then the body never runs and
the name is never created. The cell below uses a name that has not appeared anywhere else in
this notebook, so there is no leftover value to hide the problem.


````python
for reading in []:
    pass

print(reading)
````

`name 'reading' is not defined`, because nothing was ever assigned to it.

The choice of a fresh name matters more than it looks. Written with `temperature`, this cell
prints `30` instead of failing, because the earlier loop left that name behind and cells share
one memory, which is what **Running Python** covered. The bug is still there; the stale value simply
hides it, and that is exactly how this reaches production unnoticed.

If you need a value after a loop, create it before the loop starts.


### The quiet one: zip drops what it cannot pair

This does not raise, and it loses data.


````python
names = ["Ada", "Grace", "Alan"]
scores = [91, 88]

for name, score in zip(names, scores):
    print(name, score)
````

Alan is gone. `zip` stopped at the shorter collection and said nothing, and in a longer program
you would never know a row had been dropped.

When the lengths are supposed to match, say so with `strict=True` and let Python check.


````python
names = ["Ada", "Grace", "Alan"]
scores = [91, 88]

for name, score in zip(names, scores, strict=True):
    print(name, score)
````

`zip() argument 2 is shorter than argument 1` turns a silent data loss into a stop. `strict`
arrived in Python 3.10, and it is worth using every time two collections are meant to line up.


## Recap

- `for item in collection:` runs the block once per item, with no counter to get wrong.
- `range(stop)` counts from `0` up to but not including `stop`, matching the slicing rule.
- `enumerate` gives position and item; `zip` walks two collections together.
- Most loops are one of four shapes: total, count, collect, or find the best.
- `break` leaves the loop; `continue` skips to the next item.
- `while` repeats on a condition, and something in the body has to make that condition false.
- A loop `else` runs only when no `break` happened.
- `zip` silently stops at the shortest input, so pass `strict=True` when lengths should match.
