# Environments and pip &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
import sys
import importlib.metadata as md
import importlib.util
```

**1.** Print the version of Python you are running, and the version of `pip`.


```python
print("Python:", sys.version.split()[0])
print("pip:   ", md.version("pip"))
```

`sys.version` on its own includes the build date and compiler, which is more than you usually
want. `.split()[0]` takes just the version number.


**2.** For `csv`, `sqlite3` and one package of your choosing, print whether each is standard
library or an installed package.


```python
import csv
import sqlite3
import json

for module in (csv, sqlite3, json):
    where = module.__file__ or ""
    kind = "installed" if "site-packages" in where else "standard library"
    print(f"{module.__name__:<10} {kind}")
```

All three are standard library. A module with no `__file__` at all is built into the interpreter
itself, which is why the `or ""` is there: `sys` and a few others would otherwise fail on that
line.


**3.** Using `find_spec`, print whether `pandas` and `nosuchlibrary` are available, without
importing either.


```python
for name in ["pandas", "nosuchlibrary"]:
    print(f"{name:<16} {importlib.util.find_spec(name) is not None}")
```

`find_spec` looks the module up without running it. That matters for a large package: importing
`pandas` takes a moment, and this only asks whether it could be found.

Your answer for `pandas` depends on where you are running. Colab has it; a fresh virtual
environment does not.


**4.** In a comment, write the two commands that create a virtual environment and activate
it on macOS or Linux.


```python
# python -m venv .venv
# source .venv/bin/activate

print("both are terminal commands, not Python")
```

The first is Python running its own `venv` module, which is why it starts with `python -m`. The
second is a shell command, and it has no Python equivalent because it changes the shell itself.

On Windows the second line is `.venv\Scripts\activate`.


**5.** In a comment, write the command that records the current packages into
`requirements.txt`, and the one that installs from it.


```python
# pip freeze > requirements.txt
# pip install -r requirements.txt

print("the first writes the file, the second reads it")
```

`>` is the shell redirecting output into a file, not part of pip. The `-r` in the second command
means "read the list from this file".

Commit `requirements.txt` and never commit the `.venv` folder. One is a few lines describing
what you need; the other is thousands of files that can be rebuilt from it.


**6.** Print whether you are currently inside a virtual environment.


```python
print("inside a virtual environment:", sys.prefix != sys.base_prefix)
```

`sys.base_prefix` is the Python the environment was built from. Outside an environment the two
are the same value, so the comparison is `False`.

In Colab this is `False`, because Colab gives every notebook its own machine and has no need to
isolate anything further.
