# Comprehensions &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
prices = [4.50, 12.00, 7.25, 30.00, 9.99]
names = ["ada", "grace", "alan", "kay"]
```

**1.** Build a list of the prices with 10 percent added, rounded to two decimal places.


```python
print([round(p * 1.10, 2) for p in prices])
```

Multiplying by `1.10` adds ten percent in one step. The rounding happens inside the expression,
so every item is rounded as it is built, which matters because these are floats.


**2.** Build a list holding only the prices under 10.


```python
print([p for p in prices if p < 10])
```

The expression at the front is just `p`, meaning keep the item unchanged. That is the pure
filter.


**3.** Build a list of the names with their first letter capitalized.


```python
print([n.title() for n in names])
```

No condition, so every name is kept and every one is transformed. `.title()` returns a new
string, as the **Strings** notebook showed.


**4.** Build a dictionary mapping each name to its length.


```python
print({n: len(n) for n in names})
```

The colon is the only difference from a set comprehension. `key: value` is the same shape
the **Dictionaries** notebook used.


**5.** Given `"the quick brown fox"`, build a set of the distinct letters, ignoring spaces.


```python
phrase = "the quick brown fox"

print({c for c in phrase if c != " "})
```

Curly braces with no colon build a set, so the duplicates collapse without being asked. The
filter drops the spaces before they ever reach the set.

`set(phrase.replace(" ", ""))` gives the same answer and is shorter still.


**6.** Rewrite the over-long comprehension as a loop.


```python
result = []
for w in names:
    if len(w) <= 3:
        continue
    if w == "kay":
        continue
    if w[0] >= "g":
        continue
    result.append(w.upper())

print(result)
```

Three conditions, so three lines, each one readable on its own. Note that each condition
flipped: the comprehension said which words to **keep**, and the loop says which to **skip**,
which is the guard clause shape from the **Conditionals** notebook.

Checking against the original confirms the rewrite:


```python
print([w.upper() for w in names if len(w) > 3 and w != "kay" and w[0] < "g"])
```

Same answer. The loop is four times as long and takes a fraction of the time to understand,
which is the trade this notebook is about.
