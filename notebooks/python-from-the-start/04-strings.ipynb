# Strings

Pull pieces out of text, build new text from values you have, and clean up text that arrived
messy. You will also be able to say why a string method never changes the string you gave it.

Nothing to import. Text is built into Python.

## The idea

A string is a **sequence of characters in order**. Each character sits at a numbered
position, and that number is called its **index**.

It is also **immutable**: once made, a string can never be changed. Everything that looks like
changing text actually builds a new string and hands it back.


```python
city = "Philadelphia"

print(city)
print(len(city), "characters")
print(city[0], "is the character at index 0")
```

`len` counts the characters. `city[0]` means "the character at index 0", and square brackets
after a name are how you always ask for one.

Indexes start at **0**, not 1. The first character is at index 0, the second at index 1, and
in a twelve-character word the last is at index 11, not 12. That gap of one between "first"
and "zero" is the most common source of small mistakes early on, so it is worth saying the
index out loud while you are learning: index zero is `P`.


## Worked examples

### Making a string

Single and double quotes do exactly the same thing. Having both means you can put one kind
inside the other without any fuss.


```python
print('single quotes')
print("double quotes")
print("she said 'hello'")
print('it costs $5, and it is worth it')
```

Pick whichever avoids the escaping. For text longer than a line, use three quotes and the line
breaks are kept.


```python
note = """Dear reader,

This is a string that spans lines.
The line breaks are part of it."""

print(note)
```

### Indexing from either end

A positive index counts from the left, starting at 0. A **negative index** counts from the
right, starting at -1 for the last character. Negative indexes save you writing
`city[len(city) - 1]` every time you want the end.


```python
city = "Philadelphia"

print(city[0])      # first
print(city[1])      # second
print(city[-1])     # last
print(city[-2])     # second from last
```

### Slicing: taking a run of characters

`city[start:end]` gives you everything from `start` up to **but not including** `end`.

That exclusive end is deliberate, and it is worth getting used to rather than fighting.


```python
print(city[0:4])
print(city[4:8])
```

The reason it is convenient: the end of one slice is the start of the next, with no `+ 1`
anywhere, and `end - start` is the length you get.


```python
print(city[0:4] + city[4:12])   # the two halves rejoin exactly
print(len(city[4:8]), "characters, which is 8 - 4")
```

Leave a side blank and Python fills in the obvious end.


```python
print(city[:4])     # from the beginning
print(city[4:])     # to the end
print(city[-3:])    # the last three
print(city[:])      # the whole thing
```

A third number is a **step**. A step of `-1` walks the string backward, which is the usual
way to reverse text.


```python
print(city[::2])    # every second character
print(city[::-1])   # backward
```

### Seeing exactly what is in a string

`print` is for a person reading the output, so it shows text the way it is meant to be read.
That is usually what you want, and it is a problem when you are trying to work out what a value
actually contains.

`repr` is the other view. It shows a value **the way you would type it in code**: quotes
included, and with characters like a tab or a new line written as `\t` and `\n` instead of
being acted on.


```python
value = "  hi\n"

print(value)          # how it reads
print(repr(value))    # what it actually is
```

The first line looks like `hi` with some space, followed by a blank line, and you cannot tell
how much space or where the blank line came from. The second says precisely: two spaces, `hi`,
then a new line character.

Use `repr` whenever the question is "what is really in there?". It is the fastest way to see:

- spaces at the start or end, which are invisible otherwise
- an empty string, which prints as nothing at all
- a `\n` or `\t` that is being acted on rather than shown

This also settles something left open in notebook 1. A bare last line displays itself, and
there the examples were numbers like `7 * 6`, which look identical whichever view you use.
Text is where the two come apart: a bare last line shows the `repr`, with quotes, while
`print` shows the text without them.


```python
city = "Philadelphia"

print(city)     # print: no quotes
city            # bare last line: repr, with quotes
```

### Slicing is forgiving, indexing is not

Ask for a single index past the end and Python stops with an error, which you will see in
Common errors. Ask for a **slice** past the end and it quietly gives you whatever was there.


```python
print(repr(city[0:99]))   # more than exists, no complaint
print(repr(city[99:]))    # nothing there at all, an empty string
```

This is `repr` earning its keep: without the quotes, the second line would be an empty
output and you could not tell whether anything came back at all.

This difference matters: a slice that silently returns nothing is a common reason a program
produces empty output instead of an error.


### Building text from values

`+` joins strings and `*` repeats one.


```python
print("Phil" + "adelphia")
print("-" * 30)
```

For anything with a value in it, use an **f-string**. Put `f` before the quote and any name in
braces is replaced by its value.


```python
name = "Ada"
age = 36

print(f"{name} is {age} years old")
```

Braces can hold a whole expression, not only a name.


```python
print(f"next year {name} will be {age + 1}")
print(f"{name} in capitals is {name.upper()}")
```

A colon inside the braces controls formatting. These two are worth memorizing.


```python
pi = 3.14159
population = 1603797

print(f"pi to two places: {pi:.2f}")
print(f"with thousands separators: {population:,}")
```

`{pi=}` prints the name and the value together, which is the fastest way to check what a
variable holds while you are working.


```python
print(f"{pi=}")
print(f"{population=}")
```

### The quiet mistake: forgetting the f

This one does not raise an error. It prints the braces.


```python
name = "Ada"

print("Hello, {name}")     # no f, so no substitution
print(f"Hello, {name}")    # with the f
```

If your output contains literal braces, this is why. There is no error message to lead you
there, so it is worth recognizing on sight.


### Methods: cleaning and changing text

A method is an operation the string performs on itself, reached with a dot.


```python
city = "Philadelphia"

print(city.upper())
print(city.lower())
print(city.replace("Philadelphia", "Pittsburgh"))
```

`strip()` removes whitespace from both ends, which is most of what cleaning up input means.


```python
messy = "   Ada Lovelace \n"

print(repr(messy))
print(repr(messy.strip()))
```

### None of that changed the original

This is immutability, and it is the point of the section.


```python
city = "Philadelphia"

city.upper()          # this result is thrown away
print(city)           # unchanged

city = city.upper()   # keep it by assigning it
print(city)
```

Every string method **returns a new string** and leaves the original alone. If you do not
assign the result, nothing appears to happen.

This is the most common confusion in this notebook, and it is the same idea as notebook 2: the
name has to be pointed at the new value.


### Splitting and joining

`split` breaks text into a list. `join` puts a list back together.


```python
line = "name,age,city"

fields = line.split(",")
print(fields)
print(len(fields), "fields")
```

```python
print(" | ".join(fields))
print("".join(fields))
```

The string you call `join` on is the **glue** placed between the pieces. Writing the
separator first feels inside out at first, and stops feeling that way quickly.

Called with nothing, `split` breaks on any run of whitespace, which is what you usually want
for a sentence.


```python
print("the  quick   brown fox".split())
```

### Searching

`in` answers whether something appears. `find` answers where.


```python
city = "Philadelphia"

print("delp" in city)
print(city.find("delp"))
print(city.find("xyz"))     # not found
```

`find` returns `-1` when there is no match rather than raising an error. Test for `-1`, and be
careful not to treat it as a real index, because `-1` is itself a valid index meaning the
last character.


```python
print(city.startswith("Phil"))
print(city.endswith("phia"))
print(city.count("a"))
```

### A method that will surprise you

`title()` capitalizes the first letter of every word. Its idea of where a word starts includes
after an apostrophe.


```python
print("ada lovelace".title())
print("don't stop".title())
```

`Don'T Stop` is not a bug report anyone will accept, so avoid `title()` on names and anything
with apostrophes. Capitalizing deliberately is safer.


### Backslashes, and turning them off

A backslash starts an escape: `\n` is a new line, `\t` a tab, and `\\` a single backslash.


```python
print("first\nsecond")
print("a\tb")
print("a single backslash: \\")
```

When you want the backslashes left alone, put `r` before the quote for a **raw string**.


```python
print("C:\name\table")     # \n and \t become a newline and a tab
print(r"C:\name\table")    # left exactly as written
```

Raw strings matter more than they look right now. Notebook 5 is full of patterns written with
backslashes, and they are always written as raw strings for exactly this reason.


### Where string methods stop

Everything above needs you to know the exact text you are looking for. Real questions are
usually vaguer than that: any four digits, a word at the start of a line, a phone number in
either format.


```python
log = "2026-09-04 ERROR disk full"

print(log.startswith("2026"))     # fine if you know the year
print(log.find("ERROR"))          # fine if you know the word
```

Asking for *any* date, rather than this one, is where string methods stop and notebook 5 begins.

## Common errors

Each cell below is run on purpose so you can see the real message.

### TypeError: you cannot change a string in place

This is immutability meeting the assignment you would naturally write.


```python
city = "Philadelphia"
city[0] = "F"
```

`'str' object does not support item assignment` means exactly what it says. A string cannot be
edited; you build a new one.


```python
city = "Philadelphia"
city = "F" + city[1:]     # take everything after the first character
print(city)
```

### IndexError: past the end

Asking for a position that does not exist stops the program.


```python
city = "Philadelphia"
print(city[99])
```

`string index out of range` means the position is beyond the last character. Valid positions
run from `0` to `len(city) - 1`, which is `0` to `11` here.

Note the contrast with slicing above, which returns an empty string instead of complaining. It
is worth remembering which one is silent.


### AttributeError: that method belongs to something else

`append` adds to a list. Strings cannot be added to, so they do not have it.


```python
city = "Philadelphia"
city.append("!")
```

`'str' object has no attribute 'append'` is the error you get whenever you call a method the
value does not have, usually because you thought it was a different type.

`dir(city)` lists everything a string can actually do, and `help(str.replace)` explains one of
them.


### ValueError: index cannot report failure

`find` returns `-1` when there is no match. `index` does the same search but raises instead.


```python
city = "Philadelphia"
print(city.index("xyz"))
```

`substring not found` is the same situation `find` reports quietly. Use `find` when a miss is
expected and normal; use `index` when a miss means something has gone wrong and you want it to
stop.


## Recap

- A string is characters in order, reached by position from `0`, or from the end with `-1`.
- `text[start:end]` excludes the end, so the pieces of consecutive slices fit together exactly.
- Indexing past the end raises; slicing past the end quietly returns what is there.
- f-strings build text from values, with `:.2f` and `:,` for formatting and `{x=}` for checking.
- Methods return a **new** string, so `text.upper()` does nothing unless you assign the result.
- Raw strings, written `r"..."`, leave backslashes alone, which notebook 5 relies on.
