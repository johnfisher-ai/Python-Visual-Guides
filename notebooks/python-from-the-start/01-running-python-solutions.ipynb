# Running Python &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too, and comparing two working answers is often more useful than reading one.


**1.** Print your own name.


```python
print("John")
```

`print` shows whatever you put in the brackets. Text goes in quotes.


**2.** Print the answer to `123 * 456`, with a short label in front of it.


```python
print("123 * 456 is", 123 * 456)
```

The commas put a space between each piece. Python works out the multiplication first, then
prints the answer.


**3.** Store the number 10 in a name called `apples`, then print it.


```python
apples = 10
print(apples)
```

Two steps. The first creates the name and prints nothing. The second displays it. Note there
are no quotes around `apples` in the print: quotes would print the word itself.


**4.** Write a cell whose last line shows a value without using `print`.


```python
100 / 4
```

The notebook displays the value of the last line on its own. Try adding a second line below
it and running again: only the last one shows.


**5.** Write a cell that contains only a comment, and run it. What appears underneath?


```python
# Python ignores this line entirely.
```

Nothing appears. The cell ran successfully and produced no value, so there is nothing to
show. A cell that prints nothing has not necessarily failed.


### One thing worth noticing

Task 3 and task 4 look similar but behave differently. `apples = 10` **stores** a value and
shows nothing. `100 / 4` **produces** a value, and the notebook shows it.

That difference comes up constantly, and it is the reason a cell can run perfectly and still
appear to do nothing at all.
