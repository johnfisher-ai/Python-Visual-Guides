# Values and Variables &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** Create a name `city` holding the text `"Philadelphia"`, then print it.


```python
city = "Philadelphia"
print(city)
```

Text goes in quotes. Without them Python would look for a name called Philadelphia and
raise a NameError.


**2.** Create `length` as 12 and `width` as 5, then print their product with a label.


```python
length = 12
width = 5
print("Area:", length * width)
```

The multiplication happens before the printing. You could also store it first, as
`area = length * width`, which is worth doing if you need it again.


**3.** Create `score` holding 10. On the next line change it to 20. Print it once, at the end. What does it show?


```python
score = 10
score = 20
print(score)
```

It shows 20. The second assignment moved the name onto a new value, and nothing anywhere
remembers the 10.


**4.** Use `type()` to show what kind of value `3.5` is.


```python
print(type(3.5))
```

`float`, because it has a decimal point. `type(3)` would say `int`. You can call `type()`
on a value directly; it does not need a name first.


**5.** Create `first` and `second` holding any two values, then swap them on a single line and print both.


```python
first, second = "apple", "banana"
first, second = second, first
print(first, second)
```

The right side is worked out completely before either name moves, which is why this works
without a third name to hold one value temporarily.


### One thing worth noticing

Task 3 is the one that catches people. Nothing warned you that the `10` was about to be
discarded, and nothing kept a copy. Assignment is silent, and it always wins.

That is fine when you meant it. It is a long afternoon when you reused a name you had forgotten
was already in use, which is the real argument for names that say what they hold.
