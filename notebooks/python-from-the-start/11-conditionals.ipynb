# Conditionals


## What you will be able to do

Use `if`, `elif` and `else` to make Python choose between different pieces of code, and
recognize when a chain of decisions has grown into a shape that is hard to read, and flatten it.


## The idea

### The problem

Every program you have written so far has done exactly the same thing on every run.
Read the cells top to bottom and you know what happens, because nothing depends on anything.

Almost nothing useful works that way. A shipping calculator charges for delivery under fifty
dollars and not over it. A form checks whether an email address was actually supplied. A
cleaning script drops the rows outside the study window and keeps the rest. In every case the
program has to look at a value it did not know in advance, and **do something different
depending on what it finds**.

The **Booleans and Comparison** notebook built the questions, with comparisons, `and` and `or`, and
the rule for which values count as true. Acting on the answers is what this notebook adds.

### What a conditional is

> An **`if` statement** runs a block of code only when a condition is true. The condition is any
> expression Python can judge as true or false. The block is the indented lines beneath it.
>
> An `else` block runs when the condition was false. An `elif` block, short for "else if",
> tests a further condition and is only reached when every condition above it was false.

The shape is always the same: a keyword, a condition, a colon, then an indented block.

```
if temperature > 30:
    print("hot")
```

### Indentation is the block

This is the part that is different from most other languages, and it is worth being explicit
about because the **Running Python** notebook met the error it causes.

Many languages mark a block with braces, and indent it as well so that people can read it. The
braces are what the language reads; the indentation is decoration. Python removes the braces
and promotes the indentation: **the indent is the block**, and it is what decides which lines
belong to the `if` and which run regardless.

That has a consequence worth stating plainly. In Python, badly indented code is not ugly code.
It is different code, or an error. There is no such thing as a purely cosmetic indentation
change inside a block.

Four spaces is the convention. Colab inserts them when you press Tab after a colon.

### Where this shows up

Everywhere, immediately. Filtering rows, validating input, handling the case where a lookup
found nothing, choosing a message, deciding whether to keep going. The **Loops** notebook puts conditions
inside loops, which is where most real filtering happens, and the **Errors and Exceptions** notebook uses them
to decide what to do when something has gone wrong.

## Worked examples

### else for the other path

`else` needs no condition of its own. It runs whenever the `if` condition was false.


````python
temperature = 12

if temperature > 30:
    print("It is hot.")
else:
    print("It is not hot.")
````

Exactly one of those two blocks runs. Never both, never neither.

### elif for more than two outcomes

Chain as many as you need. Python tries them in order and stops at the first that is true.


````python
score = 78

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)
````

Only one block runs, and the rest are skipped without being tested at all.

### Order decides the answer

Because Python stops at the first true condition, a chain in the wrong order can contain code
that can never run.


````python
def label(n):
    if n > 0:
        return "positive"
    elif n > 100:      # unreachable: anything over 100 is already positive
        return "big"
    else:
        return "other"

print(label(500))
````

`500` is over 100, but it never reaches that test, because `n > 0` caught it first. There is no
error and no warning. The branch is simply dead.

The fix is to put the **most specific** condition first.


````python
def label(n):
    if n > 100:
        return "big"
    elif n > 0:
        return "positive"
    else:
        return "other"

print(label(500), label(5), label(-2))
````

Those two cells use `def`, which the **Functions** notebook covers. All you need here is that it names a piece of
code so it can be run with different values.

### Conditions that are not comparisons

The **Booleans and Comparison** notebook established that every value counts as true or false. An `if` accepts any of them.


````python
name = ""

if name:
    print("Hello,", name)
else:
    print("No name given.")
````

`if name:` reads as "if there is a name". Writing `if name != "":` would work too and says less.

The same applies to collections, which is the usual way to check whether one is empty.


````python
results = []

if results:
    print("Found", len(results))
else:
    print("Nothing found.")
````

### Combining conditions

`and`, `or` and `not` from the **Booleans and Comparison** notebook work here directly.


````python
temperature = 22
raining = False

if temperature > 18 and not raining:
    print("Good day for a walk.")
````

Chained comparison is often clearer than `and` when you are checking a range.


````python
age = 35

if 18 <= age < 65:
    print("Working age.")
````

### Nesting, and why it becomes hard to follow

An `if` can contain another `if`. Two levels are usually fine. Beyond that, the code becomes
hard to follow, because understanding any line means holding every condition above it in your
head at once.


````python
user = {"id": 7, "email": "ada@example.com", "active": True}

if user is not None:
    if user.get("id"):
        if user.get("email"):
            if user.get("active"):
                status = "ready"
            else:
                status = "inactive"
        else:
            status = "no email"
    else:
        status = "no id"
else:
    status = "no user"

print(status)
````

That works. It is also four levels deep, the interesting line is the most buried one, and each
`else` is a long way from the condition it belongs to.

### The guard clause

The fix is to turn the nesting inside out. Deal with each **awkward case first**, one at a
time, and let the normal path fall through to the end without any indentation at all.


````python
user = {"id": 7, "email": "ada@example.com", "active": True}

if user is None:
    status = "no user"
elif not user.get("id"):
    status = "no id"
elif not user.get("email"):
    status = "no email"
elif not user.get("active"):
    status = "inactive"
else:
    status = "ready"

print(status)
````

Same answers, same conditions, and every one now sits at the same level. Each line reads as a single
sentence, saying that if this is wrong, here is the outcome. The normal case is the last line rather
than the deepest one.

Conditions that guard against the awkward cases like this are called **guard clauses**. In
the **Functions** notebook, once you can write functions, they become an early `return` and the shape gets
flatter still, but the idea is exactly this one.

Reach for it whenever you find yourself indenting a third time.


### Choosing between two values on one line

When both branches only decide what a name should hold, the whole `if` collapses into one line.


````python
age = 20

label = "adult" if age >= 18 else "minor"

print(label)
````

Read it from the middle, taking the condition first, then what to use if it holds, then the
alternative.

Use this only when it fits comfortably on one line. It is for choosing a value, and an `if`
statement is for doing two different things.


### match in newer code

When you are comparing one value against several fixed possibilities, `match` says it more
directly than a chain of `elif`.


````python
status_code = 404

match status_code:
    case 200:
        message = "OK"
    case 404:
        message = "Not found"
    case 500:
        message = "Server error"
    case _:
        message = "Something else"

print(message)
````

`case _` is the catch-all, like `else`. `match` arrived in Python 3.10 and can do a great deal
more than this, but comparing against fixed values is the use worth knowing now.

An `elif` chain does the same job and works everywhere, so nothing here depends on it.

## Common errors

Each cell below is run on purpose so you can see the real message.

### SyntaxError: the colon is missing

Every `if`, `elif` and `else` ends with a colon.


````python
x = 5

if x > 3
    print("big")
````

`expected ':'` is about as clear as an error message gets. The colon is what tells Python a
block is coming.


### IndentationError: the block is missing

A colon promises an indented block, and Python insists on getting one.


````python
x = 5

if x > 3:
print("big")
````

`expected an indented block after 'if' statement on line 3` names the statement that was left
without a body. Indent the line by four spaces and it belongs to the `if`.

This is the same rule the **Running Python** notebook met from the other direction, where an unexpected indent was
the problem. Space at the start of a line always means something in Python.


### SyntaxError: elif after else

`else` is the end of a chain. Nothing can follow it.


````python
x = 5

if x > 10:
    print("big")
else:
    print("small")
elif x > 3:
    print("medium")
````

`invalid syntax`, with the caret under the `elif`, means the chain is out of order. Python 3.14
names the problem outright, as `'elif' block follows an 'else' block`. Every `elif` goes above
the `else`, because `else` is what happens when all of them failed.


### The quiet one: a branch that can never run

This does not raise. It gives a wrong answer and no warning.


````python
def size(n):
    if n > 0:
        return "positive"
    elif n > 1000:
        return "huge"
    else:
        return "zero or less"

print(size(999999))
````

The function returns `positive` for every positive number, and `huge` can never be returned for any
input at all. The first condition already covers everything the second one would match.

Whenever a chain of conditions overlaps, put the **narrowest** case first. Reading a chain from
the top and asking "what does this let through?" catches it.


## Recap

- `if condition:` runs the indented block below it only when the condition is true.
- Indentation **is** the block in Python, so it is meaning rather than formatting.
- `else` runs when the `if` did not; `elif` tests another condition and only runs if all above
  it failed.
- Python stops at the first true condition, so an overlapping chain in the wrong order leaves
  branches that can never run.
- Any value works as a condition, so `if results:` is the normal way to ask "is this empty".
- Nesting past two levels is a signal: flatten it into **guard clauses** that handle the
  awkward cases first.
- `value_if_true if condition else value_if_false` chooses between two values on one line.
