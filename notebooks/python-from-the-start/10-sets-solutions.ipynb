# Sets &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
morning = ["tea", "toast", "eggs", "tea", "juice"]
evening = ["soup", "toast", "tea", "bread"]
```

**1.** Print how many different items appear in `morning`.


```python
print(len(set(morning)))
```

`morning` has five entries and one repeat, so four distinct items. `len(morning)` would
have answered a different question.


**2.** Print the items that appear in both lists, sorted.


```python
print(sorted(set(morning) & set(evening)))
```

`&` keeps only what both sides have. The `sorted` is there because a set has no order, so
printing it directly would give an arrangement you cannot rely on.


**3.** Print the items that appear in `morning` but not in `evening`, sorted.


```python
print(sorted(set(morning) - set(evening)))
```

Subtraction removes everything the right side contains. Reversing the operands answers the
opposite question, and gives a different list.


**4.** Print every item across both lists, sorted, with no repeats.


```python
print(sorted(set(morning) | set(evening)))
```

`|` collects both sides, and the set rule drops the duplicates without being asked.


**5.** Given `required = {"name", "email", "age"}` and `form = {"name", "email"}`, print
whether the form has everything required, then print what is missing.


```python
required = {"name", "email", "age"}
form = {"name", "email"}

print(required <= form)
print(sorted(required - form))
```

`required <= form` asks whether every required field is present, which is `False` here.
`required - form` names exactly which ones are absent, which is what an error message would
need to say.


**6.** Given `"the quick brown fox jumps over the lazy dog"`, print how many different
letters it uses, ignoring spaces.


```python
phrase = "the quick brown fox jumps over the lazy dog"

letters = set(phrase)
letters.discard(" ")

print(len(letters))
print(sorted(letters))
```

`set(phrase)` gives every distinct character, including the space, so the space is removed
afterward. `discard` rather than `remove` means the line would still work on a phrase with no
spaces in it.

The answer is 26: this sentence is a pangram, using every letter of the alphabet at least once.
