**1.** Write `safe_int(text)` that returns the number, or `None` if the text is not a
number.


```python
def safe_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(safe_int("7"))
print(safe_int("seven"))
```

`except ValueError:` names the one failure expected here. A different problem would still
reach you as a traceback.


**2.** Rewrite it to print the exception message before returning `None`, using `as`.


```python
def safe_int(text):
    try:
        return int(text)
    except ValueError as e:
        print("could not convert:", e)
        return None

print(safe_int("seven"))
```

The message is the same text a traceback would have shown, which is what makes it worth
printing rather than inventing your own wording.


**3.** Write a function that divides two numbers and handles both a non-numeric input and a
zero divisor, with a different message for each.


```python
def divide(text, divisor):
    try:
        return int(text) / divisor
    except ValueError:
        return "not a number"
    except ZeroDivisionError:
        return "cannot divide by zero"

print(divide("10", 2))
print(divide("ten", 2))
print(divide("10", 0))
```

Two `except` blocks, each naming one type. Order does not matter here because `ValueError` and
`ZeroDivisionError` are unrelated, and putting the specific case first is still the habit worth
keeping.


**4.** Given `ages = {"ada": 36}`, look up `"grace"` and print `"not on file"` instead of
raising.


```python
ages = {"ada": 36}

try:
    print(ages["grace"])
except KeyError:
    print("not on file")
```

`.get()` from the **Dictionaries** notebook does the same job in one line:


```python
print(ages.get("grace", "not on file"))
```

Use `.get()` when a miss is ordinary. Use `try` and `except KeyError` when a miss means
something has gone wrong and you want to handle it deliberately.


**5.** Write `withdraw(balance, amount)` that raises `ValueError` when the amount is more
than the balance.


```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError(f"cannot withdraw {amount} from a balance of {balance}")
    return balance - amount

print(withdraw(100, 30))
```

```python
print(withdraw(100, 300))
```

The message names both numbers, so whoever reads the traceback does not have to go looking for
what the values were. Returning `None` or `0` instead would have let the program carry on with a
wrong balance.


**6.** Given `values = ["3", "4", "x", "5"]`, total the ones that are numbers and print
how many were skipped.


```python
values = ["3", "4", "x", "5"]

total = 0
skipped = 0
for v in values:
    try:
        total += int(v)
    except ValueError:
        skipped += 1

print("total:", total)
print("skipped:", skipped)
```

Counting what was skipped is the part worth keeping. A loop that silently drops bad values gives
an answer with no indication of how much of the data it is based on.
