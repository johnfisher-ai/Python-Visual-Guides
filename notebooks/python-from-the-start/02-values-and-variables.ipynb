# Values and Variables


The **Running Python** notebook had you store something in a name and print it back. This one
explains what that actually did, because the answer is not what most people picture.


## What you will be able to do

Create names, point them at values, and change what they point at. You will also be able to say
what a variable is in Python, which is not a box with something inside it.


## The idea

### The problem

In the **Running Python** notebook you ran `7 * 6`, saw `42` appear, and then lost it. The answer was computed,
displayed, and thrown away. The next cell had no way to refer to it.

That is fine for a calculator. It is useless for a program. Any piece of real work computes
something and then needs it again: a total to compare against, a cleaned-up piece of text to
save, a result from a slow step you do not want to repeat. You need a way to **keep** a value
and **refer back to it**, and you need to be able to give it a name that says what it is.

### What a variable is

> A **variable** is a **name attached to a value**. Assignment, written with a single `=`,
> creates that attachment. Python works out the value on the right first, then attaches the
> name on the left to it.

```
apples = 10
```

Read it from right to left, working out `10` first and then attaching the name `apples` to it. The
name is not part of the value and the value does not know its own name.

### The picture that matters

A common first picture is a box labeled `apples` with `10` sitting inside it, where assignment puts
something in the box and using the name takes it out.

That picture is wrong, and it is worth replacing now rather than later.

A more accurate picture is a **label stuck onto a value**. The value exists somewhere. The name
is a tag pointing at it. Assignment moves the tag; it does not fill a container.

Both pictures give the same answers for simple code, which is why the box survives so long. They
come apart when **two names** are attached to one value, and when the value is something that can be
**changed in place**. In those cases the box picture predicts wrong results, and the label picture
predicts the right ones.

This notebook ends by putting both pictures in front of the same three lines of code, so you
can watch one of them fail.

### Where this shows up

Every line of Python you write from here on. There is no program without variables, and the
label idea underneath them explains behavior you will meet again in lists, in dictionaries, and
in every function you pass data to.

### What this notebook covers

- Assignment, and what actually happens when you write `=`
- Moving a name onto a different value, and why the old one is simply gone
- Types, and the fact that a name is not tied to one
- Naming rules, and the conventions worth following
- Assigning several names at once, and the one-line swap
- Two names on one value, which is where the box picture breaks
- That every value carries operations you reach with a dot

## Worked examples

### The name is not the value

Printing `apples` shows the value. Printing `"apples"` shows the word, because quotes make it
text rather than a name.


````python
apples = 10
print(apples)
print("apples")
````

### A name can be moved

Assign to the same name again and the label comes off the old value and goes onto the new one.
Nothing is added; the name simply refers to something else now.


````python
apples = 10
print(apples)

apples = 25
print(apples)
````

The `10` is gone. There is no history, and no way to ask what `apples` used to be.

### A name can point at anything

A name is not tied to a kind of value. The same name can hold a number, then text, then
something else entirely. `type()` tells you what it is holding right now.


````python
thing = 42
print(thing, type(thing))

thing = "forty two"
print(thing, type(thing))

thing = 42.0
print(thing, type(thing))
````

`int` is a whole number, `str` is text, `float` is a number with a decimal point. The **Numbers** notebook
covers numbers and the **Strings** notebook covers text.

Notice that one name held a number, then text, then a number again. Python is **dynamically typed**,
which means the type belongs to the value and the name has none, so a name can be moved onto any
kind of value at any time. Languages like Java and C work the other way, where you declare up front
what a name is allowed to hold and it never changes.

### Using a name in a calculation

Wherever a value would work, a name holding that value works too.


````python
price = 3
quantity = 7
total = price * quantity

print("Total:", total)
````

`total` was worked out **once**, at the moment that line ran. Change `price` afterwards and
`total` does not follow, because the name was attached to the answer that the calculation produced.


````python
price = 3
quantity = 7
total = price * quantity

price = 100          # total does not care
print("Total is still:", total)
````

### Several names in one assignment

You can assign several names on one line. The right side is worked out first, all of it, then
the names are attached.


````python
width, height = 4, 9
print(width, height)
print("Area:", width * height)
````

Because the right side is finished before any name is attached, swapping two values takes one
line and needs no temporary name.


````python
a, b = "first", "second"
print(a, b)

a, b = b, a
print(a, b)
````

### Choosing names

Python only requires that a name starts with a letter or underscore and contains letters,
digits and underscores. Everything else is convention, and the conventions are worth following
because you will be reading this code later.

- `snake_case`, lowercase with underscores: `total_price`, `user_name`
- Say what it holds: `n` and `x` tell a reader nothing, `days_remaining` tells them everything
- Names are **case sensitive**: `Total` and `total` are two different names


````python
days_remaining = 14
Days_Remaining = 3     # a completely separate name

print(days_remaining, Days_Remaining)
````

### A second name for the same value

This is where the label picture and the box picture stop agreeing, and where the box picture
starts giving wrong answers.

Assign one name to another and you do not get a copy. You get a **second label on the same
value**.


````python
x = 10
y = x        # y now labels the same 10

x = 99       # x is moved onto a different value
print("x is", x)
print("y is", y)
````

`y` is still 10, which is what you would expect either way. Reassigning `x` moved that one
label and left `y` alone.

Now the same thing with a value that can be **changed in place** rather than replaced. This
uses a list, which you meet properly in the **Lists** notebook; all you need here is that the square
brackets hold several values and that `.append()` adds one.


````python
x = [1, 2, 3]
y = x            # again, a second label on the same list

x.append(4)      # this changes the list itself, it does not move the label

print("x is", x)
print("y is", y)
````

`y` changed too, and nothing in that code mentions `y`.

Both names were attached to one list, and `.append()` altered the list rather than replacing it. The
box picture cannot explain this. The label picture can, because there was only ever one list and
both labels are still on it.

Values that can be changed in place are called **mutable**. Numbers and text are **immutable**, so
you can never change one, only move a name onto a different one. That single distinction explains a
whole category of confusing bugs, and the **Lists** notebook returns to it in detail.


### Everything is an object

Every value in Python carries information about itself and operations it can perform, and you reach
both with a dot.

Even a plain number.


````python
n = 4.5

print(n.is_integer())     # is this float a whole number?
print((5.0).is_integer())

price = 19.99
print(price.__class__)    # the same thing type() told you
````

`n.is_integer()` is a **method**, an operation the value can perform on itself. Text carries a large
collection of them, and they make up most of the **Strings** notebook.

This is why the dot appears everywhere in Python once you start reading real code. It is not
special syntax for particular libraries; it is how you reach anything a value carries.

## Common errors

### NameError: Python does not guess

In the **Running Python** notebook this came from running cells out of order. The other cause is
simpler, a name typed differently from the one you created, since Python matches names exactly,
including case.


````python
message = "hello"
print(mesage)
````

One letter. `NameError: name 'mesage' is not defined`, and Python offers no correction because
it has no idea what you meant.

Recent versions add a hint when the name is close to one that exists, which is worth reading
when it appears.

### SyntaxError: that is not a valid name

A name cannot start with a digit. Python starts reading `2nd_place` as a number and then fails
when letters arrive.


````python
2nd_place = "silver"
````

`invalid decimal literal` is Python saying "I thought this was a number". `second_place` or
`place_2` are both fine.

### SyntaxError: assignment has a direction

`=` puts the value on the right into the name on the left. It is not a statement that two
things are equal, so it cannot be written in reverse.


````python
5 = apples
````

`cannot assign to literal here` means the left side has to be a name. `5` is a value, and a
value cannot be given a new meaning.

If you wanted to **compare** two things rather than assign, that is `==`, which the **Booleans and Comparison** notebook
covers.


## Recap

- Assignment reads right to left: work out the value, then attach the name.
- A variable is a **name attached to a value**, a better picture than a box holding one.
- Reassigning moves the name. The old value is simply no longer referred to.
- A name can hold any type, and `type()` says what it is holding now.
- A calculated name holds the **answer**, so it does not update by itself when its inputs change.
- `a, b = b, a` swaps two values, because the right side is finished before either name moves.
- Assigning one name to another gives a **second label on the same value**, and makes no copy.
- A value that can be changed in place is **mutable**, and then both labels see the change.
- Every value carries methods you reach with a dot, even a number.
- Names are case sensitive, and `snake_case` that says what it holds is worth the extra typing.
