# Tuples and Unpacking &middot; Solutions

**1.** Make a tuple `today` holding the year, month and day as three numbers. Print it,
then print just the year.


```python
today = (2026, 9, 4)

print(today)
print(today[0])
```

A date is a good example of what a tuple is for: three numbers that mean different things
and always come as a set of exactly three.


**2.** Unpack `today` into three names on one line, then print them separated by slashes.


```python
year, month, day = today

print(f"{year}/{month}/{day}")
```

Three names for a three-item tuple. Naming them beats using `today[0]`, `today[1]` and
`today[2]` in the line below.


**3.** Create `low = 1` and `high = 10`, print them, swap them on a single line, and print
them again.


```python
low, high = 1, 10
print(low, high)

low, high = high, low
print(low, high)
```

The right side builds `(10, 1)` before either name changes, so nothing is lost. Doing it with
one assignment at a time would need a third name to hold the first value.


**4.** Given `pairs = [("apple", 3), ("pear", 5), ("plum", 2)]`, loop over it, unpacking
each item into a fruit and a count, and print one line per item, like `apple: 3`.


```python
pairs = [("apple", 3), ("pear", 5), ("plum", 2)]

for fruit, count in pairs:
    print(f"{fruit}: {count}")
```

`for fruit, count in pairs` unpacks each tuple as the loop hands it over, so the body reads
in terms of what the values mean.


**5.** Using `divmod`, print how many whole hours and leftover minutes are in `500`
minutes.


```python
hours, minutes = divmod(500, 60)

print(hours, "hours and", minutes, "minutes")
```

The **Numbers** notebook did this with `//` and `%` on separate lines. `divmod` does both divisions at once
and returns them as a tuple, which unpacks directly into two names.


**6.** Given `readings = [7, 3, 9, 4, 1]`, unpack the first value into `first` and
everything else into `rest`, then print both along with the total of `rest`.


```python
readings = [7, 3, 9, 4, 1]

first, *rest = readings

print("first:", first)
print("rest: ", rest)
print("total of rest:", sum(rest))
```

`*rest` collects whatever is left, always as a list even when the source was a tuple. That is
why `sum(rest)` works without any conversion.
