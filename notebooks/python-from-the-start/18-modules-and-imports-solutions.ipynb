# Modules and Imports &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** Create the folder and path setup, then write `practice/shapes.py` containing
`area(width, height)`.


```python
import sys
import shutil
import importlib
from pathlib import Path

work = Path("practice")
work.mkdir(exist_ok=True)

if str(work.resolve()) not in sys.path:
    sys.path.insert(0, str(work.resolve()))

shapes_code = '''
def area(width, height):
    return width * height
'''

(work / "shapes.py").write_text(shapes_code)

print("wrote", work / "shapes.py")
```

The module is written as a string and saved to disk. `sys.path.insert` puts the folder at the
front of the search order, which is what lets the next cell find it.


**2.** Import the module and call `area(3, 4)` using the `import shapes` form.


```python
import shapes

print(shapes.area(3, 4))
```

`shapes.area` keeps the source of the name visible, which is why this form is the default.


**3.** Import just the function with `from shapes import area`.


```python
from shapes import area

print(area(3, 4))
```

Same result, shorter call, and no indication in this line of where `area` came from. That
trade is the whole difference between the two forms.


**4.** Print the module's `__name__` and its public names.


```python
print("name:  ", shapes.__name__)
print("public:", [n for n in dir(shapes) if not n.startswith("_")])
```

`__name__` is `shapes` because the module was imported. Running the file directly would
make it `__main__`.


**5.** Add a `perimeter` function to the file, import the module again, and call
`shapes.perimeter(3, 4)`.


```python
shapes_code = '''
def area(width, height):
    return width * height


def perimeter(width, height):
    return 2 * (width + height)
'''

(work / "shapes.py").write_text(shapes_code)

import shapes

print(shapes.perimeter(3, 4))
```

`module 'shapes' has no attribute 'perimeter'`, even though the file on disk plainly has one.

The `import shapes` on the third line did nothing, because `shapes` was already in memory from
task 2. Python returned the cached module without reading the file.


**6.** Make the new function available with `importlib.reload`, call it, then delete the
folder.


```python
importlib.reload(shapes)

print(shapes.perimeter(3, 4))
print(shapes.area(3, 4))
```

`reload` re-reads the file, so the new function appears and the existing one still works.

Note that `area` imported earlier with `from shapes import area` still points at the old
function object. That is fine here because it did not change, and it is the reason `reload` is
a development convenience rather than something to depend on.


```python
shutil.rmtree(work)

print("practice still there:", work.exists())
```
