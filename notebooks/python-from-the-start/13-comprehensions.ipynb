# Comprehensions


## What you will be able to do

Write a **comprehension**, which builds a list, dictionary or set in one line from something you
can loop over. You will also be able to tell when a comprehension has stopped helping, and turn
it back into a loop.


## The idea

### The problem

The **Loops** notebook named four shapes that almost every loop takes, and one of them came up
again and again: build an empty list, walk a collection, and append the items you want.

```
kept = []
for word in words:
    if len(word) > 4:
        kept.append(word)
```

Four lines, and three of them are bookkeeping. The empty list, the loop, and the append are the
same every time. Only `words`, the condition and what gets appended ever change.

When the same four lines appear in a file twenty times, they stop carrying meaning. A reader has
to decode the machinery each time to find the one idea inside it.

### What a comprehension is

> A **comprehension** builds a new collection from an existing one, in a single expression.
> It names what each item should become, where the items come from, and optionally which ones
> to keep.
>
> ```
> [ what_to_keep   for item in collection   if condition ]
> ```
>
> Square brackets build a list, curly braces with a colon build a dictionary, curly braces
> without one build a set.

The four-line loop above becomes:

```
kept = [word for word in words if len(word) > 4]
```

### How to read one

Not left to right. Read it in the order the machine does:

1. **`for item in collection`** first. This is the loop, and it is in the middle.
2. **`if condition`** next, if there is one. Items that fail are skipped.
3. **The expression at the front** last. This is what goes into the new collection.

Reading the middle first feels wrong for about a week and then becomes automatic. It is worth
doing deliberately at the start rather than guessing.

### Where you will meet this

Constantly, in code written by other people. A comprehension is the most recognizably Python
thing in the language, and experienced Python programmers reach for one by default when
building a collection.

They also matter beyond this guide. The **NumPy** and **Pandas** guides replace many
comprehensions with operations over whole arrays and columns, which is faster still, but the
thinking is the same: describe the result rather than the steps.

### The part the title promises

A comprehension is not always an improvement. Past a certain amount of logic it becomes a line
that is technically shorter and genuinely harder to read, and the loop was better. That point
is real, it arrives sooner than people expect, and recognizing it is as much a part of using
comprehensions well as the syntax is.

### What this notebook covers

- The basic form, and reading it in the right order
- Filtering with `if`, and transforming at the same time
- Choosing between two values with `if` and `else` at the front
- Dictionary and set comprehensions
- Multiple loops in one comprehension, and comprehensions inside comprehensions
- Where the readability cliff is, and what to do at it
- **Generator expressions**, which look almost identical and behave very differently
- Three errors, including one that a loop would not have

Start with a loop and its one-line equivalent, side by side.


````python
words = ["apple", "fig", "banana", "kiwi"]

kept = []
for word in words:
    if len(word) > 4:
        kept.append(word)

print(kept)
print([word for word in words if len(word) > 4])
````

Same answer. The second version says what the result is; the first says how to build it.

## Worked examples

### Transforming every item

The simplest comprehension has no condition at all. It changes each item and keeps them all.


````python
numbers = [1, 2, 3, 4, 5, 6]

print([n * 2 for n in numbers])
print([n ** 2 for n in numbers])
````

````python
words = ["apple", "fig", "banana", "kiwi"]

print([w.upper() for w in words])
print([len(w) for w in words])
````

The expression at the front can be anything: arithmetic, a method call, a function call. It runs
once per item.

### Filtering

Add `if` at the end to keep only some items.


````python
print([n for n in numbers if n % 2 == 0])
print([w for w in words if len(w) > 4])
````

Here the front expression is just `n`, meaning "keep the item unchanged". That is the pure
filter, and it is the most common comprehension of all.

### Both at once

Transform and filter in the same line. The condition is applied first, and only survivors reach
the expression at the front.


````python
print([n * 2 for n in numbers if n % 2 == 0])
````

Read the middle first: take each `n` in `numbers`; keep it if it is even; double it. `[4, 8, 12]`,
not twelve items with the odd ones doubled.

### Choosing between two values

There is a second place `if` can appear, and it means something completely different. At the
**front**, paired with `else`, it is the one-line conditional from the **Conditionals** notebook.


````python
print(["even" if n % 2 == 0 else "odd" for n in numbers])
````

That keeps every item and decides what each becomes. Compare it with the filter form:

- `[x for x in items if cond]` decides **which items** survive. The `if` is at the end and has
  no `else`.
- `[a if cond else b for x in items]` decides **what each item becomes**. The `if` is at the
  front and must have an `else`.

The rule that keeps them apart: an `if` at the end filters, and an `if` at the front always
needs an `else`.

### Dictionaries and sets

Same idea, different brackets.


````python
words = ["apple", "fig", "banana", "kiwi"]

print({w: len(w) for w in words})
````

A colon between two expressions makes it a dictionary comprehension, and `key: value` is the
same shape the **Dictionaries** notebook used.


````python
print({len(w) for w in words})
````

No colon, so this builds a set. Four words gave four lengths and the duplicates would have
collapsed, exactly as the **Sets** notebook described.


### More than one loop

A comprehension can carry several `for` clauses. They read left to right in the same order the
nested loops would appear.


````python
print([(a, b) for a in [1, 2] for b in "xy"])
````

That is the outer loop first, then the inner, giving every combination. The equivalent loop is:

```
for a in [1, 2]:
    for b in "xy":
        ...
```

The most useful version of this flattens a list of lists.


````python
matrix = [[1, 2, 3], [4, 5, 6]]

print([x for row in matrix for x in row])
````

### Where it stops being readable

Here is a comprehension that works perfectly and should not be written.


````python
records = [
    {"name": "ada",   "dept": "eng",   "salary": 120, "active": True},
    {"name": "grace", "dept": "eng",   "salary": 140, "active": False},
    {"name": "alan",  "dept": "sales", "salary": 90,  "active": True},
    {"name": "kay",   "dept": "eng",   "salary": 110, "active": True},
]

print([r["name"].title() for r in records
       if r["active"] and r["dept"] == "eng" and r["salary"] > 100])
````

It is correct. It is also three separate rules crammed into one condition, and a reader who
wants to know why someone was excluded has to hold all of it at once.

The loop is longer and better.


````python
result = []
for r in records:
    if not r["active"]:
        continue
    if r["dept"] != "eng":
        continue
    if r["salary"] <= 100:
        continue
    result.append(r["name"].title())

print(result)
````

Each rule is on its own line, in the guard clause shape from the **Conditionals** notebook. You
can read one, understand it, and move on. You can put a comment on one. You can delete one
without rebalancing a chain of `and`.

**The rule of thumb:** if the comprehension does not fit comfortably on one line, or has more
than one condition, or nests another comprehension inside it, write the loop.

Short comprehensions are clearer than loops. Long ones are worse, and there is no prize for the
one-liner.


````python
print([[row * col for col in range(1, 4)] for row in range(1, 3)])
````

That builds a grid, and it is already at the edge. One more level and nobody, including the
person who wrote it, will be able to read it a month later.


### Generator expressions

Swap the square brackets for parentheses and you get something that looks almost identical and
behaves quite differently.


````python
numbers = [1, 2, 3, 4, 5, 6]

print([n * 2 for n in numbers])
print((n * 2 for n in numbers))
````

The second did not produce a list. It produced a **generator**: a recipe for the values rather
than the values themselves. Nothing has been calculated yet.

Values arrive as something asks for them.


````python
print(sum(n * 2 for n in numbers))
print(list(n * 2 for n in numbers))
````

When a generator is the only argument, the extra parentheses can be dropped, which is why
`sum(n * 2 for n in numbers)` reads so cleanly.

Two differences matter.

**A generator is single use.** Once consumed, it is empty.


````python
gen = (n for n in [1, 2, 3])

print(list(gen))
print(list(gen))     # nothing left
````

A list can be walked as many times as you like. A generator cannot, and the second pass gives
you an empty result rather than an error, which makes it a quiet bug.

**A generator holds almost nothing.** The list builds every value in memory; the generator holds
only the recipe.


````python
import sys

as_list = [n for n in range(10000)]
as_gen = (n for n in range(10000))

print("list:     ", sys.getsizeof(as_list), "bytes")
print("generator:", sys.getsizeof(as_gen), "bytes")
````

That difference is why you would use one. When you are summing, counting or searching through
something large and do not need to keep it, a generator does the job without building the whole
collection first.

Use a list when you need the values more than once, or need to index into them. Use a generator
when you are passing straight into `sum`, `max`, `any` or a loop.

## Common errors

Each cell below is run on purpose so you can see the real message.

### TypeError: that is not something you can loop over

The same error a `for` loop gives, for the same reason.


````python
print([n for n in 5])
````

`'int' object is not iterable` means the middle of the comprehension was handed a single value
where a collection was expected. `range(5)` is almost always what was meant.


### SyntaxError: the parts are in the wrong order

The order is fixed: what to keep, then the loop, then the filter.


````python
numbers = [1, 2, 3]
print([if n > 1 n for n in numbers])
````

An `if` cannot come first. Written the right way round it is
`[n for n in numbers if n > 1]`, and if you meant to choose between two values then the `if`
belongs at the front **with** an `else`: `["big" if n > 1 else "small" for n in numbers]`.


### NameError: the loop variable does not survive

This is a real difference from a `for` loop, and it is in your favor.


````python
squares = [q ** 2 for q in range(4)]

print(squares)
print(q)
````

`name 'q' is not defined`. The comprehension's variable exists only inside it.

Compare that with the `for` loop in the **Loops** notebook, where the loop variable was still
there afterward and held its last value. A comprehension cannot leak a name into your code by
accident, which is one fewer thing to think about.


## Recap

- A comprehension builds a collection in one expression: `[expr for item in collection if cond]`.
- Read the middle first, then the filter, then the expression at the front.
- An `if` at the **end** filters items; an `if` at the **front** needs an `else` and chooses a
  value.
- Curly braces with a colon build a dictionary, without one build a set.
- Several `for` clauses read in the same order as the nested loops they replace.
- Write the loop when the comprehension needs more than one condition, will not fit on a line,
  or nests another inside it.
- Parentheses build a **generator**: single use, almost no memory, ideal for feeding `sum` or a
  loop.
- The loop variable does not escape a comprehension, unlike a `for` loop's.
