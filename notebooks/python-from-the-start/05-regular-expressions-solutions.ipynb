# Regular Expressions &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
import re

sample = "Order A-1234 shipped 2026-03-15, order B-99 shipped 2026-04-02"
```

**1.** Print every run of digits in `sample`.


```python
print(re.findall(r"\d+", sample))
```

`\d+` takes one or more digits at a time. Without the `+` you would get every digit
separately.


**2.** Print every date in `sample`, matching four digits, a dash, two digits, a dash, two
digits.


```python
print(re.findall(r"\d{4}-\d{2}-\d{2}", sample))
```

The shape alone is enough to separate dates from the order numbers, and nothing in the pattern
names a particular date.


**3.** Print every order code: one capital letter, a dash, then one or more digits.


```python
print(re.findall(r"[A-Z]-\d+", sample))
```

`[A-Z]` is a range covering the capitals. `A-1234` and `B-99` have different numbers of digits,
which is exactly why `+` is used rather than a fixed count.


**4.** Use `re.sub` to mark every date in `sample` by wrapping it in square brackets, and
print the result.


```python
print(re.sub(r"\d{4}-\d{2}-\d{2}", r"[\g<0>]", sample))
```

`\g<0>` is the whole matched text, so the replacement puts the date back with brackets
around it.


**5.** Use `re.search` with named groups to pull the year, month and day out of
`"2026-03-15"`, then print them on one line separated by slashes.


```python
m = re.search(r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})", "2026-03-15")

print(m.group("year") + "/" + m.group("month") + "/" + m.group("day"))
print(f"{m.group('year')}/{m.group('month')}/{m.group('day')}")   # the same, as an f-string
```

Both lines produce the same text. The f-string version is easier to read once there are more
than two pieces.


**6.** `re.search(r"ZZZ", sample)` finds nothing. Write code that prints `not found`
instead of raising an error.


```python
m = re.search(r"ZZZ", sample)

if m:
    print(m.group())
else:
    print("not found")
```

`re.search` returns `None` when nothing matches, and `None` is treated as false in an `if`,
which notebook 6 explains. Testing the match object before calling `.group()` on it is the
habit that avoids the most common regex error there is.
