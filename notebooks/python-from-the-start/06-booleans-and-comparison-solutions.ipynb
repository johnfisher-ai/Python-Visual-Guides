**1.** Create `score = 87`. Print `True` or `False` for whether it is between 80 and 90,
using a chained comparison.


```python
score = 87

print(80 <= score <= 90)
```

`<=` on both ends includes 80 and 90 themselves. With `<` on both ends a score of exactly 90
would come back `False`, so the choice matters and is worth making deliberately.


**2.** Print `bool()` of each of these on separate lines: `0`, `""`, `" "`, `[]`,
`"False"`. Two of them are true; be sure you can say which before you run it.


```python
print(bool(0))
print(bool(""))
print(bool(" "))
print(bool([]))
print(bool("False"))
```

The two true ones are `" "` and `"False"`. Both are strings with at least one character in
them, and neither is empty. The word `False` written as text has nothing to do with the boolean
`False`.


**3.** Create `username = ""`. Using `or`, produce a name that falls back to `"guest"`
when the username is empty, and print it.


```python
username = ""
display_name = username or "guest"

print(display_name)
```

`or` gives back the first truthy value. The empty string is falsy, so `"guest"` is what comes
through. Set `username` to something and the same line hands back that instead.


**4.** Create two equal lists with different names, and a third name pointing at the first
list. Print the results of `==` and `is` for both pairs.


```python
first = [1, 2, 3]
second = [1, 2, 3]
also_first = first

print(first == second, first is second)
print(first == also_first, first is also_first)
```

`first` and `second` hold equal values in two separate lists, so `==` is `True` and `is` is
`False`. `also_first` is a second label on the one list, so both are `True`.


**5.** Create `middle_name = None`. Print `"none given"` if it is `None`, using the `is`
test rather than `==`.


```python
middle_name = None

if middle_name is None:
    print("none given")
else:
    print(middle_name)
```

`is None` is the standard way to write this. `== None` happens to work here, but `is` asks the
question you actually mean: is this the one and only `None`.


**6.** Given `words = ["yes", "no", "yes", "yes"]`, count how many are `"yes"` in a
single line, using the fact that `True` counts as `1`.


```python
words = ["yes", "no", "yes", "yes"]

print(sum(w == "yes" for w in words))
```

Each `w == "yes"` is `True` or `False`, and `sum` adds them as `1` and `0`.

A loop does the same thing and is just as good while the syntax above is still unfamiliar:


```python
count = 0
for w in words:
    if w == "yes":
        count = count + 1

print(count)
```

`words.count("yes")` also works and is the shortest of the three. Lists are notebook 7.
