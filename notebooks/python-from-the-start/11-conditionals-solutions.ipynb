# Conditionals &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** Set `hour = 14`. Print `"Good morning"` if the hour is before 12, and
`"Good afternoon"` otherwise.


```python
hour = 14

if hour < 12:
    print("Good morning")
else:
    print("Good afternoon")
```

Two outcomes, so `if` and `else` and nothing more. `else` needs no condition of its own.


**2.** Set `temperature` to any number. Print `"freezing"` below 0, `"cold"` below 15,
`"mild"` below 25, and `"hot"` otherwise.


```python
temperature = 18

if temperature < 0:
    print("freezing")
elif temperature < 15:
    print("cold")
elif temperature < 25:
    print("mild")
else:
    print("hot")
```

The order is what makes this correct. Because Python stops at the first true condition, by the
time `temperature < 15` is tested the value is already known to be 0 or above, so no lower bound
is needed on any line.

Written the other way round, with `temperature < 25` first, every cold reading would be
reported as mild.


**3.** Set `basket = []`. Print `"Your basket is empty"` if it is, and the number of items
if it is not. Test the condition on the list itself.


```python
basket = []

if basket:
    print(len(basket), "items")
else:
    print("Your basket is empty")
```

An empty list is falsy, so `if basket:` is the whole test. `if len(basket) > 0:` gives the same
answer and says more than it needs to.


**4.** Set `age = 30` and `member = True`. Print `"Discount"` only if the person is over
25 **and** a member.


```python
age = 30
member = True

if age > 25 and member:
    print("Discount")
```

`member` is already `True` or `False`, so it needs no comparison. Writing `member == True` works
and is the long way round.

There is no `else` here because the task did not ask for one. An `if` on its own is fine.


**5.** Rewrite the nested code as a flat chain of guard clauses that gives the same
answers.


```python
password = "abc"

if not password:
    result = "missing"
elif len(password) < 8:
    result = "too short"
else:
    result = "ok"

print(result)
```

The nested version asked "is there a password? then is it long enough?" and buried the success
case two levels in. This asks the same two questions as reasons to stop, in order, and lets the
good outcome fall through to the end.

Note that both conditions flipped. The nested version tested `if password:` and `if len(...) >=
8:`; the flat version tests for the problems instead. That is the shape of a guard clause.


**6.** Set `n` to any number, and use a one-line conditional to put `"even"` or `"odd"`
into a name called `parity`, then print it.


```python
n = 7

parity = "even" if n % 2 == 0 else "odd"

print(n, "is", parity)
```

`n % 2` is the remainder from **Numbers**, and it is `0` for every even number.

The same thing written as a full `if` takes four lines and is equally correct. The one-line
form is worth using when both branches do nothing but choose a value.
