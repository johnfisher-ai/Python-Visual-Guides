# Dictionaries &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** Make a dictionary `capitals` with three countries as keys and their capital cities
as values. Print it and print one lookup.


```python
capitals = {
    "France": "Paris",
    "Japan": "Tokyo",
    "Peru": "Lima",
}

print(capitals)
print(capitals["Japan"])
```

Country to capital is a natural fit: you always know the country and want the city, never
the other way round by position.


**2.** Add a fourth country, change one of the existing values, and print the result.


```python
capitals["Kenya"] = "Nairobi"     # a new key, so this adds
capitals["Peru"] = "LIMA"         # an existing key, so this replaces

print(capitals)
```

Both lines are the same syntax. Whether it adds or replaces depends only on whether the key
was already there.


**3.** Print the capital of a country that is **not** in your dictionary, without raising
an error, showing `"unknown"` instead.


```python
print(capitals.get("Brazil", "unknown"))
```

`capitals["Brazil"]` would raise a `KeyError`. The second argument to `get` is what comes
back instead.


**4.** Loop over `capitals` printing a line like `France -> Paris` for each pair. Use
`.items()`.


```python
for country, city in capitals.items():
    print(f"{country} -> {city}")
```

`.items()` hands over a tuple each time round, and the two names unpack it, which is the
pattern from the **Tuples and Unpacking** notebook applied here.


**5.** Given `letters = "mississippi"`, count how many times each letter appears and print
the dictionary.


```python
letters = "mississippi"

counts = {}
for letter in letters:
    counts[letter] = counts.get(letter, 0) + 1

print(counts)
```

Looping over a string gives one character at a time, which the **Loops** notebook covers properly. The
counting line is the same idiom as the word count in the notebook: current value or zero, plus
one.


**6.** Using your counts from task 5, print the most common letter and how many times it
appeared.


```python
top = max(counts, key=counts.get)

print(top, "appears", counts[top], "times")
```

`max(counts, key=counts.get)` walks the keys and compares them by their values, so it returns
the key rather than the count. `max(counts)` on its own would give the alphabetically last
letter instead, which is a different question.

Sorting works too when you want more than the top one:


```python
for letter, n in sorted(counts.items(), key=lambda pair: pair[1], reverse=True):
    print(letter, n)
```
