**1.** Create a list `pets` holding `"dog"`, `"cat"` and `"fish"`. Print its length,
then print the last item without using the number 2.


```python
pets = ["dog", "cat", "fish"]

print(len(pets))
print(pets[-1])
```

`pets[-1]` keeps working when the list grows or shrinks. `pets[2]` is correct only while the
list has exactly three items, which is why it is the more fragile way to write it.


**2.** Add `"bird"` to the end of `pets`, insert `"hamster"` at the front, and print the
result.


```python
pets.append("bird")
pets.insert(0, "hamster")

print(pets)
```

Neither line is assigned to anything. Both change `pets` in place, and both return `None`,
so assigning would have thrown the list away.


**3.** Given `scores = [88, 72, 95, 64, 91]`, print the highest, the lowest, and the mean
rounded to one decimal place.


```python
scores = [88, 72, 95, 64, 91]

print("highest:", max(scores))
print("lowest: ", min(scores))
print("mean:   ", round(sum(scores) / len(scores), 1))
```

The mean is the sum over the count. `round(x, 1)` from notebook 3 keeps one decimal place
for display.


**4.** Given `words = ["delta", "Alpha", "charlie"]`, print them sorted so that capitals
do not come first, without changing `words` itself. Print `words` afterward to prove it is
unchanged.


```python
words = ["delta", "Alpha", "charlie"]

print(sorted(words, key=str.lower))
print(words)
```

`sorted` returns a new list and leaves the original alone, which is exactly what the task asked
for. `words.sort(key=str.lower)` would also sort correctly, but it would change `words`, and
then the second line could not prove anything.


**5.** Create a list, make a real copy of it, change the copy, and print both to show the
original did not move.


```python
original = ["a", "b", "c"]
duplicate = original.copy()

duplicate.append("d")

print("original: ", original)
print("duplicate:", duplicate)
```

`original[:]` and `list(original)` do the same thing. What would **not** work is
`duplicate = original`, which gives a second name for the one list, and then both lines would
print four items.


**6.** Given `numbers = [1, 2, 3, 4, 5, 6]`, build a **new** list holding only the even
ones, without changing `numbers`. A loop is fine.


```python
numbers = [1, 2, 3, 4, 5, 6]
evens = []

for n in numbers:
    if n % 2 == 0:
        evens.append(n)

print("evens:  ", evens)
print("numbers:", numbers)
```

Building a separate list is the safe habit. Removing the odd ones from `numbers` while looping
over it would have given a wrong answer and no error, which is the trap the notebook shows.

The comprehension version says the same thing in one line, and is notebook 13:


```python
print([n for n in numbers if n % 2 == 0])
```
