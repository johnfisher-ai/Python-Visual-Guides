# Where Next &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
rows = [
    "2026-03-01,north,18.5",
    "2026-03-01,south,22.1",
    "2026-03-02,north,",
    "2026-03-02,south,23.4",
    "bad line with no commas",
    "2026-03-03,north,19.2",
]


def parse(line):
    parts = line.split(",")
    if len(parts) != 3:
        return None
    date, region, value = parts
    if not value.strip():
        return None
    return {"date": date, "region": region, "value": float(value)}


records = [r for r in (parse(line) for line in rows) if r is not None]
print(len(records), "records")
```

**1.** Print the lines that `parse` rejects, and why each one was rejected.


```python
for line in rows:
    parts = line.split(",")
    if len(parts) != 3:
        print(f"{line!r} -> wrong number of fields ({len(parts)})")
    elif not parts[2].strip():
        print(f"{line!r} -> no value")
```

The two conditions match the two guard clauses inside `parse`. Repeating them here is fine for
a one-off check; if this were code you kept, `parse` would return the reason rather than `None`
so there is only one copy of the rule.


**2.** Print how many distinct regions appear in `records`.


```python
regions = {r["region"] for r in records}

print(regions)
print(len(regions), "distinct regions")
```

A set comprehension, so the duplicates collapse without being asked. `len` on the set is the
answer.


**3.** Print the record with the highest value, using `max` with a `key`.


```python
highest = max(records, key=lambda r: r["value"])

print(highest)
print(f"{highest['region']} on {highest['date']}: {highest['value']}")
```

Without the `key`, `max` would try to compare dictionaries against each other and raise. The
`lambda` tells it which part to compare, exactly as it did for sorting in the **Functions**
notebook.


**4.** Group by `date` instead of by region, and print the mean for each date.


```python
by_date = {}
for r in records:
    by_date.setdefault(r["date"], []).append(r["value"])

for date, values in sorted(by_date.items()):
    print(f"{date}  n={len(values)}  mean={sum(values) / len(values):.2f}")
```

Only the key changed, from `r["region"]` to `r["date"]`. That similarity is what the next
task is about.


**5.** Write `summarize(records, field)` that groups by any field. Call it for `"region"` and
for `"date"`.


```python
def summarize(records, field):
    groups = {}
    for r in records:
        groups.setdefault(r[field], []).append(r["value"])

    print(f"by {field}:")
    for key, values in sorted(groups.items()):
        print(f"  {key:<12} n={len(values)}  mean={sum(values) / len(values):.2f}")


summarize(records, "region")
summarize(records, "date")
```

The two versions in tasks 4 and 5 differed by one string, which is the signal that a parameter
was wanted. Passing the field name in turns two near-identical blocks into one function that
answers both questions, and any similar question later.


**6.** Add a line with a value that is not a number, and make `parse` reject it instead of
raising.


```python
rows.append("2026-03-04,north,warm")


def parse(line):
    parts = line.split(",")
    if len(parts) != 3:
        return None

    date, region, value = parts
    if not value.strip():
        return None

    try:
        number = float(value)
    except ValueError:
        return None

    return {"date": date, "region": region, "value": number}


records = [r for r in (parse(line) for line in rows) if r is not None]

print(f"kept {len(records)} of {len(rows)} lines")
```

A third guard, and this one has to be a `try` rather than an `if`, because the only reliable way
to find out whether text is a number is to attempt the conversion.

Note that the count still reports how many of how many. Adding a rejected line changed the
denominator and not the answer, and the output says so.
