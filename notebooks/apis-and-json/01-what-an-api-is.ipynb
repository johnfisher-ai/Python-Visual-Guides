# What an API Is


## What you will be able to do

Explain what a web API is, read an HTTP request and its response line by line, and send a request
from Python and take the answer apart into its status, its headers and its JSON body.


## The idea

### The problem

Yesterday's temperature in Tromso, this morning's exchange rates, the open issues in a GitHub
repository: all of it is data held by somebody else's program, on somebody else's computer. You
cannot open their database, and you cannot import their code.

You could download the web page a person reads and pick the numbers out of its HTML. People did,
for years, and every redesign of the page broke every program reading it. The page was made for
eyes, and nothing promised a program that a number would stay where it was.

What a program needs is a way to ask another program a question in a form that program has agreed
to answer, and to get back data rather than decoration.

### What an API is

> An **API**, an application programming interface, is the set of requests a program agrees to
> accept and the responses it promises to send back. A **web API** carries them over **HTTP**, the
> protocol a browser uses to fetch pages. The client sends a **request** that names a **method**
> and a **resource**, and the server sends back a **response** carrying a **status code**,
> **headers** and a **body**, which for most APIs is JSON.

### Why it works that way

HTTP is text. A request is a line saying what you want, usually the method `GET` and the path of a
resource, then a few header lines of the form `Name: value`, then an empty line, then an optional
body. The response has the same shape: a status line such as `HTTP/1.1 200 OK`, headers, an empty
line, and the body. Every HTTP tool, from a browser to the `requests` library, is a convenience for
writing and reading that text.

Two consequences follow. The server never learns what language the client was written in, only
what text arrived, which is why a single API can serve Python, JavaScript and a phone app alike.
And every request stands alone: the server answers what it was asked, and nothing carries over to
the next request unless a header carries it.

The URL packs the address into one string: a scheme (`http` or `https`), a host, sometimes a port,
the path to the resource, and an optional query after a `?`.

The status code is three digits, and the first digit says how it went: `2xx` means it worked, `4xx`
means the request was wrong, and `5xx` means the server failed. The **Status Codes** notebook takes
them apart.

### Where you will meet this

Weather services, payment providers, cloud storage, GitHub and the language models behind AI tools
are all used through web APIs. `pip install` itself speaks HTTP to the Python Package Index. In
this library, the **Projects, End to End** guide includes an API project that pulls a live feed and
stores what arrives.

## Setup

Six imports, the last of them the practice API.

- `json` turns a JSON body into Python values
- `socket` opens a raw connection, so a request and its response can be seen as the text they are
- `urllib.request` sends requests the ordinary way, and fetches the practice API's code when this
  notebook runs in Colab
- `Path` checks whether that code is already beside the notebook
- `urlsplit` takes a URL apart into its pieces
- `practice_api` is the server this guide talks to. `start()` runs it in the background, inside

`HTTPError` is imported later, in Common errors, the only section that uses it.


````python
import json
import socket
import urllib.request
from pathlib import Path
from urllib.parse import urlsplit

PRACTICE_API = "https://raw.githubusercontent.com/johnfisher-ai/Python-Visual-Guides/main/notebooks/apis-and-json/practice_api.py"

if not Path("practice_api.py").exists():      # true in Colab, which starts with only the notebook
    urllib.request.urlretrieve(PRACTICE_API, "practice_api.py")

import practice_api

BASE = practice_api.start()
print("The practice API is running at", BASE)
````

## Worked examples

### The parts of a URL

`urlsplit` takes a URL apart without sending anything. Here is the address of a station on the
practice API.


````python
parts = urlsplit(f"{BASE}/stations/tromso")

print("scheme:", parts.scheme)
print("host:  ", parts.hostname)
print("port:  ", parts.port)
print("path:  ", parts.path)
````

`127.0.0.1` is the address a computer uses to reach itself, and `8765` is the port the practice API
listens on, so this URL leads to the server Setup started. The path names the resource: the
station whose id is `tromso`.

A real service's URL has the same parts, and usually a query as well. This one asks Open-Meteo's
archive for three days of Tromso's weather.


````python
archive = ("https://archive-api.open-meteo.com/v1/archive"
           "?latitude=69.65&longitude=18.96&start_date=2025-01-15&end_date=2025-01-17"
           "&daily=temperature_2m_mean&models=era5")
parts = urlsplit(archive)

print("scheme:", parts.scheme)
print("host:  ", parts.hostname)
print("port:  ", parts.port)
print("path:  ", parts.path)
print("query: ", parts.query)
````

No port is written, so the scheme's standard port applies: 443 for `https`, 80 for `http`. `https`
is the same HTTP carried over an encrypted connection, so everything this notebook shows about
requests and responses holds for it too.

The query is the part after `?`: `name=value` pairs joined by `&`. Writing it by hand works here,
and the **Query Parameters** notebook shows why real code builds it instead.

### A request is text

A raw connection sends exactly the bytes you give it and returns exactly the bytes that arrive,
with no library in between. `exchange` below does that and nothing more: it connects to the
practice API, sends a request, and reads until the server hangs up.

The request is three lines and an empty line. Every line ends in `\r\n`, a carriage return and a
line feed, which is what HTTP requires, and the empty line tells the server the headers are over.
`Host` names the server being asked, and `Connection: close` asks it to hang up after answering,
which is how `exchange` knows the response is complete.


````python
host, port = urlsplit(BASE).hostname, urlsplit(BASE).port


def exchange(request):
    """Send a request as raw text, and return the raw text of the response."""
    with socket.create_connection((host, port)) as conn:
        conn.sendall(request.encode("utf-8"))
        return conn.makefile(encoding="utf-8").read()


lines = ["GET /stations/tromso HTTP/1.1", f"Host: {host}:{port}", "Connection: close"]
raw = exchange("\r\n".join(lines) + "\r\n\r\n")
print(raw)
````

Read the response from the top:

- `HTTP/1.1 200 OK` is the **status line**: the protocol version, the status code, and a reason
  phrase for people.
- `Server`, `Date`, `Content-Type` and `Content-Length` are **headers**. `Content-Type` says the
  body is JSON, and `Content-Length` says how many bytes of it to expect.
- The empty line ends the headers.
- Everything after it is the **body**.

The practice API always reports the same `Date`, so this output matches yours. A real server
reports the moment it answered.

### Taking a response apart

Every client library begins with this step. The headers end at the first empty line, and everything
after it is the body. Reading the connection as text turned every `\r\n` into `\n`, which is why
the split below is on `"\n\n"`.


````python
head, body = raw.split("\n\n", 1)
status_line, *header_lines = head.splitlines()
headers = dict(line.split(": ", 1) for line in header_lines)

print("status line:", status_line)
print("headers:    ", headers)
print("body:       ", body)
````

The body is still a string. `json.loads`, from the **JSON on Disk** notebook in the **Files, Paths
and Formats** guide, turns it into Python values.


````python
station = json.loads(body)

print(type(station).__name__, station)
print(station["name"], "is at", station["latitude"], station["longitude"])
print("Content-Length said", headers["Content-Length"], "and the body is", len(body.encode()), "bytes")
````

`Content-Length` counts bytes, not characters, which is why the comparison encodes the body first.
Here the two agree because every character is ASCII. A body containing `ø` would have more bytes
than characters.

### The same conversation through `urllib.request`

Real code does not write requests by hand. `urllib.request`, part of Python's standard library,
writes the request text, reads the response and takes it apart, and `urlopen` hands back an object
carrying the pieces. The connection closes at the end of the `with` block, as a file would.


````python
with urllib.request.urlopen(f"{BASE}/stations/tromso") as response:
    print("status: ", response.status, response.reason)
    print("type:   ", response.headers["Content-Type"])
    body = response.read()

print("body:   ", body)
print("as dict:", json.loads(body))
````

`read` returns bytes, not a string, because a body can as easily be an image as text. `json.loads`
accepts bytes and decodes them as UTF-8, so the body goes straight in.

### Asking for something that is not there

A request for a station that does not exist still gets a complete, well-formed response. The answer
is simply no.


````python
lines = ["GET /stations/atlantis HTTP/1.1", f"Host: {host}:{port}", "Connection: close"]
print(exchange("\r\n".join(lines) + "\r\n\r\n"))
````

The status line says `404 Not Found`, and the body says why, in JSON, so a program can read the
reason as well as the code. The connection worked and the server did its job; what failed was the
request. `urlopen` reports this differently, as an exception, which Common errors shows.

### A collection, and its members

Most APIs are laid out the same way. A plural noun names a collection, and an id after it names a
member of that collection. The collection gives a summary of every member.


````python
with urllib.request.urlopen(f"{BASE}/stations") as response:
    stations = json.loads(response.read())

print(stations)
````

The detail for a station lives at its own address, so reading all of it takes a request per
station.


````python
for summary in stations:
    with urllib.request.urlopen(f"{BASE}/stations/{summary['id']}") as response:
        station = json.loads(response.read())
    print(f"{station['name']:<9} {station['latitude']:>6} {station['longitude']:>6}")
````

Four stations, five requests: the collection, then a member at a time. A short summary list plus a
request per detail is the layout you will meet most often, and the **Pagination** notebook deals
with collections too long to send at once.

### A real API: Open-Meteo

The practice API answers the same way for every reader, which is what a notebook needs. A real
service does the same job over the internet. Open-Meteo publishes weather data free for
non-commercial use, with no key to sign up for, and its archive holds past days. Asking for the
ERA5 model by name fixes which data answers, and a day in January 2025 is long settled, so you will
get the same numbers as below.

Weather data by [Open-Meteo.com](https://open-meteo.com/), under
[CC BY 4.0](https://creativecommons.org/licenses/by/4.0/), from the ERA5 reanalysis. Generated using
Copernicus Climate Change Service information 2026.


````python
with urllib.request.urlopen(archive, timeout=30) as response:
    print("status:", response.status, response.reason)
    print("type:  ", response.headers["Content-Type"])
    weather = json.loads(response.read())

print("keys:  ", sorted(weather))
````

The same shape as the practice API's answers: a status, headers, and a JSON body. `generationtime_ms`
is how long the server spent on the request, so it changes every time, which is why this cell
prints the keys and not the whole reply.

`timeout=30` stops `urlopen` waiting forever for a slow server. The **Errors and Retries** notebook
covers what to do when the wait runs out.


````python
daily = weather["daily"]
unit = weather["daily_units"]["temperature_2m_mean"]

for day, temperature in zip(daily["time"], daily["temperature_2m_mean"]):
    print(day, temperature, unit)
````

A dictionary of lists, matched by position: `time[0]` goes with `temperature_2m_mean[0]`. `zip`
pairs them, as the **Loops** notebook showed. The **JSON in a Response** notebook reaches into
deeper structures than this.

### Two APIs, one question

The practice API knows where every station is. Open-Meteo knows what the weather was there. Asking
both, a station at a time, answers a question neither can answer alone.


````python
def mean_temperatures(latitude, longitude):
    """Three daily mean temperatures for one place, from Open-Meteo's archive."""
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={latitude}&longitude={longitude}"
           "&start_date=2025-01-15&end_date=2025-01-17&daily=temperature_2m_mean&models=era5")
    with urllib.request.urlopen(url, timeout=30) as response:
        return json.loads(response.read())["daily"]["temperature_2m_mean"]


with urllib.request.urlopen(f"{BASE}/stations") as response:
    stations = json.loads(response.read())

for summary in stations:
    with urllib.request.urlopen(f"{BASE}/stations/{summary['id']}") as response:
        station = json.loads(response.read())
    print(f"{station['name']:<9}", mean_temperatures(station["latitude"], station["longitude"]))
````

### Where each part came from

| In the loop | What it relies on | The section that showed it |
|---|---|---|
| `f"{BASE}/stations"`, then `/stations/{id}` | a collection, and a member at its own address | A collection, and its members |
| `urlopen` in a `with` block | `urllib` writes the request and reads the response | The same conversation through `urllib.request` |
| `json.loads(response.read())` | the body arrives as bytes of JSON | The same conversation through `urllib.request` |
| `?latitude=...&longitude=...` | the query is part of the URL | The parts of a URL |
| `timeout=30` | a real server can be slow to answer | A real API: Open-Meteo |
| `["daily"]["temperature_2m_mean"]` | the reply is a dictionary of lists | A real API: Open-Meteo |

Nine requests to two servers, all made with the same `urlopen`. The client does not need to know
who is answering, only the address and the shape of the reply.

## Common errors

### URLError: unknown url type, when the scheme is missing

A URL without `http://` looks harmless, especially when it is an address and a port.


````python
urllib.request.urlopen("127.0.0.1:8765/stations")
````

`unknown url type: 127.0.0.1` looks like nonsense until you see what `urllib` did: it read
everything before the first colon as the scheme, so it went looking for a protocol called
`127.0.0.1`. Every URL passed to `urlopen` needs its scheme. `BASE` starts with `http://`, which is
why building on it avoids this.


````python
with urllib.request.urlopen(f"{BASE}/stations") as response:
    print(response.status, response.reason)
````

### URLError: Connection refused, when nothing is listening


````python
urllib.request.urlopen("http://127.0.0.1:8799/stations")
````

Read the last line first: `Connection refused`. The computer at `127.0.0.1` answered, and said no
program was listening on port `8799`, so the connection failed before any HTTP was sent. The
traceback is long because the error passed through several layers on its way out: the first
exception, `ConnectionRefusedError`, comes from the connection itself, and `urlopen` raised the
`URLError` below it while handling that.

The number in brackets is the operating system's code for the failure: 61 on macOS, where this
notebook's output was recorded, and 111 on Linux, which is what Colab runs. The words are the same.

The usual causes are a mistyped port and a server that is not running, which here would mean Setup
was not run. Print the address you meant to use, and compare:


````python
print("the practice API is at", BASE)

with urllib.request.urlopen(f"{BASE}/stations") as response:
    print(response.status, response.reason)
````

### HTTPError: HTTP Error 404, which urlopen raises

Asking for something that is not there showed this request getting a complete answer. `urlopen`
sees it differently.


````python
urllib.request.urlopen(f"{BASE}/stations/atlantis")
````

`urlopen` turns any status of 400 or above into an exception. The exception carries the response,
so catch it and read the status and the body from it. `HTTPError` lives in `urllib.error`:


````python
from urllib.error import HTTPError

try:
    urllib.request.urlopen(f"{BASE}/stations/atlantis")
except HTTPError as error:
    print("status:", error.code, error.reason)
    print("body:  ", json.loads(error.read()))
````

The `requests` library in the **Your First Request** notebook makes the opposite choice: a `404`
comes back as an ordinary response, and raises only when you ask it to.

### JSONDecodeError: the body was HTML, not JSON

The practice API's home page, at `BASE` itself, is written for people.


````python
with urllib.request.urlopen(BASE) as response:
    body = response.read()

json.loads(body)
````

`Expecting value: line 1 column 1 (char 0)` means the very first character was not the start of
any JSON value. It was `<`, the start of an HTML tag. Real code meets this when a URL is slightly
wrong and the server answers with a web page, or when a proxy between you and the server returns an
error page of its own. Check `Content-Type` before parsing:


````python
with urllib.request.urlopen(BASE) as response:
    kind = response.headers["Content-Type"]
    body = response.read()

if kind.startswith("application/json"):
    print(json.loads(body))
else:
    print("not JSON:", kind)
    print(body.decode().splitlines()[0])
````

## Recap

- An API is the set of requests a program agrees to accept and the responses it promises to send
  back. A web API carries them over HTTP.
- An HTTP request is text: a request line, headers, an empty line, and an optional body. The
  response has the same shape and starts with a status line.
- A URL packs a scheme, a host, a port, a path and a query into one string, and `urlsplit` takes it
  apart.
- `urllib.request.urlopen` sends a request and returns the status, the headers and the body as
  bytes. `json.loads` turns a JSON body into Python values.
- Most APIs name a collection with a plural noun, and a member with an id after it.
- A `404` is a complete answer from a working server. `urlopen` raises `HTTPError` for it, and the
  exception carries the response.
- `Connection refused` means nothing was listening on that port, so no HTTP was spoken at all.
- Check `Content-Type` before parsing a body as JSON.
