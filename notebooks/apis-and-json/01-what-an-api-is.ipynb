# What an API Is


## What you will be able to do

An API is the set of requests a program agrees to accept and the responses it promises to send back.
Send a request to the practice API from Python, read the exchange as plain text, and take the
response apart into its status, headers and body.

## The idea

### The problem

Yesterday's temperature in Tromso, this morning's exchange rates and the open issues in a GitHub
repository are all data held by somebody else's program, on somebody else's computer. You cannot
open their database, and you cannot import their code.

You could download the web page a person reads and pick the numbers out of its HTML. People did,
for years, and every redesign of the page broke every program reading it. The page was made for
eyes, and nothing promised a program that a number would stay where it was.

What a program needs is a way to ask another program a question in a form that program has agreed
to answer, and to receive data rather than decoration.

### What an API is

> An **API**, an application programming interface, is the set of requests a program agrees to
> accept and the responses it promises to send back. A **web API** carries them over **HTTP**, the
> protocol a browser uses to fetch pages. The client sends a **request** that names a **method**
> and a **resource**, and the server returns a **response** carrying a **status code**,
> **headers** and a **body**, which for most APIs is JSON.

### The vocabulary of an API

Every call to a web API is a request from the client and a response from the server, and the rest
of the vocabulary names their parts. This section is the map. The notebooks of this guide take the
parts in turn, and the last column of each table says where.

A **request** carries up to four things:

| Part | What it is | Example | Covered in depth in |
|---|---|---|---|
| Method | the verb: what the client asks the server to do | `GET` | **Sending Data** |
| URL | the address of the resource, with any query parameters | `/stations/tromso` | this notebook, and **Query Parameters** |
| Headers | extra information about the request, as `Name: value` pairs | `Accept: application/json` | **Headers and Content Types** |
| Body | data the client sends, when the method sends any | `{"name": "Narvik"}` | **Sending Data** |

A **response** carries three:

| Part | What it is | Example | Covered in depth in |
|---|---|---|---|
| Status code | three digits saying how the request went, sometimes called a response code | `200 OK`, `404 Not Found` | **Status Codes** |
| Headers | extra information about the response | `Content-Type: application/json` | **Headers and Content Types** |
| Body | the data itself, most often JSON | `{"id": "tromso", ...}` | **JSON in a Response**, and **Schemas and Validation** |

An address an API responds at is called an **endpoint**. The practice API in this guide has a
`/stations` endpoint for the whole collection of weather stations, and a `/stations/{id}` endpoint
for a single station, where `{id}` stands for the station's id.

The **methods**, also called verbs, say what a request is for:

| Method | Asks the server to | For example |
|---|---|---|
| `GET` | send back a resource, changing nothing | read a station |
| `POST` | create a new resource, or carry out an action | add a station |
| `PUT` | replace a resource with the body sent | replace a station's whole record |
| `PATCH` | change part of a resource | correct a station's latitude |
| `DELETE` | remove a resource | remove a station |

`GET` is by far the most common, and most requests in this notebook use it. The **Sending Data**
notebook takes up the other four, including why repeating a `PUT` or a `DELETE` is harmless while
repeating a `POST` is not.

**Parameters** are the values that make a request specific, and each kind travels in a different
place:

| Kind | Where it travels | Example | Covered in depth in |
|---|---|---|---|
| Path parameter | inside the path, naming a resource | `tromso` in `/stations/tromso` | this notebook |
| Query parameter | after `?` in the URL, as `name=value` pairs joined by `&` | `?start_date=2025-01-15` | **Query Parameters** |
| Header | among the request's headers | `Authorization: Bearer <token>` | **Authentication** |
| Body | in the body of a `POST`, `PUT` or `PATCH` request | `{"latitude": 69.65}` | **Sending Data** |

Status codes fall into five classes, and the first digit says which:

| Class | Means | Codes you will meet |
|---|---|---|
| `1xx` | information only, rarely seen by client code | `100 Continue` |
| `2xx` | the request worked | `200 OK`, `201 Created`, `204 No Content` |
| `3xx` | the resource is at another address, which clients usually follow on their own | `301 Moved Permanently`, `302 Found` |
| `4xx` | the request was wrong | `400 Bad Request`, `401 Unauthorized`, `404 Not Found`, `405 Method Not Allowed`, `429 Too Many Requests` |
| `5xx` | the server failed | `500 Internal Server Error`, `503 Service Unavailable` |

The **Status Codes** notebook takes them apart, the **Rate Limits** notebook deals with `429`, and
the **Errors and Retries** notebook covers which failures are worth trying again.

A few more terms come up often enough to name now:

- An API's *documentation* lists its endpoints, their parameters and their responses. Many APIs
  also publish it in a machine-readable form, an *OpenAPI* document, which tools such as Postman
  can load. The **Exploring an API** notebook reads both kinds.
- A *schema* describes the shape a body must have, and checking a body against its schema is
  called *validation*, the subject of the **Schemas and Validation** notebook.
- An *API key* or *token* identifies who is calling, the subject of the **Authentication**
  notebook.
- A long answer is often split into pages, called *pagination*, the subject of the
  **Pagination** notebook.
- A *rate limit* caps how many requests a client may send in a period, the subject of the
  **Rate Limits** notebook.

### Why it works that way

HTTP messages are plain text, which the first look below shows directly. A request starts with a
request line naming the method, the path and the protocol version. Its headers follow, a blank line
ends them, and the body comes after when there is one. A response has the same layout, starting
with a status line. Every HTTP tool, from a browser to the `requests` library, is a convenience for
writing and reading these messages.

One consequence is that the server never learns what language the client was written in, only what
request arrived, which is why a single API can serve Python, JavaScript and a phone app alike.
Another is that every request stands alone. The server responds to what it was asked, and nothing
carries over to the next request unless a header carries it.

### Where this shows up

Weather services, payment providers, cloud storage, GitHub and the language models behind AI tools
are all used through web APIs. `pip install` itself speaks HTTP to the Python Package Index. In
this library, the **Projects, End to End** guide includes an API project that pulls a live feed and
stores what arrives.

## Worked examples

### The parts of a URL

`urlsplit` takes a URL apart without sending anything. Here is the address of a station on the
practice API.


````python
parts = urlsplit(f"{BASE}/stations/tromso")

print("scheme:", parts.scheme)
print("host:  ", parts.hostname)
print("port:  ", parts.port)
print("path:  ", parts.path)
````

`127.0.0.1` is the address a computer uses to reach itself, and `8765` is the port the practice API
listens on, so this URL leads to the server Setup started. The path names the resource. It matches
the endpoint `/stations/{id}`, with `tromso` in place of `{id}`: `tromso` is a **path parameter**.

A real service's URL has the same parts, and usually query parameters as well. This one asks
Open-Meteo's archive for three days of Tromso's weather.


````python
archive = ("https://archive-api.open-meteo.com/v1/archive"
           "?latitude=69.65&longitude=18.96&start_date=2025-01-15&end_date=2025-01-17"
           "&daily=temperature_2m_mean&models=era5")
parts = urlsplit(archive)

print("scheme:", parts.scheme)
print("host:  ", parts.hostname)
print("port:  ", parts.port)
print("path:  ", parts.path)
print("query: ", parts.query)
````

No port is written, so the scheme's standard port applies: 443 for `https`, 80 for `http`. `https`
is the same HTTP carried over an encrypted connection, so everything this notebook shows about
requests and responses holds for it too.

The query holds the **query parameters**. `parse_qs` splits it into their names and values:


````python
for name, values in parse_qs(parts.query).items():
    print(f"{name:<11} {values[0]}")
````

Six query parameters choose the place, the dates, the measurement and the model. `parse_qs` gives
a list of values for every name, because a query may repeat a name, which is why the loop prints
`values[0]`. Typing a query by hand works here, and the **Query Parameters** notebook shows why real
code builds it from a dictionary instead.

### A request and its response over a raw connection

A raw connection sends exactly the bytes you give it and returns exactly the bytes the server sends
back, with no library in between. `exchange` below does that and nothing more. It connects to the
practice API, sends a request, and reads the response until the server closes the connection.

The request has a request line, two headers, and the blank line that ends the headers. The request
line gives the method, `GET`, then the path, then the protocol version, and a `GET` request has no
body. Every line ends in `\r\n`, a carriage return and a line feed, as HTTP requires. `Host` names
the server being asked, and `Connection: close` asks it to close the connection after sending its
response, which is how `exchange` knows the response is complete.


````python
host, port = urlsplit(BASE).hostname, urlsplit(BASE).port


def exchange(request):
    """Send a request as raw text, and return the raw text of the response."""
    with socket.create_connection((host, port)) as conn:
        conn.sendall(request.encode("utf-8"))
        return conn.makefile(encoding="utf-8").read()


lines = ["GET /stations/tromso HTTP/1.1", f"Host: {host}:{port}", "Connection: close"]
raw = exchange("\r\n".join(lines) + "\r\n\r\n")
print(raw)
````

Read the response from the top:

- `HTTP/1.1 200 OK` is the **status line**: the protocol version, the status code `200`, and the
  reason phrase `OK`, which is there for people.
- `Server`, `Date`, `Content-Type` and `Content-Length` are the **response headers**.
  `Content-Type` says the body is JSON, and `Content-Length` says how many bytes of body to expect.
- The blank line ends the headers.
- Everything after it is the **response body**.

The practice API always reports the same `Date`, so this output matches yours. A real server
reports the moment it responded.

### Taking a response apart

Every client library begins with this step. The headers end at the first blank line, and everything
after it is the body. Reading the connection as text turned every `\r\n` into `\n`, which is why
the split below is on `"\n\n"`.


````python
head, body = raw.split("\n\n", 1)
status_line, *header_lines = head.splitlines()
headers = dict(line.split(": ", 1) for line in header_lines)

print("status line:", status_line)
print("headers:    ", headers)
print("body:       ", body)
````

The body is still a string. `json.loads`, from the **JSON on Disk** notebook in the **Files, Paths
and Formats** guide, turns it into Python values.


````python
station = json.loads(body)

print(type(station).__name__, station)
print(station["name"], "is at", station["latitude"], station["longitude"])
print("Content-Length said", headers["Content-Length"], "and the body is", len(body.encode()), "bytes")
````

`Content-Length` counts bytes, not characters, which is why the comparison encodes the body first.
Here the two agree because every character is ASCII. A body containing `ø` would have more bytes
than characters.

### The same exchange through `urllib.request`

Real code does not write requests by hand. `urllib.request`, part of Python's standard library,
writes the request, reads the response and takes it apart, and `urlopen` returns an object
carrying the pieces. The connection closes at the end of the `with` block, as a file would.


````python
with urllib.request.urlopen(f"{BASE}/stations/tromso") as response:
    print("status: ", response.status, response.reason)
    print("type:   ", response.headers["Content-Type"])
    body = response.read()

print("body:   ", body)
print("as dict:", json.loads(body))
````

`read` returns the body as bytes, not a string, because a body can as easily be an image as text.
`json.loads` accepts bytes and decodes them as UTF-8, so the body goes straight in.

### Asking for a resource that is not there

A request for a station that does not exist still receives a complete, well-formed response, and the
answer in it is no.


````python
lines = ["GET /stations/atlantis HTTP/1.1", f"Host: {host}:{port}", "Connection: close"]
print(exchange("\r\n".join(lines) + "\r\n\r\n"))
````

The status code is `404 Not Found`, and the body says why, in JSON, so a program can read the
reason as well as the code. The connection worked and the server did its job; what failed was the
request, which is what every `4xx` code means. `urlopen` reports this differently, as an exception,
which Common errors shows.

### A method the endpoint does not allow

The method is the first word of the request line, and it changes what the request asks for. The
stations are reference data, and the practice API only lets you read them. Here is a `DELETE`
request for the same station:


````python
lines = ["DELETE /stations/tromso HTTP/1.1", f"Host: {host}:{port}", "Connection: close"]
print(exchange("\r\n".join(lines) + "\r\n\r\n"))
````

`405 Method Not Allowed` is another `4xx`. The request was wrong, because this endpoint does not
accept `DELETE`. The `Allow` response header lists the methods it does accept, here only `GET`. A
real API refuses a method the same way. The **Sending Data** notebook sends `POST`, `PUT` and
`DELETE` requests to a part of the practice API that accepts them.

### A collection, and its members

Most APIs are laid out the same way. A plural noun names a collection, and an id after it names a
member of that collection. On the practice API, `/stations` is the collection's endpoint, and its
response gives a summary of every member.


````python
with urllib.request.urlopen(f"{BASE}/stations") as response:
    stations = json.loads(response.read())

print(stations)
````

The detail for a station lives at its own endpoint, `/stations/{id}`, so reading all of it takes a
request per station, each with the station's id as the path parameter.


````python
for summary in stations:
    with urllib.request.urlopen(f"{BASE}/stations/{summary['id']}") as response:
        station = json.loads(response.read())
    print(f"{station['name']:<9} {station['latitude']:>6} {station['longitude']:>6}")
````

Four stations took five requests, one for the collection and then one for each member. A short
summary list plus a request per detail is the layout you will meet most often, and the
**Pagination** notebook deals with collections too long to send in a single response.

### Calling the real Open-Meteo API

The practice API responds the same way for every reader, which is what a notebook needs. A real
service does the same job over the internet. Open-Meteo publishes weather data free for
non-commercial use, with no key to sign up for, and its archive holds past days. The query
parameters in `archive` choose what comes back, and asking for the ERA5 model by name fixes which
data answers. A day in January 2025 is long settled, so you will get the same numbers as below.

The request goes to `OPEN_METEO`, the address Setup chose: Open-Meteo itself, or the practice API's
recording of it when Open-Meteo is not answering, which holds the same numbers.

Weather data by [Open-Meteo.com](https://open-meteo.com/), under
[CC BY 4.0](https://creativecommons.org/licenses/by/4.0/), from the ERA5 reanalysis. Generated using
Copernicus Climate Change Service information 2026.


````python
query = urlsplit(archive).query

with urllib.request.urlopen(f"{OPEN_METEO}?{query}", timeout=30) as response:
    print("status:", response.status, response.reason)
    print("type:  ", response.headers["Content-Type"])
    weather = json.loads(response.read())

print("keys:  ", sorted(weather))
````

The response has the same parts as the practice API's: a status code, headers, and a JSON body.
`generationtime_ms` is how long the server spent on the request, so it changes every time, which is
why this cell prints the keys and not the whole body.

`timeout=30` stops `urlopen` waiting forever for a slow server. The **Errors and Retries** notebook
covers what to do when the wait runs out.


````python
daily = weather["daily"]
unit = weather["daily_units"]["temperature_2m_mean"]

for day, temperature in zip(daily["time"], daily["temperature_2m_mean"]):
    print(day, temperature, unit)
````

A dictionary of lists, matched by position: `time[0]` goes with `temperature_2m_mean[0]`. `zip`
pairs them, as the **Loops** notebook showed. The **JSON in a Response** notebook reaches into
deeper structures than this.

### Asking both APIs about each station

The practice API knows where every station is. Open-Meteo knows what the weather was there. Asking
both, a station at a time, answers a question neither can answer alone.


````python
def mean_temperatures(latitude, longitude):
    """Three daily mean temperatures for one place, from Open-Meteo's archive."""
    url = (f"{OPEN_METEO}?latitude={latitude}&longitude={longitude}"
           "&start_date=2025-01-15&end_date=2025-01-17&daily=temperature_2m_mean&models=era5")
    with urllib.request.urlopen(url, timeout=30) as response:
        return json.loads(response.read())["daily"]["temperature_2m_mean"]


with urllib.request.urlopen(f"{BASE}/stations") as response:
    stations = json.loads(response.read())

for summary in stations:
    with urllib.request.urlopen(f"{BASE}/stations/{summary['id']}") as response:
        station = json.loads(response.read())
    print(f"{station['name']:<9}", mean_temperatures(station["latitude"], station["longitude"]))
````

### Where each part came from

| In the loop | What it relies on | The section that showed it |
|---|---|---|
| `f"{BASE}/stations"`, then `/stations/{id}` | a collection's endpoint, then a member reached with a path parameter | A collection, and its members |
| `urlopen` in a `with` block | `urllib` sends the request and reads the response | The same exchange through `urllib.request` |
| `json.loads(response.read())` | the response body arrives as bytes of JSON | The same exchange through `urllib.request` |
| `?latitude=...&longitude=...` | query parameters, filled in from the practice API's response | The parts of a URL |
| `timeout=30` | a real server can be slow to respond | Calling the real Open-Meteo API |
| `["daily"]["temperature_2m_mean"]` | the response body is a dictionary of lists | Calling the real Open-Meteo API |

Nine requests to two servers, all sent with the same `urlopen`. The client does not need to know
who is responding, only the endpoint and the shape of the response.

## Common errors

### URLError: unknown url type, when the scheme is missing

A URL without `http://` looks harmless, especially when it is an address and a port.


````python
urllib.request.urlopen("127.0.0.1:8765/stations")
````

`unknown url type: 127.0.0.1` looks like nonsense until you see what `urllib` did. It read
everything before the first colon as the scheme, so it went looking for a protocol called
`127.0.0.1`. Every URL passed to `urlopen` needs its scheme. `BASE` starts with `http://`, which is
why building on it avoids this.


````python
with urllib.request.urlopen(f"{BASE}/stations") as response:
    print(response.status, response.reason)
````

### URLError: Connection refused, when nothing is listening


````python
urllib.request.urlopen("http://127.0.0.1:8799/stations")
````

Read the last line first, which says `Connection refused`. The computer at `127.0.0.1` answered, and
said no program was listening on port `8799`, so the connection failed before any request was sent.
The traceback is long because the error passed through several layers on its way out. The first
exception, `ConnectionRefusedError`, comes from the connection itself, and `urlopen` raised the
`URLError` below it while handling that.

The number in brackets is the operating system's code for the failure. It is 111 on Linux, which is
what Colab runs and where this notebook's output was recorded, and 61 on macOS. The words after it
are the same on both.

The usual causes are a mistyped port and a server that is not running, which here would mean Setup
was not run. Print the address you meant to use, and compare:


````python
print("the practice API is at", BASE)

with urllib.request.urlopen(f"{BASE}/stations") as response:
    print(response.status, response.reason)
````

### HTTPError: HTTP Error 404, which urlopen raises

Earlier, a raw request for this station received a complete `404` response. `urlopen` treats that
response differently.


````python
urllib.request.urlopen(f"{BASE}/stations/atlantis")
````

`urlopen` turns any status code of 400 or above into an exception. The exception carries the
response, so catch it as `HTTPError`, from `urllib.error`, and read the status code and the body
from it:


````python
from urllib.error import HTTPError

try:
    urllib.request.urlopen(f"{BASE}/stations/atlantis")
except HTTPError as error:
    print("status:", error.code, error.reason)
    print("body:  ", json.loads(error.read()))
````

The `requests` library in the **Your First Request** notebook makes the opposite choice. A `404`
comes back as an ordinary response, and raises only when you ask it to.

### JSONDecodeError: the body was HTML, not JSON

The practice API's home page, at `BASE` itself, is written for people.


````python
with urllib.request.urlopen(BASE) as response:
    body = response.read()

json.loads(body)
````

`Expecting value: line 1 column 1 (char 0)` means the very first character was not the start of any
JSON value. It was `<`, the start of an HTML tag. Real code meets this when a URL is slightly wrong
and the server responds with a web page, or when a proxy between you and the server returns an error
page of its own, so check `Content-Type` before parsing:


````python
with urllib.request.urlopen(BASE) as response:
    kind = response.headers["Content-Type"]
    body = response.read()

if kind.startswith("application/json"):
    print(json.loads(body))
else:
    print("not JSON:", kind)
    print(body.decode().splitlines()[0])
````

## Recap

- An API is the set of requests a program agrees to accept and the responses it promises to send
  back. A web API carries them over HTTP.
- A request carries a method, a URL, headers, and sometimes a body. A response carries a status
  code, headers, and usually a body.
- The method is the verb: `GET` reads, `POST` creates, `PUT` replaces, `PATCH` changes part, and
  `DELETE` removes. An endpoint that does not accept a method responds with `405` and an `Allow`
  header.
- Parameters travel in the path (`/stations/tromso`), in the query (`?start_date=2025-01-15`), in
  headers, or in the body.
- The first digit of a status code gives its class: `2xx` worked, `4xx` means the request was
  wrong, and `5xx` means the server failed.
- Underneath, a request and a response are plain text: a first line, headers, a blank line, and a
  body. `urlsplit` and `parse_qs` take a URL apart, and `json.loads` turns a JSON body into Python
  values.
- A `404` is a complete response from a working server. `urlopen` raises `HTTPError` for it, and
  the exception carries the response.
- `Connection refused` means nothing was listening on that port, so no request was ever sent.
- Check `Content-Type` before parsing a body as JSON.
