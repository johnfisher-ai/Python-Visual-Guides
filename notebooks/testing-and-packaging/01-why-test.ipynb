## The idea

### The problem

A module turns the day's readings from four weather stations into a mean temperature for each
station. On Tuesday, the station on Svalbard sent only empty readings, and the daily summary stopped
with `ZeroDivisionError`. The fix was one line: a station with no readings has a mean of `None`. The
summary ran again, printed four stations, and the fix was forgotten.

A week later, somebody rewrites the function with `statistics.fmean`, which is shorter, and drops
the line that looks like clutter. They check the new version the way the fix was checked, by calling
it on a few lists of readings and looking at the numbers, which are right. Nobody tries a list with
no readings in it, because nobody remembers that it matters. The next day Svalbard is silent again,
and the summary stops again, this time with `StatisticsError`, in a report somebody else runs.

A check by hand covers what you think of, on the day you think of it. The bug came back because the
one fact that mattered, that a station can send nothing, was kept in somebody's memory and not in
anything that runs.

### What a test is

> A **test** is a function that runs a piece of code in one situation and checks the result with
> `assert`. It raises nothing when the code does what it should, and raises an exception when it
> does not. A project's tests together are its **test suite**, and a **test runner** finds the
> tests, runs every one, and reports which **passed** and which **failed**. A bug that was fixed
> and comes back is a **regression**, and a test written for that bug is a **regression test**: it
> fails as soon as the bug returns.

### Why it works that way

- **A test knows the answer.** `print(mean([4.2, 5.8]))` shows a number and leaves you to judge it.
  `assert mean([4.2, 5.8]) == 5.0` makes the judgment, and says nothing unless the answer is wrong.
- **A test runs the same way every time.** A check by hand is the check somebody thinks of that
  day. A test is code, so a check written in March still runs in September, after every change, for
  anybody who runs the suite.
- **A failing test names what broke.** When `test_mean_of_no_readings_is_none` fails, its name says
  which behavior changed, a second after the change and not a day later in a report.
- **A bug becomes a test before it is fixed.** Write the test, watch it fail with the bug's own
  error, fix the code until the test passes, and keep the test. The bug cannot come back without
  the test failing.
- **Tests make a change safe to try.** Rewriting `mean` with `statistics.fmean` is a reasonable
  change. With tests, you learn at once whether it changed anything the tests check.
- **A test checks only what it says.** A suite that passes says that the cases in it work, and
  nothing about a case nobody wrote. Choosing the cases is most of the skill.

## Worked examples

### A module, checked by eye

The module holds three functions. `parse_reading` turns a line such as `Bergen,4.2` into a station
and a temperature in degrees Celsius, `mean` averages a station's readings and skips the empty ones,
and `summarize` does both for a day's lines. This is the module as it was before Tuesday.
`%%writefile` saves the cell as a file, and runs nothing:


````python
%%writefile scratch/stations/readings.py
"""Readings from the weather stations, and each station's mean temperature."""


def parse_reading(line):
    """A (station, celsius) pair from a line such as 'Bergen,4.2'. An empty reading is None."""
    station, celsius = line.strip().split(",")
    return station, float(celsius) if celsius else None


def mean(values):
    """The mean of the readings that are not None."""
    present = [value for value in values if value is not None]
    return sum(present) / len(present)


def summarize(lines):
    """Each station's mean temperature, from lines of readings."""
    by_station = {}
    for line in lines:
        station, celsius = parse_reading(line)
        by_station.setdefault(station, []).append(celsius)
    return {station: mean(values) for station, values in by_station.items()}
````

A check by eye imports the module and prints what its functions return:


````python
import readings

print(readings.parse_reading("Bergen,4.2"))
print(readings.mean([4.2, 5.8, None]))
print(readings.summarize(["Bergen,4.2", "Bergen,5.8", "Oslo,-2.4", "Oslo,-1.6"]))
````

Every value is right, and the only thing that says so is you. You had to know that the mean of 4.2
and 5.8 is 5.0, and read three lines to see that nothing was wrong, and after the next change to the
module you would have to know it again.

### An assert that knows the answer

`assert` takes a condition. When the condition is true, nothing happens, and when it is false,
`assert` raises `AssertionError`, as the **Debugging** notebook showed. A condition that states the
right answer makes the judgment that a check by eye left to you:


````python
assert readings.mean([4.2, 5.8, None]) == 5.0
print("the mean is 5.0")

try:
    assert readings.mean([4.2, 5.8, None]) == 4.2            # a wrong answer, on purpose
except AssertionError:
    print("AssertionError: the mean is not 4.2")
````

The first `assert` passed without a word. The second raised, and nothing had to be read or remembered
to know that something was wrong.

A test is a check like that inside a function, with a name that says what it checks. Calling the
function runs the check:


````python
def test_mean_of_two_readings():
    assert readings.mean([4.2, 5.8]) == 5.0


def test_mean_skips_missing_readings():
    assert readings.mean([4.2, None, 5.8]) == 5.0


def test_summary_of_two_stations():
    lines = ["Bergen,4.2", "Bergen,5.8", "Oslo,-2.4", "Oslo,-1.6"]
    assert readings.summarize(lines) == {"Bergen": 5.0, "Oslo": -2.0}


test_mean_of_two_readings()
test_mean_skips_missing_readings()
test_summary_of_two_stations()
print("three tests passed")
````

These tests take no arguments and return nothing, and a test has passed when calling it raised
nothing. They reach `mean` through the module, as `readings.mean`, at the moment they run, so a test
always runs the code the module holds at that moment. The sections below rely on that.

### A bug, turned into a failing test

Tuesday's readings include two empty readings from Svalbard, and one empty reading from Bergen and
from Tromso:


````python
TUESDAY = ["Bergen,4.2", "Bergen,5.8", "Bergen,", "Oslo,-2.4", "Oslo,-1.6",
           "Svalbard,", "Svalbard,", "Tromso,-6.3", "Tromso,-5.7", "Tromso,"]

try:
    readings.summarize(TUESDAY)
except ZeroDivisionError as error:
    print("ZeroDivisionError:", error)
````

`mean` dropped Svalbard's empty readings and divided by the number left, which was zero. Before the
code changes, a test states what should happen instead. No number is the right mean of no readings,
so the answer is `None`, which `summarize` can pass on:


````python
def test_mean_of_no_readings_is_none():
    assert readings.mean([None, None]) is None


try:
    test_mean_of_no_readings_is_none()
except ZeroDivisionError as error:
    print("the test fails with the bug's own error: ZeroDivisionError:", error)
````

The test fails with the error the summary raised, so it reproduces the bug. Now the fix, in the
module's file:


````python
%%writefile scratch/stations/readings.py
"""Readings from the weather stations, and each station's mean temperature."""


def parse_reading(line):
    """A (station, celsius) pair from a line such as 'Bergen,4.2'. An empty reading is None."""
    station, celsius = line.strip().split(",")
    return station, float(celsius) if celsius else None


def mean(values):
    """The mean of the readings that are not None, or None when there are none."""
    present = [value for value in values if value is not None]
    return sum(present) / len(present) if present else None


def summarize(lines):
    """Each station's mean temperature, from lines of readings."""
    by_station = {}
    for line in lines:
        station, celsius = parse_reading(line)
        by_station.setdefault(station, []).append(celsius)
    return {station: mean(values) for station, values in by_station.items()}
````

The file has changed, and the module this notebook imported has not: Python keeps running the code it
imported until it is told to import the file again. `importlib.reload` runs the file again, and then
the tests run against the fix:


````python
importlib.reload(readings)

test_mean_of_two_readings()
test_mean_skips_missing_readings()
test_summary_of_two_stations()
test_mean_of_no_readings_is_none()
print("four tests passed")
print(readings.summarize(TUESDAY))
````

The order is the point: the test first, failing with the bug's error, then the fix, then the test
passing. A test that has never failed may not check the bug at all, and this one was seen to fail for
the right reason.

### A suite: every test, found by its name

Calling every test by hand has the fault the tests were meant to remove: somebody has to remember to
call the new one. `run_tests` finds the tests by their names instead, runs every one, reports each
failure with the exception it raised, and counts the results:


````python
def run_tests(namespace):
    """Run every function in namespace whose name starts with test_, and report every failure."""
    tests = [value for name, value in namespace.items() if name.startswith("test_") and callable(value)]
    failed = 0
    for test in tests:
        try:
            test()
        except Exception as error:
            failed += 1
            print(f"FAILED {test.__name__}: {error!r}")
    print(f"{len(tests) - failed} passed, {failed} failed")


run_tests(globals())
````

`globals()` is the dictionary of every name the notebook has defined, so a test is in the suite the
moment it is defined, and nobody has to add it to a list. A test that fails is reported and the rest
still run, so one report shows every failure. The **Your First Test** notebook hands this job to
pytest, which finds tests in files the same way, by their names.

### A week later, without tests

A week later, `mean` is rewritten with `statistics.fmean`, which averages numbers in one call. The
new version is shorter, and it leaves out the check for no readings, which looks like clutter to
somebody who was not there on Tuesday:


````python
%%writefile scratch/stations/readings.py
"""Readings from the weather stations, and each station's mean temperature."""

import statistics


def parse_reading(line):
    """A (station, celsius) pair from a line such as 'Bergen,4.2'. An empty reading is None."""
    station, celsius = line.strip().split(",")
    return station, float(celsius) if celsius else None


def mean(values):
    """The mean of the readings that are not None."""
    return statistics.fmean(value for value in values if value is not None)


def summarize(lines):
    """Each station's mean temperature, from lines of readings."""
    by_station = {}
    for line in lines:
        station, celsius = parse_reading(line)
        by_station.setdefault(station, []).append(celsius)
    return {station: mean(values) for station, values in by_station.items()}
````

It is checked the way the fix was, by eye, with readings that come to mind:


````python
importlib.reload(readings)

print(readings.mean([4.2, 5.8]))
print(readings.mean([4.2, None, 5.8]))
print(readings.summarize(["Bergen,4.2", "Bergen,5.8", "Oslo,-2.4", "Oslo,-1.6"]))
````

Every value is right, so the change stays. The next day, the summary runs on a day like Tuesday:


````python
try:
    readings.summarize(TUESDAY)
except Exception as error:
    print(f"{type(error).__name__}: {error}")
````

The bug is back, a day after the change and in somebody else's report. The error is a different one
from Tuesday's, so nothing in the message points at the fix that was lost.

### A case the passing suite never tried

The readings arrive as text. One day a file comes from a spreadsheet that writes a minus sign as `−`,
the character U+2212, which looks like a hyphen and is a different character:


````python
line = "Tromso,\u22126.3"
print(line, "| the character before 6.3 is U+" + format(ord(line[7]), "04X"))

run_tests(globals())

try:
    readings.parse_reading(line)
except ValueError as error:
    print("ValueError:", error)
````

Every test passed, and `parse_reading` still fails on a line from a real file, because no test gave
it one. A passing suite speaks only for the cases in it. The cases worth writing are the ordinary
ones, the edges (no readings, one reading, an empty field), the bad input that really arrives, and
every bug that has been fixed. This one gets a test first, as Tuesday's bug did:


````python
def test_a_unicode_minus_sign_reads_as_negative():
    assert readings.parse_reading("Tromso,\u22126.3") == ("Tromso", -6.3)


run_tests(globals())
````

The test fails with the error the line raised. The fix replaces U+2212 with the hyphen-minus that
`float` reads:


````python
%%writefile scratch/stations/readings.py
"""Readings from the weather stations, and each station's mean temperature."""

import statistics


def parse_reading(line):
    """A (station, celsius) pair from a line such as 'Bergen,4.2'. An empty reading is None.

    A minus sign written as U+2212, as some spreadsheets write it, reads as a hyphen-minus.
    """
    station, celsius = line.strip().split(",")
    celsius = celsius.replace("\u2212", "-")
    return station, float(celsius) if celsius else None


def mean(values):
    """The mean of the readings that are not None, or None when there are none."""
    present = [value for value in values if value is not None]
    return statistics.fmean(present) if present else None


def summarize(lines):
    """Each station's mean temperature, from lines of readings."""
    by_station = {}
    for line in lines:
        station, celsius = parse_reading(line)
        by_station.setdefault(station, []).append(celsius)
    return {station: mean(values) for station, values in by_station.items()}
````

### Unit, integration and end-to-end tests

Every test so far is a **unit test**: it calls one function with values written into the test. An
**integration test** checks parts working together, here reading a file and summarizing it, and an
**end-to-end test** runs the whole program the way a person does, and checks what it prints. The
program is a script that prints the summary of a file:


````python
%%writefile scratch/stations/summary.py
"""Print each station's mean temperature from a file of readings: python summary.py readings.csv"""

import sys

from readings import summarize

with open(sys.argv[1], encoding="utf-8") as file:
    for station, celsius in summarize(file).items():
        print(station, "no readings" if celsius is None else f"{celsius:.1f}")
````

The file holds Tuesday's readings. The integration test opens it and summarizes it. The end-to-end
test runs the script with `subprocess.run`, which starts a second Python, the one `sys.executable`
names, and keeps what the script printed:


````python
(PROJECT / "tuesday.csv").write_text("\n".join(TUESDAY) + "\n", encoding="utf-8")


def test_summary_of_a_file():
    with open(PROJECT / "tuesday.csv", encoding="utf-8") as file:
        assert readings.summarize(file) == {"Bergen": 5.0, "Oslo": -2.0, "Svalbard": None, "Tromso": -6.0}


def test_the_script_prints_every_station():
    finished = subprocess.run([sys.executable, "summary.py", "tuesday.csv"], cwd=PROJECT, capture_output=True, text=True)
    assert finished.returncode == 0, finished.stderr.strip().splitlines()[-1]       # the error's last line
    assert finished.stdout.splitlines() == ["Bergen 5.0", "Oslo -2.0", "Svalbard no readings", "Tromso -6.0"]


run_tests(globals())
````

The end-to-end test is the closest to what a person sees, and the slowest, since it starts a program:


````python
started = time.perf_counter()
test_mean_of_two_readings()
unit = time.perf_counter() - started

started = time.perf_counter()
test_the_script_prints_every_station()
end_to_end = time.perf_counter() - started

print("the end-to-end test took longer than 100 runs of the unit test:", end_to_end > 100 * unit)
````

That difference is why a suite usually holds many unit tests, which run in a moment and point at one
function, and fewer integration and end-to-end tests, which check that the parts were put together
correctly. When an end-to-end test fails, the unit tests say whether a function broke or the way the
functions are joined.

### A change, with the suite run before, during and after

The pieces of this notebook, in one change: the summary should give each mean in degrees Fahrenheit
as well. First, the suite, to know that everything passes before anything changes:


````python
run_tests(globals())
````

Then the tests for the new behavior, before the code. A conversion turns 0 °C into 32 °F and 100 °C
into 212 °F, facts that come from outside the code, and the script prints both temperatures.
`test_the_script_prints_every_station` is written again with the new lines, which replaces the old
test of that name:

Three failures, all expected, since the code has not changed yet: they show that the new tests check
something. Now the code. The module gains `to_fahrenheit`, and the script prints both temperatures:


````python
%%writefile scratch/stations/readings.py
"""Readings from the weather stations, and each station's mean temperature."""

import statistics


def parse_reading(line):
    """A (station, celsius) pair from a line such as 'Bergen,4.2'. An empty reading is None.

    A minus sign written as U+2212, as some spreadsheets write it, reads as a hyphen-minus.
    """
    station, celsius = line.strip().split(",")
    celsius = celsius.replace("\u2212", "-")
    return station, float(celsius) if celsius else None


def mean(values):
    """The mean of the readings that are not None, or None when there are none."""
    present = [value for value in values if value is not None]
    return statistics.fmean(present) if present else None


def summarize(lines):
    """Each station's mean temperature, from lines of readings."""
    by_station = {}
    for line in lines:
        station, celsius = parse_reading(line)
        by_station.setdefault(station, []).append(celsius)
    return {station: mean(values) for station, values in by_station.items()}


def to_fahrenheit(celsius):
    """A temperature in degrees Celsius, in degrees Fahrenheit."""
    return celsius * 9 / 5 + 32
````

````python
%%writefile scratch/stations/summary.py
"""Print each station's mean temperature from a file of readings: python summary.py readings.csv"""

import sys

from readings import summarize, to_fahrenheit

with open(sys.argv[1], encoding="utf-8") as file:
    for station, celsius in summarize(file).items():
        fahrenheit = to_fahrenheit(celsius)
        print(station, "no readings" if celsius is None else f"{celsius:.1f} C, {fahrenheit:.1f} F")
````

## Common errors

### No error, and a test that cannot fail: a test with no assert


````python
def to_fahrenheit_upside_down(celsius):
    return celsius * 5 / 9 + 32               # the fraction inverted, on purpose


def test_boiling_point():
    to_fahrenheit_upside_down(100)


run_tests({"test_boiling_point": test_boiling_point})
````

The test passed, and the function turns boiling water into 87.6 °F. A test passes when it raises
nothing, and a test with no `assert` raises only when the code crashes, so it checks that the
function runs and nothing about what it returns. Give the test the answer:


````python
def test_boiling_point():
    assert to_fahrenheit_upside_down(100) == 212


run_tests({"test_boiling_point": test_boiling_point})
````

Now the test fails, as it should while the function is wrong.

### No error, and the bug kept: an expected value copied from the code's output


````python
print(to_fahrenheit_upside_down(100))


def test_boiling_point():
    assert to_fahrenheit_upside_down(100) == 87.55555555555556          # the value printed above


run_tests({"test_boiling_point": test_boiling_point})
````

The expected value came from running the code, so the test checks that the code returns what it
returns, and it passes with the bug in place. Fixing the formula would now make it fail, so the test
defends the bug. An expected value has to come from somewhere other than the code under test: a
known fact, such as water boiling at 212 °F, a value worked out by hand, or a source you trust. The
test that states the fact passes for the module's own `to_fahrenheit`, which has the fraction the
right way up:


````python
def test_boiling_point():
    assert readings.to_fahrenheit(100) == 212


run_tests({"test_boiling_point": test_boiling_point})
````

### NameError: name 'parsed' is not defined


````python
def test_a_day_parses():
    global parsed
    parsed = [readings.parse_reading(line) for line in TUESDAY]
    assert len(parsed) == 10


def test_bergen_comes_first():
    assert parsed[0] == ("Bergen", 4.2)


run_tests({"test_bergen_comes_first": test_bergen_comes_first})
````

`test_bergen_comes_first` uses a list that `test_a_day_parses` makes. Run after that test, it passes.
Run alone, or first, it fails with an error that has nothing to do with Bergen. A test runner can run
a single test when you ask it to, or run the tests in a different order, so a test that needs another
test is broken even while the whole suite passes. Let a test make what it needs:


````python
def test_bergen_comes_first():
    parsed = [readings.parse_reading(line) for line in TUESDAY]
    assert parsed[0] == ("Bergen", 4.2)


run_tests({"test_bergen_comes_first": test_bergen_comes_first})
````

## Recap

- A test is a function that runs code in one situation and checks the result with `assert`. It passes
  when it raises nothing.
- A check by eye covers what somebody remembers to check. A test runs the same checks after every
  change, for anybody.
- Turn a bug into a test before fixing it: see the test fail with the bug's error, fix the code, and
  keep the test, so the bug cannot return unnoticed.
- A suite finds its tests by name and reports every failure, and a failing test's name says which
  behavior broke.
- A passing suite speaks only for its cases. Write tests for ordinary input, the edges, the bad input
  that really arrives, and every bug that was fixed.
- Unit tests run in a moment and point at one function. Integration and end-to-end tests check that
  the parts work together, and take longer.
- An expected value comes from outside the code under test, and a test needs an `assert` to fail at
  all.
