# CSV &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
import csv
from pathlib import Path
import shutil

scratch = Path("scratch")
scratch.mkdir(exist_ok=True)

cities = scratch / "cities.csv"
```

**1.** Write a CSV with the header `city,country,population` and three rows, one of which
has a comma inside the city name.


```python
rows = [
    ["Washington, D.C.", "United States", 671803],
    ["Tokyo", "Japan", 13960000],
    ["Lima", "Peru", 9752000],
]

with open(cities, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["city", "country", "population"])
    writer.writerows(rows)

print(cities.read_text(encoding="utf-8"))
```

The writer put quotes around `Washington, D.C.` and left the other cities alone. Nothing in the
code asked for that; the writer adds quoting where the format requires it.


**2.** Read it back with `csv.reader` and print the number of fields in each row.


```python

with open(cities, newline="", encoding="utf-8") as f:
    for row in csv.reader(f):
        print(len(row), row)
```

Three fields on every row including the quoted one. Splitting that file on commas would have
given four for the first city.


**3.** Read it with `DictReader` and print each city with its country.


```python

with open(cities, newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        print(f"{row['city']} is in {row['country']}")
```

**4.** Total the populations.


```python

total = 0

with open(cities, newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        total += int(row["population"])

print(f"{total:,}")
```

`int(...)` is the whole answer. Without it, `total` would have started as `0` and raised on the
first `+`, or started as `""` and silently joined the numbers into one long string.


**5.** Write a semicolon-separated version of the same data, then read it back correctly.


```python

semi = scratch / "cities_semi.csv"

with open(cities, newline="", encoding="utf-8") as source:
    rows = list(csv.reader(source))

with open(semi, "w", newline="", encoding="utf-8") as f:
    csv.writer(f, delimiter=";").writerows(rows)

print(semi.read_text(encoding="utf-8"))

with open(semi, newline="", encoding="utf-8") as f:
    print(list(csv.reader(f, delimiter=";"))[1])
```

Note what the writer did with `Washington, D.C.` when the delimiter changed: the comma is no
longer a separator, so the quotes are no longer needed and were not added. The quoting rules
follow the delimiter.


**6.** Write a file with one row that has too few fields, read it with `DictReader` using
`restval`, and print the result.


```python

short = scratch / "short.csv"
short.write_text("city,country,population\nOslo,Norway\n", encoding="utf-8")

with open(short, newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f, restval="MISSING"):
        print(dict(row))
```

Without `restval` the population would have been `None`, which is easy to miss when it reaches a
calculation. `"MISSING"` is visible in output and will raise if something tries to convert it,
which is the better failure.


```python

shutil.rmtree(scratch)

print("cleaned up:", not scratch.exists())
```
