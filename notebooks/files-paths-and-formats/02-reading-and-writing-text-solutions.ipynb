# Reading and Writing Text &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
from pathlib import Path
import shutil

scratch = Path("scratch")
scratch.mkdir(exist_ok=True)
notes = scratch / "notes.txt"
```

**1.** Write three lines to `scratch/notes.txt` using `with` and `write`, then print the
file back.


```python
with open(notes, "w") as f:
    f.write("first line\n")
    f.write("second line\n")
    f.write("third line\n")

print(repr(notes.read_text()))
```

The `\n` at the end of each string is what makes them three lines. Without it the file holds one
line reading `first linesecond linethird line`.


**2.** Append a fourth line without losing the first three.


```python
with open(notes, "a") as f:
    f.write("fourth line\n")

print(notes.read_text())
```

`"a"` opens at the end. Using `"w"` here would have replaced all three lines with the fourth,
with no warning.


**3.** Open the file with mode `"x"` and print the exception type rather than letting it
stop the cell.


```python
try:
    with open(notes, "x") as f:
        f.write("this will not happen\n")
except FileExistsError as e:
    print(type(e).__name__, "-", e)
```

`"x"` refused because the file exists. That refusal is the whole point of the mode: it turns an
accidental overwrite into an error you can catch.


**4.** Loop over the file printing each line numbered, with the trailing newline removed.


```python
with open(notes) as f:
    for number, line in enumerate(f, start=1):
        print(number, line.rstrip())
```

Looping over the file object reads one line at a time and holds only that line in memory.
`readlines()` would have given the same output and loaded the whole file first.


**5.** Open a file, write to it, and check the size before and after closing, without using
`with`.


```python
demo = scratch / "manual.txt"

f = open(demo, "w")
f.write("some text that is going nowhere yet\n")
print("before close:", demo.stat().st_size, "bytes")

f.close()
print("after close: ", demo.stat().st_size, "bytes")

# f.close() is the line that made the data appear. It flushed the buffer to disk.
```

`f.flush()` would have done the same without closing. `with` does the close for you, which is
why this situation does not arise when you use it.


**6.** Write `["red", "green", "blue"]` so each is on its own line, using `writelines`.
Then do the same with `write` and a join, and confirm the files match.


```python
colors = ["red", "green", "blue"]

a = scratch / "a.txt"
b = scratch / "b.txt"

with open(a, "w") as f:
    f.writelines(c + "\n" for c in colors)

with open(b, "w") as f:
    f.write("\n".join(colors) + "\n")

print(repr(a.read_text()))
print(repr(b.read_text()))
print("identical:", a.read_text() == b.read_text())
```

`writelines` needed the newlines added to every item, because it adds none itself. The join
version puts a newline between the items and one more at the end, which is why the trailing
`+ "\n"` is there.

The second form says what it does more clearly, and is the one to reach for.


```python
shutil.rmtree(scratch)

print("cleaned up:", not scratch.exists())
```
