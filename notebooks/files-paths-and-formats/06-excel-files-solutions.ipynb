# Excel Files &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python

import openpyxl
from openpyxl import Workbook
import datetime
from pathlib import Path
import shutil

scratch = Path("scratch")
scratch.mkdir(exist_ok=True)
cities = scratch / "cities.xlsx"
```

**1.** Create a workbook with a sheet named `Cities`, write a header and three rows, save
it, and print the sheet names.


```python

wb = Workbook()
sheet = wb.active
sheet.title = "Cities"

sheet.append(["city", "country", "population"])
sheet.append(["Tokyo", "Japan", 13960000])
sheet.append(["Lima", "Peru", 9752000])
sheet.append(["Oslo", "Norway", 709037])

wb.save(cities)

print(wb.sheetnames)
```

`wb.active` is the sheet a new workbook starts with. Renaming it beats creating a second one
and leaving an empty `Sheet` behind.


**2.** Load it back and print every row.


```python

sheet = openpyxl.load_workbook(cities)["Cities"]

for row in sheet.iter_rows(values_only=True):
    print(row)
```

**3.** Print the value and type of cell `B2`, addressed both ways.


```python

print("by coordinate:", repr(sheet["B2"].value), type(sheet["B2"].value).__name__)
print("by numbers:   ", repr(sheet.cell(row=2, column=2).value))
```

`row=2, column=2` is `B2`, because both are counted from one. Reaching for `row=1, column=1` to
mean the second row is the mistake to expect here.


**4.** Add a second sheet called `Summary`, write the number of cities into `A1`, and save.


```python

wb = openpyxl.load_workbook(cities)

summary = wb.create_sheet("Summary")
summary["A1"] = wb["Cities"].max_row - 1      # minus the header

wb.save(cities)

print(wb.sheetnames)
print("A1 of Summary:", openpyxl.load_workbook(cities)["Summary"]["A1"].value)
```

`max_row` counts the header, so the count of data rows is one fewer. Getting that wrong is how a
summary ends up claiming one more record than the file holds.


**5.** Write today's date into a cell, set its `number_format` to `"General"`, save, reload,
and print what comes back.


```python

dated = scratch / "dated.xlsx"

wb = Workbook()
wb.active["A1"] = datetime.date(2026, 3, 1)
wb.active["A1"].number_format = "General"
wb.save(dated)

cell = openpyxl.load_workbook(dated).active["A1"]

print(repr(cell.value), type(cell.value).__name__, "| format:", repr(cell.number_format))
```

The date went in and an integer came out. Nothing was lost: the number is the date, without the
instruction saying to display it as one.


**6.** Convert the number back into a date.


```python

from openpyxl.utils.datetime import from_excel

serial = cell.value

print("with openpyxl:", from_excel(serial, openpyxl.load_workbook(dated).epoch).date())

epoch = datetime.datetime(1899, 12, 30)
print("by hand:      ", (epoch + datetime.timedelta(days=serial)).date())
```

Both give 1 March 2026. `from_excel` is the one to use, because it also handles the dates in
early 1900 where the hand calculation is a day out.


```python

shutil.rmtree(scratch)

print("cleaned up:", not scratch.exists())
```
