# Paths &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


```python
from pathlib import Path, PurePosixPath

scratch = Path("scratch")
(scratch / "data").mkdir(parents=True, exist_ok=True)
(scratch / "data" / "readings.csv").write_text("region,value\nnorth,18.5\n")
```

**1.** Print the name of the folder this notebook is running in, and whether that path is
absolute.


```python
here = Path.cwd()

print(here.name)
print(here.is_absolute())
```

`Path.cwd()` is always absolute, which is the point of it: it is the fixed starting point every
relative path is measured from.


**2.** Build the path `scratch/data/readings.csv` with `/` rather than a string, and print
whether it exists.


```python
target = Path("scratch") / "data" / "readings.csv"

print(target)
print(target.exists())
```

`Path("scratch") / "data"` builds the path piece by piece. Writing
`Path("scratch" + "/" + "data")` gives the same result here and the wrong separator on Windows.


**3.** For `PurePosixPath("/var/log/system/errors.2026.log")`, print the name, stem, suffix
and parent.


```python
p = PurePosixPath("/var/log/system/errors.2026.log")

print("name:  ", p.name)
print("stem:  ", p.stem)
print("suffix:", p.suffix)
print("parent:", p.parent)
```

Note what `stem` and `suffix` did with the two dots. `suffix` is only the last one, `.log`, and
`stem` keeps `errors.2026` intact. Splitting on `.` by hand would have given `errors` and lost
the year.


**4.** Print the same path with the suffix changed to `.txt`, and separately with the
filename changed to `summary.log`.


```python
print(p.with_suffix(".txt"))
print(p.with_name("summary.log"))
```

`with_suffix` replaces only the last suffix, so `errors.2026` survives. `with_name` replaces the
whole filename and leaves the folders alone.


**5.** Create a path to a file that does not exist, print it, and print `exists()`.


```python
imaginary = Path("nowhere") / "file.txt"

print(imaginary)
print(imaginary.exists())

# The first line touched nothing. exists() is the line that asked the filesystem.
```

Building a path is text arithmetic and cannot fail. Only the second line could have been
affected by what is on the disk, which is why it is the only one that can be wrong tomorrow.


**6.** Print whether `a/b/c.txt` equals `a/./b/c.txt`, and whether it equals
`a/b/../b/c.txt`.


```python
base = PurePosixPath("a/b/c.txt")

print(base == PurePosixPath("a/./b/c.txt"))
print(base == PurePosixPath("a/b/../b/c.txt"))

# ./ is dropped when the path is built, so the first pair is identical.
# .. is kept, because removing it correctly depends on what is on the disk:
# if b is a symbolic link, b/.. is not necessarily a.
```

`resolve()` is what settles the second question, and it has to touch the filesystem to do it.
That is the difference between a comparison of how two paths were written and a comparison of
where they actually point.


```python
import shutil
shutil.rmtree(scratch)

print("cleaned up:", not scratch.exists())
```
