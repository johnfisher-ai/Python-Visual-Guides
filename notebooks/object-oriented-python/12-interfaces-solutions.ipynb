```python
from abc import ABC, abstractmethod
from typing import Protocol, runtime_checkable
from collections.abc import Iterable, Sized

print("ready")
```

**1.** An abstract base class, and the refusal to create it.


```python
class Instrument(ABC):
    @abstractmethod
    def read(self):
        """Return the instrument's current value."""


try:
    Instrument()
except TypeError as error:
    print(error)
```

`Instrument` inherits from `ABC` and has one abstract method, so Python will not create it. It exists
to say what every instrument must do.


**2.** A subclass that meets the requirement.


```python
class Thermometer(Instrument):
    def read(self):
        return -4.1


print(Thermometer().read())
```

Defining `read` empties the list of missing methods, so a thermometer can be created.


**3.** Shared code in the base class that calls the abstract method.


```python
class Instrument(ABC):
    def describe(self):
        return f"{type(self).__name__} reads {self.read()}"

    @abstractmethod
    def read(self):
        """Return the instrument's current value."""


class Thermometer(Instrument):
    def read(self):
        return -4.1


print(Thermometer().describe())
```

`describe` is written once and works for every instrument, because each subclass is required to
supply the `read` it depends on. `type(self).__name__` names whichever class the object really is, as
in the **Inheritance** notebook.


**4.** A subclass that forgets.


```python
class Barometer(Instrument):
    pass


try:
    Barometer()
except TypeError as error:
    print(error)

# The message names the class that cannot be created, Barometer, and the
# method it still has to define, read. It says what to write, and where.
```

The error arrives when a barometer is created, not when something later calls `describe`, which is
the whole point of declaring `read` abstract.


**5.** A Protocol, and three objects tested against it.


```python
@runtime_checkable
class Readable(Protocol):
    def read(self) -> float:
        ...


class RainGauge:
    """Inherits from nothing of ours, and has a read method."""

    def read(self):
        return 2.5


print("Thermometer:", isinstance(Thermometer(), Readable))
print("RainGauge:  ", isinstance(RainGauge(), Readable))
print("a string:   ", isinstance("reading", Readable))
```

`RainGauge` fits because it has a `read` method, even though it has no connection to `Instrument`. A
string has no `read`, so it does not fit. The check looked only for the name.


**6.** The standard library's interfaces, without inheriting.


```python
class Log:
    def __init__(self):
        self._messages = []

    def __len__(self):
        return len(self._messages)

    def __iter__(self):
        return iter(self._messages)


class LengthOnly:
    def __len__(self):
        return 0


print("Log is Sized:          ", isinstance(Log(), Sized))
print("Log is Iterable:       ", isinstance(Log(), Iterable))
print("LengthOnly is Iterable:", isinstance(LengthOnly(), Iterable))
```

`Log` inherits from nothing and still counts as both, because it has `__len__` and `__iter__`.
`LengthOnly` has a length but no `__iter__`, so it is `Sized` and not `Iterable`: each interface asks
for its own method.
