# Why Classes &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.

The cell below rebuilds what the notebook set up. Run it first.


```python
from statistics import mean


class Station:
    def __init__(self, name, readings, unit):
        self.name = name
        self.readings = readings
        self.unit = unit

    def average(self):
        return f"{mean(self.readings):.1f} {self.unit}"

    def in_celsius(self):
        if self.unit == "C":
            return list(self.readings)
        return [round((r - 32) * 5 / 9, 1) for r in self.readings]

    def report(self):
        return f"{self.name}: {self.average()}, coldest {min(self.in_celsius())} C"


print("ready")
```

**1.** Three variables that describe one book, as a class.


```python
class Book:
    def __init__(self, title, pages, read_so_far):
        self.title = title
        self.pages = pages
        self.read_so_far = read_so_far

    def progress(self):
        return f"{self.read_so_far / self.pages:.0%}"


for book in [Book("Dune", 412, 103), Book("Emma", 474, 474)]:
    print(f"  {book.title:<6} {book.progress()}")
```

`progress` takes no arguments because both numbers it needs are already on the object. That is the
shape to look for: a method whose argument list is empty is usually a sign the data belongs where
it is.

The `:.0%` format does the multiplying by a hundred and adds the sign, so the method holds no
arithmetic beyond the division.


**2.** The same record with plain attributes.


```python
class StationRecord:
    def __init__(self, name, country):
        self.name = name
        self.country = country


record = StationRecord("Tromso", "NO")
print(record.name, record.country)

record.name = "Tromso Airport"
print(record.name)
```

Setting an attribute needs no setter. `record.name = ...` is the whole mechanism, and it is what
`set_name` was wrapping.

Nothing here prevents a caller assigning a number to `name`. If that matters, the answer is a
`@property` with a check inside it, from the **Properties** notebook, and it is written when the
requirement appears rather than in advance.


**3.** A method with no state is a function.


```python
def double(x):
    return x * 2


print([double(n) for n in [1, 2, 3]])
```

`MathHelpers` had nothing in `__init__`, so `self` carried nothing and every call had to construct
an object that held no data. A module of plain functions is the Python answer to a bag of helpers.


**4.** A fourth attribute, with a default.


```python
class Station:
    def __init__(self, name, readings, unit, calibrated=True):
        self.name = name
        self.readings = readings
        self.unit = unit
        self.calibrated = calibrated

    def average(self):
        return f"{mean(self.readings):.1f} {self.unit}"

    def report(self):
        line = f"{self.name}: {self.average()}"
        return line if self.calibrated else line + " (uncalibrated)"


print(Station("Tromso", [-4.1, -2.6, -3.8], "C").report())
print(Station("Bodo", [-2.6, -1.9], "C", calibrated=False).report())
```

The default matters. Existing code that creates a `Station` with three arguments keeps working,
which is the same reason a default argument is useful in a plain function.

Compare this with what the same change costs the function version: `report(name, readings, unit)`
becomes `report(name, readings, unit, calibrated)`, and every call in the program has to be found
and edited.


**5.** A function over a list of objects.


```python
def warmest(stations):
    return max(stations, key=lambda station: mean(station.readings)).name


group = [Station("Tromso", [-4.1, -2.6], "C"),
         Station("Galway", [11.5, 12.1], "C"),
         Station("Brno", [9.4, 10.1], "C")]

print(warmest(group))
```

Note that this is a plain function, not a method, and that is correct. It operates on a collection
of stations rather than on one station, so it does not belong to any single object.

A method would have to be called on one of them, `group[0].warmest(group)`, which reads as though
the first station has some special role. It does not.


**6.** Judgment, with reasons.


```python
# One chart, one list, then exit.
#   Function. There is one of it, it is used once, and nothing needs to be
#   remembered after it returns.

# Forty orders, each with an id, a customer, a list of items and a status.
#   Class. There are many of them, each carries several values that must stay
#   in step, and a caller could currently pair one order's items with another
#   order's status.

# Six unrelated string-cleaning helpers used across four files.
#   Function, in a module. The helpers share no data, so a class holding them
#   would never store anything and every call would construct an empty object.
print("nothing to run for this one")
```

The middle one has the signal that decides it: not the number of attributes, but that a caller can
currently pair the wrong values together. The other two have many values and many calls, and
neither has that risk.
