# Your First Class &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.


**1.** A class whose `__init__` stores three values.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages


for book in [Book("Dune", "Herbert", 412), Book("Emma", "Austen", 474)]:
    print(vars(book))
```

`vars` is the quickest check that `__init__` stored what you meant. Three parameters in, three keys
out.


**2.** A fourth parameter, with a default.


```python
class Book:
    def __init__(self, title, author, pages, read_so_far=0):
        self.title = title
        self.author = author
        self.pages = pages
        self.read_so_far = read_so_far


print(vars(Book("Dune", "Herbert", 412)))
print(vars(Book("Emma", "Austen", 474, 474)))
```

Both objects carry four attributes. The default decided what the first one's `read_so_far` would
be, not whether it had one.

`0` is safe as a default because it is a number, and numbers cannot be changed in place. A list or a
dictionary would need the treatment task 5 covers.


**3.** The assignments were going to locals.


```python
class Reading:
    def __init__(self, station, celsius):
        self.station = station
        self.celsius = celsius


reading = Reading("Tromso", -4.1)

print(vars(reading))
print(reading.station, reading.celsius)
```

The only change is `self.` on the left of each line. Without it the assignment sets a local variable
to its own value and the object stays empty, which is why the original raised `AttributeError` at
the first attribute read rather than when the object was created.


**4.** Two objects do not share.


```python
one = Book("Dune", "Herbert", 412)
two = Book("Emma", "Austen", 474)

one.read_so_far = 103

print("one:", one.read_so_far)
print("two:", two.read_so_far)
print("same object:", one is two)
```

Each call to `Book(...)` produced a separate object with its own attributes, so assigning through
one name cannot be seen through the other.

This is not in tension with the aliasing trap from **Why Classes**. There, one object had two names.
Here there are two objects.


**5.** The mutable default, and the version without it.


```python
class Tagged:
    def __init__(self, title, tags=[]):
        self.title = title
        self.tags = tags


first = Tagged("Dune")
second = Tagged("Emma")
first.tags.append("scifi")

print("first: ", first.tags)
print("second:", second.tags)
print("one list:", first.tags is second.tags)
```

```python
class TaggedFixed:
    def __init__(self, title, tags=None):
        self.title = title
        self.tags = [] if tags is None else tags


first = TaggedFixed("Dune")
second = TaggedFixed("Emma")
first.tags.append("scifi")

print("first: ", first.tags)
print("second:", second.tags)
print("one list:", first.tags is second.tags)
```

`first.tags is second.tags` is the line that shows what is happening. In the first version it is
`True`, so there is one list with two names on it. In the second it is `False`, because `[]` runs
once per call.

`None` is the right sentinel here rather than something like an empty string, because `None` is a
value nobody would pass deliberately, so `readings is None` cannot be confused with a caller
supplying an empty list on purpose.


**6.** A method built from the attributes.


```python
class Book:
    def __init__(self, title, author, pages, read_so_far=0):
        self.title = title
        self.author = author
        self.pages = pages
        self.read_so_far = read_so_far

    def summary(self):
        return f"{self.title} by {self.author}, {self.read_so_far} of {self.pages} pages"


print(Book("Dune", "Herbert", 412, 103).summary())
```

`summary` takes only `self` and reaches all four values through it. Nothing has to be passed in,
because the object already has everything the method needs, which is the point of putting the data
and the behavior together.
