# Methods &middot; Solutions


One way to do each task. Not the only way. If yours runs and does what was asked, yours is
right too.

The cell below rebuilds the class the notebook set up. Run it first.


```python
from statistics import mean


class Station:
    def __init__(self, name, readings, unit="C"):
        self.name = name
        self.readings = readings
        self.unit = unit

    def average(self):
        return round(mean(self.readings), 1)

    def coldest(self):
        return min(self.readings)

    def warmest(self):
        return max(self.readings)

    def add_reading(self, value):
        self.readings.append(value)

    def drop_below(self, limit):
        self.readings = [r for r in self.readings if r >= limit]

    def rounded(self):
        return Station(self.name, [round(r) for r in self.readings], self.unit)

    def report(self):
        return (f"{self.name}: mean {self.average()} {self.unit}, "
                f"coldest {self.coldest()} {self.unit}")


print("ready")
```

**1.** A method that returns the highest reading.


```python
north = Station("Tromso", [-4.1, -2.6, -3.8])

print(north.warmest())
```

`warmest` returns and changes nothing, so it can be called as often as you like. It is `coldest`
with `max` instead of `min`, which is a sign the two belong in the same class rather than being
written as separate functions taking a list.


**2.** A method that changes the object and returns `None`.


```python
north = Station("Tromso", [-4.1, -2.6, -3.8, -5.2])
print("before:  ", north.readings)

returned = north.drop_below(-4.0)

print("after:   ", north.readings)
print("returned:", returned)
```

`drop_below` rebinds `self.readings` to a new list rather than removing items from the existing one.
That still counts as changing the object, because `self.readings` is an attribute on the station and
the station now holds a different list.

Removing while looping over the same list is the version to avoid, since the loop and the removals
fight over the positions.

The method has no `return`, so it hands back `None`, which is the shape the notebook argued for.


**3.** One call, two spellings.


```python
north = Station("Tromso", [-4.1, -2.6, -3.8])

through_object = north.coldest()
through_class = Station.coldest(north)

print(through_object, through_class)
print("equal:", through_object == through_class)
```

`Station.coldest(north)` is what `north.coldest()` becomes. The first form is the one to write; the
second is worth being able to read, because it is what an error message is describing when it says
`self` is missing.


**4.** The bare method name is not defined.


```python
# The error is NameError: name 'average' is not defined. Methods are attributes
# of the class, not names in scope inside other methods, so report has to reach
# average through self.

class Fixed:
    def __init__(self, readings):
        self.readings = readings

    def average(self):
        return round(mean(self.readings), 1)

    def report(self):
        return f"mean {self.average()}"


print(Fixed([-4.1, -2.6]).report())
```

The only change is `self.` in front of `average()`. The class defines `average`, which is why the
error reads as though Python is wrong, and it is not: the class body is not a scope that methods
can see into.


**5.** A method that returns a new object.


```python
north = Station("Tromso", [-4.1, -2.6, -3.8])
whole = north.rounded()

print("north:", north.readings)
print("whole:", whole.readings)
print("same object:", whole is north)
print("same list:  ", whole.readings is north.readings)
```

Both lines of output matter. `whole is north` being `False` says two stations exist, and
`whole.readings is north.readings` being `False` says they do not share the list.

The second falls out of the list comprehension, which builds a new list. Had `rounded` been written
to pass `self.readings` through unchanged in some branch, that branch would need `list(...)` around
it, as `in_fahrenheit` does in the notebook.


**6.** Which kind is each method.


```python
# station.rename(new_name)
#   Changes the object, returns None. It is asking for one thing to be different
#   about this station, and there is no result a caller would want back.

# station.days_below(0)
#   Returns a value, changes nothing. It is a question about the readings, and
#   asking it twice should give the same answer both times.

# station.merge(other_station)
#   Either, but say which in the name. merge suggests this station absorbs the
#   other and returns None. If a third station should come back instead, call it
#   combined_with, which reads as producing something.
print("nothing to run for this one")
```

The third one is the honest answer to most naming questions: both designs are defensible, and the
cost of picking either is small next to the cost of a name that does not say which was picked.
