**1.** An alternative constructor for a CSV line.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    @classmethod
    def from_csv(cls, line):
        title, author, pages = line.split(",")
        return cls(title, author, int(pages))


for line in ["Dune,Herbert,412", "Emma,Austen,474"]:
    book = Book.from_csv(line)
    print(book.title, book.pages)
```

`title, author, pages = line.split(",")` unpacks exactly three fields, so a line with a fourth field
raises `ValueError` rather than quietly ignoring it. `int(pages)` is the conversion `__init__` should
never have to think about: it receives a number, not text.


**2.** A second constructor, for a JSON record.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    @classmethod
    def from_csv(cls, line):
        title, author, pages = line.split(",")
        return cls(title, author, int(pages))

    @classmethod
    def from_dict(cls, record):
        return cls(record["title"], record["author"], record["pages"])


record = json.loads('{"title": "Emma", "author": "Austen", "pages": 474}')
emma = Book.from_dict(record)

print(emma.title, emma.author, emma.pages)
```

JSON already gives `pages` as a number, so `from_dict` needs no conversion. Each constructor handles
exactly what its own format needs, and both end in the same `__init__`.


**3.** A counter on the class.


```python
class Book:
    count = 0

    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        Book.count += 1

    @classmethod
    def how_many(cls):
        return cls.count


for title in ["Dune", "Emma", "Ethan Frome"]:
    Book(title, "someone", 100)

print(Book.how_many())
```

`Book.count += 1` reads the class attribute, adds one, and assigns the result back to the class, so
every book increases the same number. `how_many` reads it through `cls`, and needs no book to be
called on.


**4.** A static method.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    @staticmethod
    def pages_for(words):
        return words // 300 + 1


dune = Book("Dune", "Herbert", 412)

print("on the class:", Book.pages_for(120_000))
print("on a book:   ", dune.pages_for(120_000))
```

`pages_for` uses nothing from any book, so it takes no `self`. It lives on `Book` because a page
estimate is about books, and that is where somebody would look for it.


**5.** The counter through `self` instead.


```python
class Book:
    count = 0

    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.count += 1


books = [Book(title, "someone", 100) for title in ["Dune", "Emma", "Ethan Frome"]]

print("Book.count:", Book.count)
print("each book: ", [book.count for book in books])

# self.count += 1 means self.count = self.count + 1. Reading self.count finds
# nothing on the new object, so it reads the class attribute, 0. Adding one
# gives 1, and assigning that through self creates a count attribute on the
# object. The class attribute is never changed, so Book.count stays 0 and
# every book holds its own count of 1.
```

This is the shadowing from the notebook, arriving through `+=`. The read goes to the class and the
write goes to the object, so the counter never counts.


**6.** A class method that builds many.


```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    @classmethod
    def from_csv(cls, line):
        title, author, pages = line.split(",")
        return cls(title, author, int(pages))

    @classmethod
    def load(cls, lines):
        return [cls.from_csv(line) for line in lines if line.strip()]


shelf = Book.load(["Dune,Herbert,412", "", "Emma,Austen,474"])

print([book.title for book in shelf])
```

`load` holds no parsing of its own. It decides which lines to use and hands each one to `from_csv`
through `cls`, so a change to the line format is still a change in one place.
